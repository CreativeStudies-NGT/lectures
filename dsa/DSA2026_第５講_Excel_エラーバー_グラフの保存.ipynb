# 最初に：プログラムは上から順に動く（NameError）

前回提出された notebook で、変数を作る前にグラフを描こうとしてエラーになっている例が何件かあった。下のセルを実行してみる。

````python
import matplotlib.pyplot as plt
plt.scatter(student_id, score)
student_id = [1, 2, 3, 4, 5]
score = [53, 88, 60, 63, 72]
plt.show()
````

- `NameError: name 'student_id' is not defined` は「`student_id` という名前は、まだない」という意味
- プログラムは**上の行から順に**実行される。2行目の `plt.scatter` の時点では、`student_id` も `score` もまだ作られていないので、エラーになる
- エラーが出た行で止まるので、その下の行（`student_id = …` など）は実行されない。もう一度実行しても、同じエラーになる
- 直し方：`student_id = …` と `score = …` の2行を、`plt.scatter` の行より**上に**移して、もう一度実行する

```python
import matplotlib.pyplot as plt
student_id = [1, 2, 3, 4, 5]
score = [53, 88, 60, 63, 72]
plt.scatter(student_id, score)
plt.show()
```

- **セルどうしも同じ**。notebook は、セルが並んでいる順ではなく、**実行した順**に動く。変数を作るセルを先に実行してから、グラフを描くセルを実行する
- 前回の notebook で作った変数は、新しい notebook には**引き継がれない**。新しい notebook では、データを作るセルから実行する
- セルの左の `[1]`・`[2]` などの数字が、**実行した順番**。エラーが出たら、変数を作るセルの数字が、グラフのセルの数字より小さいか（先に実行したか）を確かめる
- 順番がわからなくなったら、メニューの「Kernel」→「Restart Kernel and Run All Cells」で、上から順にすべて実行し直せる（作る前に使うセルがあると、そこで止まる）

# 前回の振り返り

- 今日はまず K でグラフを画像ファイルに保存し、保存された場所（いつもの場所）を確かめる。L-1 で、同じ場所に Excel のデータを置く。今日の notebook もいつもの場所に作る
- 前回は**平均** `np.mean(リスト)`・**標準偏差** `np.std(リスト)` を計算し、**補助線** `plt.axhline(y=値)`（横）・`plt.axvline(x=値)`（縦）を引いた
- **散布図** `plt.scatter(x, y)` も描いた。点の大きさは `s=`、色は `color=`（リストで書くと点ごとに変わる）
- 3つ目の値を、点の大きさ（バブルチャート：`s=`）で見せた。**色**で見せる方法（カラーマップ）は、今日 L-5 で扱う
- 重なる順番は `zorder=`（点に `zorder=3` を付けると、点が補助線の上になる）、透明度は `alpha=`、凡例は `label=` と `plt.legend()`、点の名前は `plt.text`

````python
import matplotlib.pyplot as plt
import japanize_matplotlib
import numpy as np

score = [72, 85, 64, 90, 58, 77, 81, 69]
number = [1, 2, 3, 4, 5, 6, 7, 8]

plt.figure(figsize=(4, 3))
plt.scatter(number, score, s=100, c='orange', edgecolor='k', zorder=3)
plt.axhline(y=np.mean(score), color='r', linestyle='--')
plt.ylim(0, 100)
plt.xlabel('出席番号', fontsize=12)
plt.ylabel('点数', fontsize=12)

# x軸目盛を明示的に指定
#plt.xticks([1, 2, 3, 4, 5, 6, 7, 8])  # plt.xticks(range(1,9))でもよい
plt.show()
````

- 今日は **pandas**（`import pandas as pd`）も使う。pandas は**表（Excel のようなデータ）を扱う道具箱**。これも使うセルの中で import する

## K-3 保存の形式と、保存した図を直す

ファイル名の最後（**拡張子**）で形式が決まる。

| 種類 | 形式（拡張子） | 向いている使い方 |
|---|---|---|
| 画像（ラスター） | `png`、`jpg`（`jpeg`）、`tif`（`tiff`）、`gif`、`webp`、`avif` | PowerPoint・Word・Web に貼る。**迷ったら png** |
| ベクター（拡大しても荒れない） | `svg`、`pdf`、`eps`、`ps` | 後から図を直す。ポスター・論文に使う |

- **ラスター**は写真と同じ「点（画素）の集まり」、**ベクター**は「ここからここへ線を引く」という情報。ベクターは拡大しても線がなめらか
- ベクターは **svg** を使う。eps は**透明が使えない**（`transparent=True` や `alpha=` が効かない）うえ、新しい PowerPoint には挿入できない

````python
import matplotlib.pyplot as plt
import japanize_matplotlib
score = [72, 85, 64, 90, 58, 77, 81, 69]
number = [1, 2, 3, 4, 5, 6, 7, 8]

plt.figure(figsize=(4, 3))
plt.scatter(number, score, s=100, c='orange', edgecolor='k')
plt.ylim(0, 100)
plt.xlabel('出席番号', fontsize=12)
plt.ylabel('点数', fontsize=12)

# x軸目盛を明示的に指定
#plt.xticks([1, 2, 3, 4, 5, 6, 7, 8])  # plt.xticks(range(1,9))でもよい

# svg形式で保存
plt.savefig('lec05_score.svg')
plt.show()
````

### svg を PowerPoint で直す（例：y 軸の名前を縦書きにする）

1. 上のセルで `lec05_score.svg` を保存する
1. PowerPoint で「挿入」→「画像」から svg を入れる
1. 図を右クリック →「**図形に変換**」
1. 元の y 軸の名前を消し、縦書きのテキストボックスで書き直す。線の色や太さも同じように直せる

- svg の文字は図形（線の集まり）になっているので、文字を打ち直すことはできない。消して書き直す

### Illustrator で直す（全学契約で学生も使える）

1. Illustrator で svg（または pdf）を「開く」
1. 全体がまとまっているので、「オブジェクト」→「グループ解除」、「オブジェクト」→「クリッピングマスク」→「解除」でばらす
1. 「ダイレクト選択ツール」で、線・点・文字を1つずつ選んで直せる
1. 元の y 軸の名前を消し、「文字（縦）ツール」で縦書きの名前を書く

- PowerPoint より細かく直せる（色をまとめて替える、点の形を変えるなど）。ポスターや論文の図に向く

### matplotlib だけで縦書きにする（紹介のみ）

- 1文字ずつ**改行**して、縦に並べる力技
- `rotation=0` で文字を横向きのまま並べ、`labelpad=` で軸から離す。`linespacing=1.5` で行の間を少し空けると読みやすい

````python
plt.figure(figsize=(8, 5))
plt.scatter(number, score, s=100)
plt.xlabel('出席番号')
plt.ylabel('点\n数', rotation=0, va='center', labelpad=15, linespacing=1.5)
plt.show()
````

- `'点\n数'` の `\n` が**改行**。縦書き用のかっこ `︵` `︶` を使うと、単位も縦に書ける：`'体\n重\n︵\nkg\n︶'`

### 制御文字（エスケープシーケンス）

`\` の後ろに文字を続けると、キーボードで直接打てない文字を表せる。

| 書き方 | 意味 | 例 |
|---|---|---|
| `\n` | 改行 | `print('1行目\n2行目')`、`plt.title('身長と体重\n（Aクラス）')` |
| `\t` | タブ（字下げ） | `print('名前\t点数')` |
| `\\` | `\` そのもの | `print('C:\\Users')` |
| `\'` | `'` そのもの（`'…'` の中で使う） | `print('It\'s')` |

- `\` の打ち方：Windows は `¥` キー（画面では `¥` に見えることがあるが同じ文字）、Mac は `option` + `¥`
- `print` で試してから、グラフのタイトルや軸の名前で使う（2行のタイトルなど）

````python
print('1行目\n2行目')
print('名前\t点数')
print('C:\\Users')
print('It\'s')
````

# M. エラーバー

**エラーバー**は、平均とばらつきを**点1つと縦の線1本**で表したもの。いくつかのグループ（クラス、教科など）の平均とばらつきを並べて比べるときに使う。

## M-0 最初の箱のデータを1点で代表させる

- 今日の最初の箱（準備のセル）の、8人の点数 `score` をもう一度描く。データはもう入っているので、描くコードだけでよい
- 前回（第４講 G-0）と同じく、平均の線に加えて、平均 ± 標準偏差の線（灰色の点線）も引く

````python
import numpy as np

plt.figure(figsize=(8, 5))
plt.scatter(number, score, s=100, c='orange', edgecolor='k', zorder=3)
plt.axhline(y=np.mean(score), color='r', linestyle='--')
plt.axhline(y=np.mean(score) + np.std(score), color='gray', linestyle=':')
plt.axhline(y=np.mean(score) - np.std(score), color='gray', linestyle=':')
plt.ylim(0, 100)
plt.xlabel('出席番号')
plt.ylabel('点数')
plt.show()
````

- このクラスの点数を、レポートやスライドで**1点で代表させたい**。いちばん素直なのは、平均の点1つ
- でも平均の点だけだと、8人がどれくらい散らばっていたか（**ばらつき**）が分からなくなる
- そこで、平均の点から上下に**標準偏差の長さの線**を付ける。これが**エラーバー**。上の図の線3本（平均と、平均 ± 標準偏差）を、点1つと縦の線1本にまとめたもの

````python
import numpy as np

plt.figure(figsize=(3, 5))
plt.errorbar(['クラス全体'], [np.mean(score)], yerr=[np.std(score)], fmt='o', capsize=8, markersize=10)
plt.xlim(-0.5, 0.5)
plt.ylim(0, 100)
plt.ylabel('点数')
plt.title('平均 ± 標準偏差')
plt.show()
````

- 縦軸の範囲（0〜100）を上の図とそろえてある。点が赤い線（平均）の高さ、縦の線の上端・下端が灰色の点線の高さになっている
- 書き方（`plt.errorbar`）は M-2 で説明する。今は実行して、形だけ見ればよい
- 1点にまとめると、いくつかのグループを**横に並べて比べられる**。次の M-1 では、2つのクラスを並べる

## M-1 平均だけでは分からないこと

- 下の2クラスは、**平均がほぼ同じ**（68点くらい）。でも、点数の散らばり方は同じとは限らない
- 散らばりの大きさを1つの数で表したものが**標準偏差**（第４講 G-3）

````python
import numpy as np

group1 = [66, 68, 70, 67, 69, 68, 71, 65, 70, 66]
group2 = [40, 95, 55, 85, 60, 90, 45, 80, 52, 78]
m = [np.mean(group1), np.mean(group2)]
s = [np.std(group1), np.std(group2)]
print('平均', m)
print('標準偏差', s)
````

- まず、一人ひとりの点数を散布図で描く。横の位置は、クラスの名前を人数分（10個）並べたリスト
- `['1組'] * 10` は「`'1組'` を10個並べたリスト」という意味（くわしくは第７講）

````python
plt.figure(figsize=(4, 4))
plt.scatter(['1組'] * 10, group1, alpha=0.5)
plt.scatter(['2組'] * 10, group2, alpha=0.5)
plt.xlim(-0.5, 1.5)
plt.ylim(0, 100)
plt.title('一人ひとりの点数')
plt.show()
````

- 同じデータを、平均 ± 標準偏差の**エラーバー**で描く（書き方は M-2）

````python
plt.figure(figsize=(4, 4))
plt.errorbar(['1組', '2組'], m, yerr=s, fmt='o', capsize=8, markersize=10)
plt.xlim(-0.5, 1.5)
plt.ylim(0, 100)
plt.title('平均 ± 標準偏差（エラーバー）')
plt.show()
````

- 2組は点数がばらばら → 標準偏差が大きい → エラーバーが**長い**
- 1組はみんな同じくらい → 標準偏差が小さい → エラーバーが**短い**
- 多くのデータでは、エラーバーの範囲（平均 ± 標準偏差）に**全体のおよそ3分の2**の人が入る
- 論文などでは、エラーバーが標準偏差なのか別のもの（標準誤差など）なのかを、必ず図の説明に書く

## M-2 エラーバーを描く

- 第４講 H-0 の2クラスの体重を使う。まず平均と標準偏差を求める

````python
import numpy as np

weight_a = [52, 58, 61, 68, 72]
weight_b = [50, 57, 60, 66, 70]

mean_a = np.mean(weight_a)
mean_b = np.mean(weight_b)
std_a = np.std(weight_a)
std_b = np.std(weight_b)
print('Aクラス 平均', mean_a, '標準偏差', std_a)
print('Bクラス 平均', mean_b, '標準偏差', std_b)
````

▶ **演習**：`errorbar_描画基礎001`

- `plt.errorbar(横の位置, 平均, yerr=エラーバーの長さ)` で描く。横の位置・平均・長さは、**同じ個数のリスト**にする
- `fmt='o'` で、点だけを描く（何も書かないと、点が線でつながる）

````python
plt.figure(figsize=(4, 5))
plt.errorbar(['Aクラス', 'Bクラス'], [mean_a, mean_b], yerr=[std_a, std_b], fmt='o', capsize=5)
plt.show()
````

▶ **演習**：`errorbar_描画基礎002`

### 見た目を整える

| オプション | 意味 |
|---|---|
| `yerr=` | エラーバーの長さ（上下それぞれ）。ふつうは標準偏差 |
| `fmt='o'` | 点の形（`'o'` 丸、`'s'` 四角、`'^'` 三角）。`'o-'` で線でつなぐ |
| `markersize=10` | 点の大きさ（散布図の `s=` とは違い、折れ線と同じ `markersize`） |
| `capsize=5` | エラーバーの端の横棒の長さ |
| `color='w'` | 点の色（`'w'` は白） |
| `markeredgecolor='k'` | 点の縁の色 |
| `ecolor='k'` | エラーバーの線の色 |

````python
plt.figure(figsize=(4, 5))
plt.errorbar(['Aクラス', 'Bクラス'], [mean_a, mean_b], yerr=[std_a, std_b], fmt='o', capsize=5,
             markersize=10, color='w', markeredgecolor='k', ecolor='k')
plt.xlim(-0.5, 1.5)
plt.ylim(40, 80)
plt.ylabel('体重 (kg)')
plt.show()
````

- 横軸が文字（`'Aクラス'` など）のとき、位置は左から `0, 1` になる。`plt.xlim(-0.5, 1.5)` で左右に余白を作っている
