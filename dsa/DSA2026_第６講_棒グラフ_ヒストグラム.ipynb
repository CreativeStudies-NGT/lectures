## N-1 棒グラフの基本

- `plt.plot(x, y)` を `plt.bar(x, y)` に変えるだけ。x が**横軸の項目**、y が**棒の高さ**
- 横軸の項目は、数でも**文字**（`'Thu'` など）でもよい

````python
x = ['Thu', 'Fri', 'Sat', 'Sun']
y = [170, 120, 250, 190]

plt.figure(figsize=(8, 5))
plt.bar(x, y)
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

### 折れ線グラフ・散布図との違い

| | 折れ線グラフ | 散布図 | 棒グラフ |
|---|---|---|---|
| 関数 | `plt.plot(x, y)` | `plt.scatter(x, y)` | `plt.bar(x, y)` |
| 向いているもの | 時間の流れ（順番に意味がある） | 2つの量の関係 | **項目ごとの量の比べ合い** |
| 縦軸 | 0 から始めなくてよい | 0 から始めなくてよい | **0 から始める**（棒の長さで比べるため） |

## N-2 棒の色

- 折れ線グラフ・散布図と同じく `color=`（短く `c=` は棒グラフでは使えないので `color=` と書く）
- 全部の棒が同じ色になる

````python
plt.figure(figsize=(8, 5))
plt.bar(x, y, color='orange')
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

- 前々回の散布図と同じく、`color=['r', 'g', 'b', 'gray']` のように**リスト**で書くと、棒ごとに色を変えられる

▶ **演習**：`bar_描画基礎001`

## N-3 棒の幅

- `width=` で棒の幅を変える。何も書かないと `0.8`。`1` にすると、すき間がなくなる

````python
plt.figure(figsize=(8, 5))
plt.bar(x, y, width=0.4)
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

▶ **演習**：`bar_描画基礎002`

## N-4 外枠線と網掛け

- `edgecolor=` で棒の**外枠線**の色、`linewidth=` でその太さ
- `hatch=` で**網掛け**（模様）。白黒で印刷しても区別できる

| 書き方 | 模様 | 書き方 | 模様 |
|---|---|---|---|
| `'/'` | 斜線 | `'x'` | バツ |
| `'\\'` | 逆向きの斜線 | `'.'` | 点 |
| `'-'` | 横線 | `'o'` | 丸 |
| `'+'` | 格子 | `'*'` | 星 |

- 同じ記号を重ねる（`'//'`、`'///'`）と、模様が細かくなる

````python
plt.figure(figsize=(8, 5))
plt.bar(x, y, color='w', edgecolor='k', linewidth=2, hatch='//')
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

▶ **演習**：`bar_描画基礎003`

## N-5 データを増やす

- 項目を増やすときは、x と y の**両方に**同じ数だけ足す（数が違うとエラー：`ValueError: shape mismatch`）
- 個数は `len(x)` と `len(y)` で確かめられる（第２講）

````python
x7 = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
y7 = [140, 110, 130, 170, 120, 250, 190]
print(len(x7), len(y7))

plt.figure(figsize=(8, 5))
plt.bar(x7, y7)
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

▶ **演習**：`bar_描画基礎004`（データは自分で決めてよい）

## N-6 平均の補助線

- 前回の復習。平均は `np.mean(リスト)`、横の補助線は `plt.axhline(y=値)`
- 平均の値を**自分で計算して書き込まない**。`np.mean(y7)` と書けば、データを変えても線が自動で動く

````python
plt.figure(figsize=(8, 5))
plt.bar(x7, y7)
plt.axhline(y=np.mean(y7), color='r', linestyle='--', label='平均')
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.legend()
plt.show()
````

▶ **演習**：`bar_描画基礎005`

## N-7 横棒グラフ

- `plt.bar` を `plt.barh` に変えると、棒が**横向き**になる（h は horizontal＝水平）
- 横向きになると、項目は**縦軸**、量は**横軸**に移る。だから `xlabel` と `ylabel` の中身を**入れ替える**
- 項目の名前が長いとき（都市名など）は、横棒グラフのほうが読みやすい

````python
plt.figure(figsize=(8, 5))
plt.barh(x7, y7)
plt.title('Bar Chart')
plt.xlabel('Total Bill')
plt.ylabel('Day')
plt.show()
````

▶ **演習**：`bar_描画基礎006`

- 横棒グラフでは、量が横軸にあるので、平均の補助線は**縦の線** `plt.axvline(x=値)` になる

````python
plt.figure(figsize=(8, 5))
plt.barh(x7, y7)
plt.axvline(x=np.mean(y7), color='r', linestyle='--', label='平均')
plt.title('Bar Chart')
plt.xlabel('Total Bill')
plt.ylabel('Day')
plt.legend()
plt.show()
````

▶ **演習**：`bar_描画基礎007`

# O. ヒストグラム

**ヒストグラム**は、データを**区間に分けて、それぞれの区間にいくつ入るかを数えた**グラフ。点数が**どのあたりに集まっているか**（分布）を見るときに使う。

| | 棒グラフ | ヒストグラム |
|---|---|---|
| 関数 | `plt.bar(x, y)` | `plt.hist(データ)` |
| 渡すもの | 項目（x）と高さ（y）の**2つ** | データ**1つだけ**（数えるのは matplotlib） |
| 横軸 | 項目（曜日、都市など） | **数値の区間**（0〜10点、10〜20点 …） |
| 縦軸 | 量 | **個数**（人数など） |
| 棒のすき間 | あける | **あけない**（区間がつながっているため） |

## O-1 ヒストグラムの基本

- `plt.hist(データ)` だけで描ける。区間の分け方と個数の数え方は、matplotlib が自動で決める

````python
score = [45, 52, 58, 61, 63, 65, 67, 68, 70, 71, 72, 73, 74, 75, 75,
         76, 77, 78, 79, 80, 81, 82, 84, 85, 86, 88, 90, 92, 95, 98]

plt.figure(figsize=(8, 5))
plt.hist(score)
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.show()
````

- 縦軸は**人数**。たとえば一番高い棒は「その区間の点数の人がいちばん多い」という意味
- これが演習「ヒストグラムの描画」のひな形

## O-2 区切りの数

- `bins=` で**区間の数**を決める（bin＝箱）。何も書かないと `10`
- 少なすぎると形が分からず、多すぎるとガタガタになる。データの数が30くらいなら 5〜10 が目安

````python
plt.figure(figsize=(8, 5))
plt.hist(score, bins=5)
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.show()
````

▶ **演習**：`hist_描画基礎001`

## O-3 色と枠線

- 棒グラフと同じく `color=`、`edgecolor=`。ヒストグラムは棒がくっついているので、**枠線を付けると区切りが見やすい**

````python
plt.figure(figsize=(8, 5))
plt.hist(score, bins=5, color='lightgreen', edgecolor='k')
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.show()
````

▶ **演習**：`hist_描画基礎002`

## O-4 区切りの範囲

- 何も書かないと、区間は**データの最小値から最大値まで**になる（上の図は 45〜98 点）。区切りが 55.6 点のような半端な値になる
- `range=(0, 100)` と書くと 0〜100 点を区切る。`bins=10` と合わせると、**10点刻み**（0〜10, 10〜20, …）になる

````python
plt.figure(figsize=(8, 5))
plt.hist(score, bins=10, range=(0, 100), edgecolor='k')
plt.xticks(range(0, 101, 10))
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.show()
````

- `plt.xticks(range(0, 101, 10))` は、横軸の目盛りを 0, 10, 20, …, 100 にする（なくてもよい）

▶ **演習**：`hist_描画基礎003`

## O-5 平均の補助線

- ヒストグラムでは点数が**横軸**にあるので、平均は**縦の線** `plt.axvline(x=np.mean(データ))`（横棒グラフのときと同じ）

````python
plt.figure(figsize=(8, 5))
plt.hist(score, bins=10, range=(0, 100), edgecolor='k')
plt.axvline(x=np.mean(score), color='r', linestyle='--', label='平均')
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.legend()
plt.show()
````

▶ **演習**：`hist_描画基礎004`

## O-6 2つのグループを重ねる

- 前々回の散布図と同じく、`plt.hist` を**2回**書くと重なる
- 後ろが隠れないように `alpha=0.5`（透明度）、どちらがどちらか分かるように `label=` と `plt.legend()`
- 2つの `bins=` と `range=` は**同じ値**にする（区間がずれると比べられない）

````python
score_b = [38, 47, 55, 58, 60, 62, 64, 66, 67, 69,
           70, 71, 72, 74, 76, 78, 80, 83, 87, 91]

plt.figure(figsize=(8, 5))
plt.hist(score, bins=10, range=(0, 100), alpha=0.5, label='A組')
plt.hist(score_b, bins=10, range=(0, 100), alpha=0.5, label='B組')
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.legend()
plt.show()
````

▶ **演習**：`hist_描画基礎005`

## P-1 ヒストグラムを並べる・横向きにする

- `plt.hist([データ1, データ2])` のように**リストにまとめて**渡すと、重ねずに**横に並べて**描く。`label=` もリストで渡す

````python
plt.figure(figsize=(8, 5))
plt.hist([score, score_b], bins=10, range=(0, 100), label=['A組', 'B組'])
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.legend()
plt.show()
````

▶ **演習**：`hist_描画基礎006`

- `orientation='horizontal'` で横向きになる。横棒グラフと同じく `xlabel` と `ylabel` を入れ替え、平均の線は**横の線** `plt.axhline`

````python
plt.figure(figsize=(8, 5))
plt.hist(score, bins=10, range=(0, 100), edgecolor='k', orientation='horizontal')
plt.axhline(y=np.mean(score), color='r', linestyle='--', label='平均')
plt.title('Histogram')
plt.xlabel('Number of students')
plt.ylabel('Score')
plt.legend()
plt.show()
````

▶ **演習**：`hist_描画基礎007`

## P-2 並べ替えて上位だけ描く

- `df.sort_values('列名', ascending=False)` で、その列の**大きい順**に並べ替える（`ascending=True` なら小さい順）
- `.head(10)` で上から10行だけ取り出す
- `barh` は**上から下へ**ではなく**下から上へ**並べるので、`plt.gca().invert_yaxis()` で上下を逆にすると1位が一番上になる

````python
top10 = df_ramen.sort_values('Y2023', ascending=False).head(10)

plt.figure(figsize=(8, 5))
plt.barh(top10['city'], top10['Y2023'], color='tan')
plt.gca().invert_yaxis()
plt.title('ラーメンの外食費 上位10市（2023年）')
plt.xlabel('1世帯あたりの支出 (円)')
plt.show()
````
