## 0-1 sympy でできること

電卓や numpy は、答えを**小数（近似値）**で出す。sympy は、**分数や $\sqrt{\ }$、文字を残したまま正確に**計算する。

````python
import sympy as sp

print(1 / 3)
print(8 ** 0.5)
````

````python
sp.Rational(1, 3) + sp.Rational(1, 6)
````

````python
sp.sqrt(8)
````

- `1 / 3` は 0.333… という小数になるが、`sp.Rational(1, 3)` は分数 $\frac{1}{3}$ のまま計算される
- `sp.sqrt(8)` は $2\sqrt{2}$ と、簡単にした形で出てくる
- セルの最後の行に式を書くと、結果が**数式の形**で表示される。`print()` で出すと、ふつうの文字（`2*sqrt(2)`）で表示される

## 0-3 共通の約束

**[1] `import sympy as sp`**
- numpy を `np`、matplotlib を `plt` と書いたのと同じく、sympy の命令は `sp.` を付けて書く（`sp.sqrt`、`sp.expand` など）

**[2] 記号（文字）は先に作る**
- Python では、`x` と書いても「x という文字」にはならない。`sp.symbols('x y')` で**記号を作ってから**使う
- 左側の変数名と、`' '` の中の文字の並びをそろえる

````python
import sympy as sp

x, y = sp.symbols('x y')
x**2 + 2*x*y + y**2
````

**[3] 式の書き方**

| 数学 | sympy | 注意 |
|---|---|---|
| $2x$ | `2*x` | `*` は省略できない |
| $x^2$ | `x**2` | `^` ではなく `**` |
| $\frac{1}{2}$ | `sp.Rational(1, 2)` | `1/2` と書くと 0.5 という小数になる |
| $\sqrt{x}$、$\sin x$、$e^x$、$\log x$ | `sp.sqrt(x)`、`sp.sin(x)`、`sp.exp(x)`、`sp.log(x)` | `np.sin` ではなく `sp.sin` |
| $\pi$、$e$、$i$（虚数単位）、$\infty$ | `sp.pi`、`sp.E`、`sp.I`、`sp.oo` | `oo` は小文字の o を2つ |

**[4] グラフの行の最後の `;`**
- `sp.plot(…)` の行の最後に `;` を付けると、グラフの下に出る余分な文字（`<sympy.plotting…>`）が出なくなる

# 1. 代数（式の計算）

展開・因数分解・約分・簡単化と、値の代入。

**準備**：最初にこのセルを実行する。

````python
import sympy as sp

x, y, a, b, c = sp.symbols('x y a b c')
````

## 1-1 式を書く・値を代入する

- 式に名前を付けておくと、あとで何度も使える
- `式.subs(x, 値)` で、x に値を**代入**する。値の代わりに式を入れることもできる

````python
expr = x**2 + 2*x + 1
expr
````

````python
expr.subs(x, 3)
````

````python
expr.subs(x, a + 1)
````

- 2つ以上の文字に代入するときは `{文字: 値, 文字: 値}` の形で渡す

````python
expr2 = x**2 + x*y + y**2
expr2.subs({x: 1, y: 2})
````

▶ **演習**：`sympy_代数1`

## 1-2 展開する

- `sp.expand(式)` で、かっこを外して展開する

````python
sp.expand((x + 1)**3)
````

````python
sp.expand((a - b)*(b - c)*(c - a))
````

## 1-3 因数分解する

- `sp.factor(式)` で因数分解する。展開の逆

````python
sp.factor(x**2 - 5*x + 6)
````

````python
sp.factor(x**4 - 7*x**2*y**2 + y**4)
````

▶ **演習**：`sympy_代数2`

## 1-4 約分・簡単化・部分分数分解

- `sp.cancel(式)`：分数の式を約分する
- `sp.simplify(式)`：いちばん簡単な形を探してくれる（何をしてよいか迷ったらこれ）
- `sp.apart(式)`：部分分数に分ける（積分などで使う）

````python
sp.cancel((x**2 - y**2) / (x - y))
````

````python
sp.simplify(sp.sin(x)**2 + sp.cos(x)**2)
````

````python
sp.apart((x - 1) / (x*(x**2 + 1)**2))
````

## 1-5 三角関数の加法定理

- `sp.expand(式, trig=True)` で、三角関数も展開する

````python
sp.expand(sp.sin(x + y), trig=True)
````

````python
sp.expand(sp.cos(2*x), trig=True)
````

## 1-6 小数の値にする

- `sp.N(式)` で小数の値（近似値）にする。`sp.N(式, 桁数)` で桁数も指定できる

````python
sp.N(sp.pi, 30)
````

````python
sp.N(sp.sqrt(2) + sp.Rational(1, 3))
````

▶ **演習**：`sympy_代数3`

## 1-7 整数の素因数分解

- `sp.factorint(整数)` で、整数を素因数分解する。答えは `{素数: 何乗か}` の形で出てくる
- `sp.isprime(整数)` で、素数かどうかを調べる（True / False）
- 1-3 の `sp.factor` は**式**の因数分解、`sp.factorint` は**整数**の素因数分解

````python
sp.factorint(360)
````

````python
sp.isprime(97)
````

**例：1 だけが並ぶ数**

- 1111111（1 が7個）は、小さい素数では割り切れないので、手計算では難しい
- `int('1' * 7)` と書くと、文字 1 を7個並べて整数にした 1111111 になる

````python
sp.factorint(int('1' * 7))
````

- 1 が2個から12個までの数を、まとめて素因数分解してみる（`for` 文を使う）
- 1 が素数個並んでいないと、素数にならない（たとえば 1 が6個なら、11 や 111 で割り切れる）。ただし、1 が11個のように素数個でも、素数とは限らない

````python
for n in range(2, 13):
    print(n, sp.factorint(int('1' * n)))
````

# 2. グラフ描画

sympy の式を、そのままグラフにする。matplotlib のようにデータの点を用意しなくてよい。

**準備**：最初にこのセルを実行する。`japanize_matplotlib` は、タイトルなどに日本語を使うため。

````python
import sympy as sp
import japanize_matplotlib

x, y, t = sp.symbols('x y t')
````

## 2-1 関数のグラフ

- `sp.plot(式, (x, 最小, 最大))` で、x の範囲を指定して描く
- 範囲を省くと、x は -10 から 10 になる

````python
sp.plot(x**2 - 2*x - 3, (x, -3, 5));
````

````python
sp.plot(sp.sin(x) / x, (x, -12, 12));
````

## 2-2 見た目を整える

| 指定 | 意味 |
|---|---|
| `title='…'` | タイトル |
| `xlabel='…'`、`ylabel='…'` | 軸の名前 |
| `line_color='…'` | 線の色（`'tomato'` など、matplotlib と同じ色の名前） |
| `size=(横, 縦)` | グラフの大きさ（matplotlib の `figsize` にあたる） |
| `ylim=(最小, 最大)` | 縦軸の範囲 |
| `aspect_ratio=(1, 1)` | 縦と横の目盛りの幅をそろえる（円を円に見せる） |

````python
sp.plot(x**2 - 2*x - 3, (x, -3, 5), title='放物線', xlabel='x', ylabel='y',
        line_color='tomato', size=(6, 4), ylim=(-5, 10));
````

## 2-3 グラフを重ねる

- 式を `,` で並べると、同じグラフに重ねて描く。`legend=True` で凡例を出す

````python
sp.plot(x**2, 2*x + 3, (x, -3, 5), legend=True);
````

- 線ごとに色や凡例の文字を変えるときは、`show=False` でいったん描かずに受け取り、`p[0]`（1本目）、`p[1]`（2本目）…を設定してから `p.show()` で表示する

````python
p = sp.plot(x**2, 2*x + 3, (x, -3, 5), legend=True, show=False)
p[0].line_color = 'tomato'
p[0].label = '放物線'
p[1].line_color = 'steelblue'
p[1].label = '直線'
p.show()
````

## 2-4 円などの図形（陰関数）

- $x^2 + y^2 = 1$ のように、$y = \cdots$ の形にしにくい図形は `sp.plot_implicit` で描く
- 等式は `sp.Eq(左辺, 右辺)` で書く。x と y の両方の範囲を指定する

````python
sp.plot_implicit(sp.Eq(x**2 + y**2, 1), (x, -1.5, 1.5), (y, -1.5, 1.5), aspect_ratio=(1, 1));
````

- `aspect_ratio=(1, 1)` で縦と横の目盛りの幅をそろえる。付けないと、円が横長の楕円に見える

- 不等式を書くと、その**領域**を塗りつぶす

````python
sp.plot_implicit(y > x**2, (x, -2, 2), (y, -1, 4));
````

## 2-5 媒介変数表示のグラフ

- $x$ と $y$ を、別の文字 $t$ の式で表したグラフ。`sp.plot_parametric((xの式, yの式), (t, 最小, 最大))`

````python
sp.plot_parametric((sp.cos(t), sp.sin(t)), (t, 0, 2*sp.pi), size=(4, 4));
````

````python
sp.plot_parametric((sp.cos(t), sp.sin(2*t)), (t, 0, 2*sp.pi), size=(4, 4));
````

# 5. 固有値・対角化

固有値・固有ベクトル、特性多項式、対角化と、その使い道。

**準備**：最初にこのセルを実行する。

````python
import sympy as sp

lam, n = sp.symbols('lambda n')
A = sp.Matrix([[2, 1], [1, 2]])
A
````

## 5-1 固有値

- `A.eigenvals()` で**固有値**を求める
- 答えは `{固有値: 重複度, …}` の形（重複度は、その固有値が何回重なっているか）

````python
A.eigenvals()
````

- 固有値は 1 と 3（どちらも重複度 1）

## 5-2 固有ベクトル

- `A.eigenvects()` で、固有値と**固有ベクトル**をまとめて求める
- 答えは `(固有値, 重複度, [固有ベクトル])` の組が、固有値の数だけ並んだリスト

````python
lam1 = vects[0][0]
v1 = vects[0][2][0]
v1
````

- 固有ベクトル $\boldsymbol{v}$ は、$A\boldsymbol{v} = \lambda \boldsymbol{v}$（A をかけても、向きが変わらず $\lambda$ 倍になるだけ）を満たす。左辺と右辺を計算して確かめる

````python
A * v1
````

````python
lam1 * v1
````

▶ **演習**：`sympy_固有値1`

## 5-3 特性多項式

- 固有値は、**特性多項式** $\det(\lambda E - A)$ を 0 にする $\lambda$。`A.charpoly(lam)` で求め、`.as_expr()` でふつうの式にする

````python
p = A.charpoly(lam).as_expr()
p
````

````python
sp.factor(p)
````

````python
sp.solve(p, lam)
````

# 6. 微分・積分

極限・微分・偏微分・積分と、微分を使って極値を求める方法。

**準備**：最初にこのセルを実行する。

````python
import sympy as sp
import japanize_matplotlib

x, y, h = sp.symbols('x y h')
````

## 6-1 極限

- `sp.limit(式, x, 近づける値)` で極限を求める。無限大は `sp.oo`

````python
sp.limit(sp.sin(x) / x, x, 0)
````

````python
sp.limit((1 + 1/x)**x, x, sp.oo)
````

- 2つ目の答え $e$ は、自然対数の底（ネイピア数）

## 6-2 微分の定義から確かめる

- 微分の定義 $\displaystyle f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$ を、そのまま計算する

````python
sp.limit((sp.sin(x + h) - sp.sin(x)) / h, h, 0)
````

- $\sin x$ を微分すると $\cos x$ になる
- $\tan x$ でも確かめてみる

````python
sp.limit((sp.tan(x + h) - sp.tan(x)) / h, h, 0)
````

- 答えは $\frac{1}{\cos^2 x}$。6-3 の `sp.diff(sp.tan(x), x)` では $\tan^2 x + 1$ と表示されるが、同じもの（$\sin^2 x + \cos^2 x = 1$ の両辺を $\cos^2 x$ で割ると確かめられる）

## 6-3 微分

- `sp.diff(式, x)` で x について微分する。`sp.diff(式, x, 2)` で2回微分する

````python
sp.diff(x**3 - 3*x, x)
````

````python
sp.diff(sp.sin(2*x), x)
````

````python
sp.diff(sp.tan(x), x)
````

````python
sp.diff(sp.exp(x) * sp.sin(x), x)
````

````python
sp.diff(x**3 - 3*x, x, 2)
````

- `sp.Derivative(式, x)` は、微分の式を作るだけで計算しない。`.doit()` で計算する。`sp.Eq` と組み合わせると「微分の式 = 答え」の形で表示できる（6-6 の `sp.Integral` と同じ使い方）

````python
expr = sp.Derivative(sp.sin(x)**2, x)
sp.Eq(expr, expr.doit())
````

## 6-5 微分を使って極値を求める

| 不定積分 | `sp.integrate(式, x)`（積分定数は出ない） |
| 定積分 | `sp.integrate(式, (x, 下端, 上端))` |
| 微分の式だけ作る | `sp.Derivative(式, x)` → `.doit()` で計算 |
| 積分の式だけ作る | `sp.Integral(式, x)` → `.doit()` で計算 |

# 7. 微分方程式

**微分方程式**は、関数 $f(x)$ と、その微分 $f'(x)$、$f''(x)$ の関係を表した式。解は数ではなく**関数**になる。6章の微分を先に読んでおくとよい。

**準備**：最初にこのセルを実行する。`sp.Function('f')` で、f を**関数の記号**として作る。

````python
import sympy as sp
import japanize_matplotlib

x = sp.symbols('x')
f = sp.Function('f')
````

## 7-1 微分方程式を書く

- $f(x)$ は `f(x)`、$f'(x)$ は `f(x).diff(x)`、$f''(x)$ は `f(x).diff(x, 2)` と書く
- 方程式は `sp.Eq(左辺, 右辺)`。下は $f'(x) = -f(x)$

````python
eq = sp.Eq(f(x).diff(x), -f(x))
eq
````

## 7-2 一般解を求める

- `sp.dsolve(方程式, f(x))` で解く。答えに出てくる $C_1$ は、どんな値でもよい**定数**（一般解）

````python
sp.dsolve(eq, f(x))
````

- $f(x)$ を含まない項（ここでは $x^2$）がある方程式も、同じように解ける。$f'(x) + f(x) + x^2 = 0$ の場合

````python
sp.dsolve(sp.Eq(f(x).diff(x) + f(x) + x**2, 0), f(x))
````

- 答えは「$x^2$ の項がないときの解（$C_1 e^{-x}$）」と「$x$ の2次式」の和になっている

## 7-3 初期条件を付けて解く

- `ics={f(0): 値}` で**初期条件**（x = 0 のときの値）を与えると、定数が決まる

````python
sol = sp.dsolve(eq, f(x), ics={f(0): 1})
sol
````

- 解が方程式を満たしているかは、`sp.checkodesol(方程式, 解)` で確かめられる。`(True, 0)` なら正しい

````python
sp.checkodesol(eq, sol)
````

▶ **演習**：`sympy_微分方程式1`・`sympy_微分方程式2`

## 7-4 2階の微分方程式

- $f''(x) + f(x) = 0$ は、ばねの振動（単振動）を表す。定数が2つ（$C_1$、$C_2$）出てくる

````python
eq2 = sp.Eq(f(x).diff(x, 2) + f(x), 0)
sp.dsolve(eq2, f(x))
````

- 2階のときは、初期条件も2つ必要。$f'(0)$ は `f(x).diff(x).subs(x, 0)` と書く

````python
sp.dsolve(eq2, f(x), ics={f(0): 0, f(x).diff(x).subs(x, 0): 1})
````

- 2階の微分方程式でも、振動しない解になるものがある。$f''(x) + f'(x) - 2f(x) = 0$ の場合

````python
sp.dsolve(sp.Eq(f(x).diff(x, 2) + f(x).diff(x) - 2*f(x), 0), f(x))
````

- 指数に出てくる $-2$ と $1$ は、2次方程式 $\lambda^2 + \lambda - 2 = 0$ の解。単振動 $f'' + f = 0$ では、この2次方程式 $\lambda^2 + 1 = 0$ の解が虚数 $\pm i$ になるので、$\sin$・$\cos$ が出てくる（8-2 のオイラーの式）

## 7-5 文字を含む微分方程式

- 放射性物質の量や、冷めていくお茶の温度差は $f'(x) = -kf(x)$（k は正の定数）で表される
- `sp.symbols('k', positive=True)` で、k を**正の数**として作る

````python
k = sp.symbols('k', positive=True)
sp.dsolve(sp.Eq(f(x).diff(x), -k*f(x)), f(x))
````

- 人口や感染者数の増え方のモデル（ロジスティック方程式）$f'(x) = f(x)\,(1 - f(x))$ も解ける

````python
eq3 = sp.Eq(f(x).diff(x), f(x) * (1 - f(x)))
sol3 = sp.dsolve(eq3, f(x), ics={f(0): sp.Rational(1, 10)})
sol3
````

▶ **演習**：`sympy_微分方程式3`・`sympy_微分方程式4`

## 7-6 解をグラフにする

- 解は `f(x) = …` という等式。右辺だけを `.rhs` で取り出し、`sp.plot` に渡す

````python
sp.plot(sol3.rhs, (x, 0, 10), title='ロジスティック曲線', ylim=(0, 1.1), axis_center=(0, 0));
````

- `axis_center=(0, 0)` で、横軸を y = 0 の位置に描く。付けないと、横軸がグラフの途中（y = 0.55 あたり）に描かれる

- 摩擦のある振動（減衰振動）$f''(x) + \frac{1}{5}f'(x) + f(x) = 0$ を、$f(0) = 1$、$f'(0) = 0$ で解いて描く

````python
eq4 = sp.Eq(f(x).diff(x, 2) + sp.Rational(1, 5)*f(x).diff(x) + f(x), 0)
sol4 = sp.dsolve(eq4, f(x), ics={f(0): 1, f(x).diff(x).subs(x, 0): 0})
sol4
````

````python
sp.plot(sol4.rhs, (x, 0, 30), title='減衰振動');
````

- `sp.Rational(1, 5)` を大きくすると（例：`sp.Rational(1, 2)`）、早く揺れがおさまる

# 9. 固有値の応用：主成分分析と2状態の量子力学

## 9-1 主成分分析とは

- たくさんの変数を、情報をなるべく失わずに**少ない数の新しい変数（主成分）にまとめる**方法。アンケートの多くの質問を2〜3個の軸にまとめる、などに使う
- ここでは5人の数学と理科の点数（2つの変数）を、1つの「総合力」にまとめられるかを調べる

| 番号 | 手順 | 節 |
|---|---|---|
| [1] | データから平均を引く | 9-2 |

## 9-2 データと平均

- 1行が1人、1列目が数学、2列目が理科の行列にする

````python
X = sp.Matrix([[50, 50], [60, 80], [80, 60], [80, 80], [80, 80]])
n = X.shape[0]
X
````

- 全部 1 の行ベクトル `sp.ones(1, n)` を左からかけると、列ごとの**合計**になる。n で割ると平均

````python
mean = sp.ones(1, n) * X / n
mean
````

- 各人の点数から平均を引く（**中心化**）。平均の行を n 人分に縦に並べて `sp.ones(n, 1) * mean` を引く

````python
Xc = X - sp.ones(n, 1) * mean
Xc
````

## 9-3 分散・共分散と共分散行列

主成分分析の材料になる**共分散行列**を、統計の基本（偏差・分散・共分散）から順に組み立てる。

### [1] 偏差：平均からのずれ

- 平均が同じでも、点数が平均の近くに集まっているクラスと、0 点から 100 点まで散らばっているクラスがある。**ばらつき**は平均だけではわからない
- 各人の値から平均を引いたもの（9-2 の `Xc`）を**偏差**という。平均より上ならプラス、下ならマイナス
- 偏差を全員分足すと、プラスとマイナスが打ち消し合って**いつも 0** になる。そのままでは、ばらつきの大きさを表せない

````python
sum(Xc.col(0))
````

### [2] 分散と標準偏差：1つの変数のばらつき

- 偏差を**2乗**すると、マイナスもプラスになる。偏差の2乗を全員分足して $n - 1$ で割ったものを**分散**という

$$\text{分散} = \frac{(\text{偏差}_1)^2 + (\text{偏差}_2)^2 + \cdots + (\text{偏差}_n)^2}{n - 1}$$

- 偏差の2乗の合計は、偏差の列ベクトルどうしの**内積** `.dot()`（同じ番号の成分どうしをかけて足したもの）で計算できる

````python
Xc.col(0).T
````

````python
var_math = Xc.col(0).dot(Xc.col(0)) / (n - 1)
var_math
````

- 数学の分散は 200。$(-20)^2 + (-10)^2 + 10^2 + 10^2 + 10^2 = 800$ を $5 - 1 = 4$ で割った値
- 分散は点数を2乗しているので、単位は「点²」になる。**平方根をとって元の単位（点）に戻した**ものを**標準偏差**という

````python
sp.sqrt(var_math), sp.N(sp.sqrt(var_math), 4)
````

- 数学の標準偏差は $10\sqrt{2} \approx 14.1$ 点。「平均 70 点から、だいたい ±14 点くらいに散らばっている」と読む
- **n で割るか、n − 1 で割るか**：手元のデータそのもののばらつきなら n で割る。データを「もっと大きな集団から取り出した一部」と考えて、元の集団のばらつきを推し量るなら n − 1 で割る（**不偏分散**）。ここでは n − 1 を使う

| 道具 | n で割る | n − 1 で割る |
|---|---|---|
| Excel | `VAR.P` | `VAR.S` |
| pandas | `df['列'].var(ddof=0)` | `df['列'].var()` |
| numpy | `np.var(値)` | `np.var(値, ddof=1)` |

### [3] 共分散：2つの変数の関係

- 数学の偏差と理科の偏差を、**1人ずつかけ算**する

````python
Xc.col(0).multiply_elementwise(Xc.col(1)).T
````

- 2科目とも平均より上（または2科目とも下）の人はプラス、片方だけ上の人はマイナスになる
- 散布図の軸を平均 (70, 70) の位置で交わらせると、4つの区画に分かれる。右上と左下の点は積がプラス、左上と右下の点はマイナス

````python
sp.plot(markers=[{'args': [list(X.col(0)), list(X.col(1)), 'o'], 'color': 'k'}],
        xlim=(40, 100), ylim=(40, 100), axis_center=(70, 70), aspect_ratio=(1, 1),
        title='横：数学、縦：理科（軸は平均 70 点の位置）', xlabel='', ylabel='');
````

- (80, 80) の2人は点が重なるので、点は4つに見える
- この積を全員分足して $n - 1$ で割ったものを**共分散**という。分散の「2乗」を「2つの変数の偏差の積」に変えたもの

````python
cov = Xc.col(0).dot(Xc.col(1)) / (n - 1)
cov
````

| 共分散 | 意味 | 散布図 |
|---|---|---|
| プラス | 一方が大きいと、もう一方も大きい | 右上がり |
| マイナス | 一方が大きいと、もう一方は小さい | 右下がり |
| 0 に近い | はっきりした関係がない | ばらばら |

- 共分散は 100 でプラス。数学ができる人は理科もできる傾向がある

### [4] 相関係数：単位に左右されない関係の強さ

- 共分散は単位（点×点）に左右される。点数を 10 倍すると共分散は 100 倍になる。そこで、2つの標準偏差で割って **-1 から 1 の数にそろえた**ものを**相関係数**という

````python
var_sci = Xc.col(1).dot(Xc.col(1)) / (n - 1)
cov / (sp.sqrt(var_math) * sp.sqrt(var_sci))
````

- 相関係数は $\frac{1}{2}$。1 に近いほど強い右上がり、-1 に近いほど強い右下がり、0 なら関係がない

### [5] 共分散行列：分散と共分散を1つの表にまとめる

- 分散と共分散を、行列の形に並べたものを**共分散行列**という

$$S = \begin{pmatrix} \text{数学の分散} & \text{数学と理科の共分散} \\ \text{理科と数学の共分散} & \text{理科の分散} \end{pmatrix}$$

- 4章の行列の積を使うと、$S = \frac{1}{n-1} X_c^{\mathsf{T}} X_c$ の1行で、全部の内積をまとめて計算できる。$X_c^{\mathsf{T}} X_c$ の (i, j) 成分は、「i 列目の偏差と j 列目の偏差の内積」になっている

````python
S = Xc.T * Xc / (n - 1)
S
````

- 対角が分散（数学 200、理科 200）、対角でないところが共分散（100）。[2]・[3] で1つずつ計算した値と一致する
- 「数学と理科の共分散」と「理科と数学の共分散」は同じなので、共分散行列はいつも**対称な行列**（`S.T` が `S` と同じ）になる
- 科目が3つなら 3行3列、10個なら 10行10列になる。それでも計算は `Xc.T * Xc / (n - 1)` の1行のまま
- $S = 100\begin{pmatrix} 2 & 1 \\ 1 & 2 \end{pmatrix}$ で、**5章の行列 A のちょうど 100 倍**になっている

▶ **演習**：`sympy_分散1`

## 9-4 固有値・固有ベクトル（主成分の向き）

- 共分散行列の固有ベクトルが**主成分の向き**、固有値がその向きの**分散（ばらつきの大きさ）**になる
- `S.diagonalize(normalize=True)` で、長さを 1 にそろえた固有ベクトルを並べた P を求める

````python
P, D = S.diagonalize(normalize=True)
P
````

````python
D
````

- D の対角は小さい順に 100, 300。P の列は D と同じ順に並んでいる
- 固有値が大きい 300 の固有ベクトル $\frac{1}{\sqrt{2}}(1, 1)$ が**第1主成分**：数学と理科が同じ向きに増える＝**総合力**
- 固有値が 100 の固有ベクトル $\frac{1}{\sqrt{2}}(-1, 1)$ が**第2主成分**：理科が上がると数学が下がる向き＝**理科と数学のどちらが得意か**
- 5章の A と同じ形の行列なので、固有ベクトルも 5章と同じになる

## 9-5 寄与率

- 固有値の和は、分散の合計（トレース）と等しい。ある主成分の固有値が全体に占める割合を**寄与率**という

````python
S.trace()
````

````python
D[1, 1] / D.trace()
````

- 第1主成分の寄与率は $\frac{3}{4}$（75%）。2科目の点数のばらつきの 75% を、「総合力」1つで説明できる

## 9-6 主成分得点

- 中心化したデータに P をかけると、新しい軸（主成分）で見た各人の値（**主成分得点**）になる。P の並びと同じく、1列目が第2主成分、2列目が第1主成分

````python
Z = Xc * P
Z
````

- 主成分得点の共分散行列を計算すると D になる。対角でないところが 0 なので、**主成分どうしは無相関**

````python
Z.T * Z / (n - 1)
````

from sklearn.decomposition import PCA

print(pca.components_)  # 主成分の向き（固有ベクトル）

## 9-9 2状態の量子力学とは

- 量子力学では、**2つの状態のどちらか**をとる系を考えることが多い。たとえば、アンモニア分子の窒素原子が「上」か「下」か、電子のスピンが「上向き」か「下向き」か、量子コンピュータの量子ビットが「0」か「1」か
- 系の状態は、**複素数を成分とする長さ 1 のベクトル** $\psi = (c_1, c_2)$ で表す。$|c_1|^2$ が状態1、$|c_2|^2$ が状態2で見つかる**確率**
- エネルギーは**ハミルトニアン**という行列 H で表す。H は**エルミート行列**（複素数版の対称行列）
- 測って得られるエネルギーは **H の固有値**、そのエネルギーをもつ状態は **H の固有ベクトル**
- ここでは単位を簡単にするため、プランク定数 $\hbar$ を 1 とする

## 9-10 状態ベクトルと確率

- 成分に虚数単位 `sp.I` を使ってよい。`psi.norm()` で長さ、`sp.Abs(成分)**2` で確率を求める

````python
psi = sp.Matrix([1, sp.I]) / sp.sqrt(2)
psi
````

````python
psi.norm()
````

````python
sp.Abs(psi[0])**2, sp.Abs(psi[1])**2
````

- 状態1・状態2で見つかる確率は、どちらも $\frac{1}{2}$
- `.H` は**共役転置**（転置して、虚数単位の符号を変える）。`psi.H * psi` は長さの2乗で、1 になる

````python
psi.H * psi
````

## 9-11 ハミルトニアンの固有値＝エネルギー準位

- 2つの状態のエネルギーがどちらも $E_0$ で、状態のあいだを $\Delta$ の強さで行き来できる系（アンモニア分子など）のハミルトニアン

````python
E0, Dl = sp.symbols('E_0 Delta', positive=True)
H = sp.Matrix([[E0, -Dl], [-Dl, E0]])
H
````

````python
H.H == H
````

- `H.H == H` が `True` なので、H はエルミート行列

````python
H.eigenvects()
````

- エネルギーは $E_0 - \Delta$ と $E_0 + \Delta$ の2つに**分かれる**。行き来がなければ（$\Delta = 0$）どちらも $E_0$ だった
- エネルギーの低い状態は $\frac{1}{\sqrt{2}}(1, 1)$：状態1と状態2を**半分ずつ重ね合わせた**状態
- 9-4 の主成分分析と**同じ形の行列**（対角が等しく、対角でないところも等しい対称な行列）なので、固有ベクトルも同じ $(1, 1)$ と $(-1, 1)$ になる
  - 主成分分析：数学と理科の分散が等しい → 第1主成分は「両方を半分ずつ」の向き
  - 量子力学：2つの状態のエネルギーが等しい → エネルギーの低い状態は「両方を半分ずつ」重ね合わせた状態

## 9-12 時間発展（2つの状態を行き来する）

- 状態の時間変化は、シュレーディンガー方程式 $i\dfrac{d\psi}{dt} = H\psi$（7章と同じ微分方程式）で決まる。解は $\psi(t) = e^{-iHt}\psi(0)$
- 行列の指数関数 $e^{-iHt}$ は `(-sp.I * H * t).exp()` で求める。sympy は中で**対角化**を使って計算している（5-6 で $A^n$ を対角化で計算したのと同じ考え）
- 最初（t = 0）に状態1にいたとする

````python
U = (-sp.I * H * t).exp()
psi0 = sp.Matrix([1, 0])
psit = U * psi0
P1 = sp.simplify(sp.Abs(psit[0])**2)
P2 = sp.simplify(sp.Abs(psit[1])**2)
P1, P2
````

- 状態1にいる確率は $\frac{1 + \cos 2\Delta t}{2} = \cos^2 \Delta t$、状態2にいる確率は $\sin^2 \Delta t$。足すといつも 1
- $\Delta = 1$ としてグラフにすると、2つの状態を**周期的に行き来**していることがわかる。$E_0$ は確率に関係しない

````python
p = sp.plot(P1.subs(Dl, 1), P2.subs(Dl, 1), (t, 0, 2*sp.pi), title='状態を見つける確率',
            xlabel='時間 t', ylabel='確率', legend=True, show=False)
p[0].label = '状態1'
p[1].label = '状態2'
p.show()
````

- アンモニア分子では、この行き来の速さ（$\Delta$ に比例）がマイクロ波の周波数になっていて、メーザー（レーザーのマイクロ波版）に使われた

## 9-13 エネルギーに差がある場合（準位の反発）

- 2つの状態のエネルギーが $\varepsilon$ と $-\varepsilon$ で、行き来の強さが V のとき

````python
eps, V = sp.symbols('epsilon V', real=True)
H2 = sp.Matrix([[eps, V], [V, -eps]])
H2.eigenvals()
````

- エネルギーは $\pm\sqrt{\varepsilon^2 + V^2}$。V = 1 として、$\varepsilon$ を変えたときのエネルギーを描く。灰色の線は行き来がない（V = 0）ときの $\pm\varepsilon$

````python
E = sp.sqrt(eps**2 + 1)
p = sp.plot(E, -E, eps, -eps, (eps, -3, 3), aspect_ratio=(1, 1), title='準位の反発（横：ε、縦：エネルギー）',
            xlabel='', ylabel='', show=False)
p[0].line_color = 'tomato'
p[1].line_color = 'steelblue'
p[2].line_color = 'lightgray'
p[3].line_color = 'lightgray'
p.show()
````

- 軸が原点にあると、軸名がグラフの真ん中で目盛りと重なる。ここでは `xlabel=''`・`ylabel=''` で軸名を消し、タイトルに書いた
- 行き来がないと、2本の線は $\varepsilon = 0$ で**交わる**。行き来があると、交わらずに**よけ合う**（間が $2V$ だけ開く）。これを**準位の反発**（避けられた交差）という

## 9-14 パウリ行列（紹介）

- 2状態の量子力学でよく使う3つの行列を**パウリ行列**という。どれもエルミート行列で、固有値は $\pm 1$

````python
sx = sp.Matrix([[0, 1], [1, 0]])
sy = sp.Matrix([[0, -sp.I], [sp.I, 0]])
sz = sp.Matrix([[1, 0], [0, -1]])
sx.eigenvals(), sy.eigenvals(), sz.eigenvals()
````

- 9-11 のハミルトニアンは、単位行列とパウリ行列で $H = E_0 E - \Delta\sigma_x$ と書ける

````python
E0 * sp.eye(2) - Dl * sx == H
````

- かける順番を変えると結果が変わる（4-3 の `A * B` と `B * A` の違い）。$\sigma_x\sigma_y - \sigma_y\sigma_x = 2i\sigma_z$ で 0 にならない。量子力学では、このとき2つの量（スピンの x 成分と y 成分）を**同時には決められない**ことを表す

````python
sx * sy - sy * sx
````
