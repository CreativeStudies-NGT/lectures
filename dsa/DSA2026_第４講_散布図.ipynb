## G-0 平均と補助線の完成形

8人のテストの点数を点だけのグラフにし、**平均**の線と、**平均 ± 標準偏差**の線を引いたもの。

1. [授業HP の **code** ページ](https://www.ces-alpha.org/hp/DSA2026/code/#lec04)の [**4-2**](https://www.ces-alpha.org/hp/DSA2026/code/#lec04b) を**全部コピーして**自分の notebook のセルに貼り付け、実行する

```python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib

score = [72, 85, 64, 90, 58, 77, 81, 69]
number = [1, 2, 3, 4, 5, 6, 7, 8]

mean = np.mean(score)  # [1] 平均
std = np.std(score)  # [2] 標準偏差
print('平均', mean, '標準偏差', std)

plt.figure(figsize=(8, 5))  # [0] グラフの大きさ
plt.plot(number, score, marker='o', markersize=10, linestyle='none', color='orange')  # 線を消した折れ線（点だけ）
plt.axhline(y=mean, color='r', linestyle='--', label='平均')  # [3] 平均の補助線
plt.axhline(y=mean + std, color='gray', linestyle=':', label='平均 ± 標準偏差')  # [4] 平均 + 標準偏差
plt.axhline(y=mean - std, color='gray', linestyle=':')  # [4] 平均 - 標準偏差
plt.ylim(0, 100)  # [5] 軸・文字・凡例
plt.title('小テストの点数', fontsize=20)
plt.xlabel('出席番号', fontsize=14)
plt.ylabel('点数', fontsize=14)
plt.legend()
plt.show()
```

## G-1 平均

- **平均**は、全部足して個数で割ったもの。`np.mean(リスト)` で計算できる
- 前回やった `sum(リスト) / len(リスト)` と同じ結果になる

```python
print(np.mean(score))
print(sum(score) / len(score))
```

- 計算した値は、変数（ここでは `mean`）に入れておくと、あとで何度も使える
- 平均の値を**自分で計算して数字で書き込まない**。`np.mean(score)` と書いておけば、データを変えても自動で計算し直される

## G-2 最大・最小・合計・中央値・四分位数

- 前回は `max(a)`、`min(a)`、`sum(a)` を使った。numpy にも同じ働きの道具がある

| 知りたいもの | 書き方 |
|---|---|
| 最大 | `np.max(リスト)` |
| 最小 | `np.min(リスト)` |
| 合計 | `np.sum(リスト)` |
| 中央値（大きさの順に並べて真ん中の値） | `np.median(リスト)` |
| 第一四分位数・第三四分位数 | `np.percentile(リスト, 25)`、`np.percentile(リスト, 75)` |

```python
print('最大', np.max(score))
print('最小', np.min(score))
print('合計', np.sum(score))
print('中央値', np.median(score))
```

- 中央値は、極端な値が1つあっても動きにくい。年収や家の値段など、一部に大きい値があるデータでは平均と中央値がずれる

```python
income = [300, 350, 400, 420, 450, 3000]
print('平均', np.mean(income))
print('中央値', np.median(income))
```

### 四分位数

- **四分位数**は前期に学んだ。データを小さい順に並べて**4等分する区切り**の値。numpy で求められる
  - 第一四分位数（下から 1/4）：`np.percentile(リスト, 25)`
  - 第二四分位数（真ん中）：**中央値**と同じ。`np.percentile(リスト, 50)` は `np.median(リスト)` と同じ値
  - 第三四分位数（下から 3/4）：`np.percentile(リスト, 75)`
- `percentile` は**百分位数**。`25` は「下から 25% の位置」という意味

```python
print('第一四分位数', np.percentile(score, 25))
print('中央値', np.percentile(score, 50))
print('第三四分位数', np.percentile(score, 75))
```

- 第一四分位数から第三四分位数までに、**真ん中の半分（50%）**のデータが入る。この幅（第三四分位数 − 第一四分位数）を**四分位範囲**という
- 四分位範囲が広いほど、真ん中の半分がばらついている。第７講の**箱ひげ図**の「箱」が、この範囲になる
- 四分位数の求め方はいくつかある。numpy は高校の教科書の手計算（前半・後半それぞれの中央値）と少し違う値になることがある（この score では、numpy は 67.75 と 82.0、教科書の方法では 66.5 と 83）

## G-3 標準偏差

- **標準偏差**は、データの**ばらつき**（平均からどれくらい離れているか）を表す数。`np.std(リスト)` で計算できる
- 平均が同じでも、ばらつきが違うことがある。下の2つのクラスは平均がどちらも 70 点

```python
class_a = [68, 70, 72, 69, 71, 70]
class_b = [40, 100, 55, 85, 60, 80]

print('A組 平均', np.mean(class_a), '標準偏差', np.std(class_a))
print('B組 平均', np.mean(class_b), '標準偏差', np.std(class_b))
```

- 標準偏差が**小さい**＝みんな平均の近く、**大きい**＝ばらばら
- 多くのデータでは、「平均 ± 標準偏差」の範囲に全体のおよそ3分の2が入る
- 標準偏差は第10講のエラーバーでも使う

## H-1 散布図の基本

- `plt.plot(x, y)` を `plt.scatter(x, y)` に変えるだけ
- x と y は**同じ個数**にする（違うと `ValueError: x and y must be the same size`）

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 6, 3, 12, 10, 5]

plt.figure(figsize=(8, 5))
plt.scatter(x, y)
plt.show()
```

### 折れ線グラフとの違い

| | 折れ線グラフ | 散布図 |
|---|---|---|
| 関数 | `plt.plot(x, y)` | `plt.scatter(x, y)` |
| 線 | 点を線でつなぐ | 線なし（点だけ） |
| 向いているもの | 時間の流れなど、**順番に意味がある**データ | 2つの量の**関係** |
| 点（マーカー） | 何も書かないと点なし。`marker='o'` で付く | 何も書かないと `'o'` |
| マーカーの種類・色 | `marker=`、`color=` | `marker=`、`color=`（同じ） |
| **点の大きさ** | `markersize=10`（`ms=10`） | **`s=100`**（名前が違う！ **面積**なので、`s` は `markersize` のおよそ2乗） |
| 点ごとの大きさ・色 | できない（全部同じ） | `s=リスト`、`color=リスト`（H-3・H-4） |
| 点のふちの色 | `markeredgecolor='k'` | **`edgecolor='k'`**（名前が違う） |
| 何も書かないときの重なる順番 | `zorder` は 2 | `zorder` は 1（線が点の上になる。H-9） |
| 線の種類・太さ | `linestyle=`、`linewidth=` | 使わない（点を線でつながないので） |

- 名前の違うオプションを書くとエラーになる（散布図に `markersize=`、折れ線に `s=` や `edgecolor=`）
- もっと細かい比較（点の中の色・ふちの太さなど）は、[授業HP notes の「折れ線と散布図のオプション比較」](https://www.ces-alpha.org/hp/DSA2026/notes/#plotvsscatter)

## H-3 点の大きさ

- 散布図の点の大きさは **`s=`** で指定する（**`markersize=` ではない**）
- `s=` は点の**面積**。何も書かないときは `s=36` くらい。`s=100` で折れ線の `markersize=10` と同じくらい
- 面積なので、`s=100` → `s=400`（4倍）で、見た目の幅は2倍になる

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', color='b', s=100)
plt.show()
```

### 点ごとに大きさを変える：`s=リスト`

- `s=` に**リスト**を渡すと、1つ目の点にはリストの1つ目、2つ目の点には2つ目…の大きさが付く
- リストの数は**自分で決めて**書いてよい。同じ数が並んでもよい
- リストの個数は、**点の個数（x・y の個数）と同じ**にする

```python
size_list = [10, 10, 20, 40, 160, 400]

plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', color='b', s=size_list)
plt.show()
```

- 左の2つ（`10`）は、何も書かないとき（`s=36` くらい）より小さい。右の2つ（`160`、`400`）は大きく目立つ
- どの点を目立たせるかを、大きさで決められる。リストの数を書き換えて、どの点が大きくなるか確かめよう
- 演習の問題文ではリストの名前が `markersize_list` だが、渡すのは **`s=markersize_list`**。`markersize=` と書くとエラーになる：
  `AttributeError: PathCollection.set() got an unexpected keyword argument 'markersize'`

▶ **演習**：`scatter_描画基礎003`

- 下は、H-0 の Aクラスの身長（x）と体重（y）に、年齢 `age` を点の大きさとして加えたもの
- 年齢の数（18〜60）をそのまま `s=` に渡すと、点が小さくて見えにくい。そこで `np.array(age) * 10` で**全部10倍**にしている（なぜ `np.array` にするかは発展 I-1）
- 下のセルは、[授業HP の **code** ページ](https://www.ces-alpha.org/hp/DSA2026/code/#lec04)の [**4-6**](https://www.ces-alpha.org/hp/DSA2026/code/#lec04f)（身長と体重の散布図）を貼り、上に `age_size` を作る行、`plt.scatter` に `s=age_size, alpha=0.6` を足したもの

## H-6 重なる順番

点が重なると、**後から描いた点が上**にきて、下の点が隠れる。演習「第４講」の 006〜010 は、次の重なり合うデータを使う。[授業HP の **code** ページ](https://www.ces-alpha.org/hp/DSA2026/code/#lec04)の [**4-7**](https://www.ces-alpha.org/hp/DSA2026/code/#lec04g) を貼って実行する。H-6〜H-8 では、これの `plt.scatter` に `zorder=`・`alpha=`・`label=` を足していく。

## H-9 散布図に補助線を引く

- G-4 の補助線（`plt.axhline`・`plt.axvline`）は、散布図にも同じように引ける
- 散布図では、何も書かないと**線が点の上**に描かれる。重なる順番は H-6 の `zorder=` で決める

### 点と補助線の重なる順番：`zorder`

- H-6 の `zorder=` は、点と補助線の間でも使える。**数が大きいほど上（手前）**に描かれる
- 何も書かないときは、**点（`plt.scatter`）は 1、線（`plt.axhline`・`plt.axvline`・`plt.plot`）は 2**。だから、何もしないと**線が点の上**に描かれる
- 違いが分かりやすいように、点を大きく（`s=600`）、線を太く（`linewidth=6`）して、同じグラフを3通りで比べる

**[1] 何も書かない**：線（2）が点（1）の上。点が線で切られて見える

```python
plt.figure(figsize=(6, 4))
plt.scatter(height_a, weight_a, s=600, color='orange')
plt.axhline(y=61, color='b', linewidth=6)
plt.axvline(x=170, color='g', linewidth=6)
plt.xlim(150, 190)
plt.ylim(40, 80)
plt.title('[1] zorder なし')
plt.show()
```

**[2] 点に `zorder=3`**：点（3）が線（2）の上。点が線に隠れない

```python
plt.figure(figsize=(6, 4))
plt.scatter(height_a, weight_a, s=600, color='orange', zorder=3)
plt.axhline(y=61, color='b', linewidth=6)
plt.axvline(x=170, color='g', linewidth=6)
plt.xlim(150, 190)
plt.ylim(40, 80)
plt.title('[2] 点を上に')
plt.show()
```

**[3] 線ごとに変える**：横の線に `zorder=4`、縦の線に `zorder=1`。点は `zorder=3`
- 上から順に、横の線（4）→ 点（3）→ 縦の線（1）

```python
plt.figure(figsize=(6, 4))
plt.scatter(height_a, weight_a, s=600, color='orange', zorder=3)
plt.axhline(y=61, color='b', linewidth=6, zorder=4)
plt.axvline(x=170, color='g', linewidth=6, zorder=1)
plt.xlim(150, 190)
plt.ylim(40, 80)
plt.title('[3] 横の線 > 点 > 縦の線')
plt.show()
```

| | 横の線 | 点 | 縦の線 | 一番上に見えるもの |
|---|---|---|---|---|
| [1] 何も書かない | 2 | 1 | 2 | 線（2本は後に書いた縦の線が上） |
| [2] 点に `zorder=3` | 2 | 3 | 2 | 点 |
| [3] 線ごとに変える | 4 | 3 | 1 | 横の線 |

- 同じ数どうしは、H-6 と同じく**後から書いたものが上**になる（[1] で縦の線が横の線の上にあるのはこのため）
- ふだんは、[2] のように**点に `zorder=3`** を付けておけば、点が補助線に隠れない

### 使ってみよう：平均の線を引く

- 身長の平均の縦の線と、体重の平均の横の線を引く。平均は G-1 の `np.mean(リスト)` で計算する
- [授業HP の **code** ページ](https://www.ces-alpha.org/hp/DSA2026/code/#lec04)の [**4-6**](https://www.ces-alpha.org/hp/DSA2026/code/#lec04f) を貼り、平均を計算する行と、`plt.axvline`・`plt.axhline`・`plt.legend()` の行を足す（点には `zorder=3`）

## I-2 平均 ± 標準偏差の範囲を塗る

- `plt.axhspan(下, 上)` で、横方向の帯を塗る。`alpha=` で薄くする

```python
mean = np.mean(score)
std = np.std(score)

plt.figure(figsize=(8, 5))
plt.scatter(number, score, s=100, zorder=3)
plt.axhline(y=mean, color='r', linestyle='--', label='平均')
plt.axhspan(mean - std, mean + std, color='r', alpha=0.1, label='平均 ± 標準偏差')
plt.ylim(0, 100)
plt.legend()
plt.show()
```
