## G-1 散布図の基本

- `plt.plot(x, y)` を `plt.scatter(x, y)` に変えるだけ
- x と y は**同じ個数**にする（違うと `ValueError: x and y must be the same size`）

```python
x = [1, 2, 3, 4, 5, 6]
y = [2, 6, 3, 12, 10, 5]

plt.figure(figsize=(8, 5))
plt.scatter(x, y)
plt.show()
```

### 折れ線グラフとの違い

| | 折れ線グラフ | 散布図 |
|---|---|---|
| 関数 | `plt.plot(x, y)` | `plt.scatter(x, y)` |
| 線 | 点を線でつなぐ | 線なし（点だけ） |
| 向いているもの | 時間の流れなど、**順番に意味がある**データ | 2つの量の**関係** |
| マーカーの種類・色 | `marker=`、`color=` | `marker=`、`color=`（同じ） |
| **点の大きさ** | `markersize=10` | **`s=100`**（名前が違う！） |

▶ **演習**：`scatter_描画基礎001`（上のセルと同じ。これで1問解ける）

## G-2 マーカーの種類と色

- 折れ線グラフと同じく、`marker=` と `color=` を足す

| 書き方 | 形 | 書き方 | 形 |
|---|---|---|---|
| `'o'` | ● 丸（何も書かないときはこれ） | `'*'` | ★ 星 |
| `'s'` | ■ 四角 | `'D'` | ◆ ひし形 |
| `'^'` | ▲ 上向き三角 | `'x'` | × バツ |

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='^', color='g')
plt.show()
```

- `color=` の代わりに `c=` と書いてもよい（`color='g'` / `c='g'`）

▶ **演習**：`scatter_描画基礎002`

## G-3 点の大きさ

- 散布図の点の大きさは **`s=`** で指定する（**`markersize=` ではない**）
- `s=` は点の**面積**。何も書かないときは `s=36` くらい。`s=100` で折れ線の `markersize=10` と同じくらい
- 面積なので、`s=100` → `s=400`（4倍）で、見た目の幅は2倍になる

```python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', color='b', s=100)
plt.show()
```

### 点ごとに大きさを変える：`s=リスト`

- `s=` に**リスト**を渡すと、1つ目の点にはリストの1つ目、2つ目の点には2つ目…の大きさが付く
- x・y に続く**3つ目の情報**（人口、売上など）を、点の大きさで見せられる
- リストの個数は、**点の個数（x・y の個数）と同じ**にする

```python
markersize_list = [100, 200, 300, 400, 500, 600]

plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', color='b', s=markersize_list)
plt.show()
```

- 演習の問題文ではリストの名前が `markersize_list` だが、渡すのは **`s=markersize_list`**。`markersize=` と書くとエラーになる：
  `AttributeError: PathCollection.set() got an unexpected keyword argument 'markersize'`

▶ **演習**：`scatter_描画基礎003`

## G-4 点ごとに色を変える

- `color=` にも**リスト**を渡せる。点ごとに色が変わる
- 例：最初の3つ（x が 1, 2, 3）は女性、残りの3つは男性として、色で分ける
- リストは**手で書けばよい**（短く書く方法は H-2 で学ぶ）

```python
mycolors = ['m', 'm', 'm', 'b', 'b', 'b']

plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', s=100, color=mycolors)
plt.show()
```

- 色のリストの個数が点の個数と違うとエラーになる：
  `ValueError: 'c' argument has 2 elements, which is inconsistent with 'x' and 'y' with size 6.`

▶ **演習**：`scatter_描画基礎004`

## G-5 2つのグループを重ねる

- データが**グループごとに別のリスト**になっているときは、`plt.scatter` を**2回書く**（前回の `plt.plot` を2回書いたのと同じ）
- グループごとに色やマーカーを変えると区別しやすい

```python
female_x = [1, 2, 3]
female_y = [2, 6, 3]
male_x = [4, 5, 6]
male_y = [12, 10, 5]

plt.figure(figsize=(8, 5))
plt.scatter(female_x, female_y, marker='o', s=100, color='m')
plt.scatter(male_x, male_y, marker='s', s=100, color='b')
plt.show()
```

- G-4（色のリスト）と同じ見た目になる。データが1つのリストなら G-4、グループごとのリストなら G-5 の書き方

▶ **演習**：`scatter_描画基礎005`

## G-6 重なる順番

点が重なると、**後から描いた点が上**にきて、下の点が隠れる。演習「第４講」の 006〜010 は、次の重なり合うデータを使う。

```python
female_x = [3, 4.1, 4.9]
female_y = [3, 4.8, 5]
male_x = [3, 4, 5]
male_y = [3.1, 4.7, 5]

plt.figure(figsize=(8, 5))
plt.scatter(female_x, female_y, s=500, color='m')
plt.scatter(male_x, male_y, s=500, color='b')
plt.xlim(2.5, 5.5)
plt.ylim(2.5, 5.5)
plt.show()
```

- 男性（青）を後から描いたので、青が上になっている
- `zorder=` で重なる順番を決められる。**数が大きいほど上（手前）**に描かれる
- 大きな点が枠からはみ出しそうなときは、`plt.xlim`・`plt.ylim` で範囲を少し広げる

```python
plt.figure(figsize=(8, 5))
plt.scatter(female_x, female_y, s=500, color='m', zorder=2)
plt.scatter(male_x, male_y, s=500, color='b', zorder=1)
plt.xlim(2.5, 5.5)
plt.ylim(2.5, 5.5)
plt.show()
```

▶ **演習**：`scatter_描画基礎006`

## G-7 透明度

- `alpha=` で点を**透けさせる**。**0 で透明、1 で不透明**（何も書かないときは 1）
- 0.5 くらいにすると、重なった部分の下の点も透けて見える
- 点の数が多いとき（数百〜数千）は、`alpha=0.3` くらいにすると、点が集まっている場所が濃く見えて分かりやすい

```python
plt.figure(figsize=(8, 5))
plt.scatter(female_x, female_y, s=500, color='m', alpha=0.5)
plt.scatter(male_x, male_y, s=500, color='b', alpha=0.5)
plt.xlim(2.5, 5.5)
plt.ylim(2.5, 5.5)
plt.show()
```

▶ **演習**：`scatter_描画基礎007`

## G-8 凡例とその場所

- 折れ線グラフと同じく、`label='...'` でグループに名前を付け、`plt.legend()` で**凡例**を表示する

```python
plt.figure(figsize=(8, 5))
plt.scatter(female_x, female_y, s=500, color='m', alpha=0.5, label='女性')
plt.scatter(male_x, male_y, s=500, color='b', alpha=0.5, label='男性')
plt.xlim(2.5, 5.5)
plt.ylim(2.5, 5.5)
plt.legend()
plt.show()
```

- `label=` を書かずに `plt.legend()` だけ書くと、凡例は出ず `No artists with labels found to put in legend.` という注意が出る

▶ **演習**：`scatter_描画基礎008`

### 凡例の場所を言葉で指定する：`loc='...'`

| `loc=` | 場所 | `loc=` | 場所 |
|---|---|---|---|
| `'upper left'` | 左上 | `'upper right'` | 右上 |
| `'lower left'` | 左下 | `'lower right'` | 右下 |
| `'center'` | 中央 | `'best'` | 点と重なりにくい場所（何も書かないときはこれ） |

```python
plt.figure(figsize=(8, 5))
plt.scatter(female_x, female_y, s=500, color='m', alpha=0.5, label='女性')
plt.scatter(male_x, male_y, s=500, color='b', alpha=0.5, label='男性')
plt.xlim(2.5, 5.5)
plt.ylim(2.5, 5.5)
plt.legend(loc='upper left')
plt.show()
```

▶ **演習**：`scatter_描画基礎009`

### 凡例の場所を数値で指定する：`loc=(横, 縦)`

- グラフの枠の**左下を `(0, 0)`、右上を `(1, 1)`** として、凡例の**左下の角**を置く位置を指定する
- 0〜1 の範囲なら枠の中、それ以外なら**枠の外**に出る。例：`loc=(1.02, 0.8)` で枠の右外

```python
plt.figure(figsize=(8, 5))
plt.scatter(female_x, female_y, s=500, color='m', alpha=0.5, label='女性')
plt.scatter(male_x, male_y, s=500, color='b', alpha=0.5, label='男性')
plt.xlim(2.5, 5.5)
plt.ylim(2.5, 5.5)
plt.legend(loc=(1.02, 0.8))
plt.show()
```

- 枠の外に出すと、保存した画像で凡例が切れることがある。そのときは `plt.show()` の前に `plt.tight_layout()` を書く

▶ **演習**：`scatter_描画基礎010`

# H. リストの操作

散布図では、点の大きさや色を**リスト**で渡した（G-3・G-4）。点が多くなると、リストを手で書くのは大変になる。ここではリストを**短く作る・数える・書き換える・追加する**方法を学ぶ。

## H-1 リストの作成と参照（復習）

- `[ ]` の中に `,` で区切って値を並べる
- 値を取り出すときは `リスト名[番号]`。**番号は 0 から**

```python
a = [1, 2, 3, 4]

print(a[0])
print(a[2])
```

▶ **演習**：`リストの作成と参照基礎001`

## H-2 同じものを並べる・個数を数える

- `リスト * 個数` で、同じ要素を並べたリストが作れる

```python
ini_list = ['y']
colorlist = ini_list * 5

print(colorlist)
```

- `len(リスト)` で、リストの**要素の個数**が分かる（第２講）
- `len` と組み合わせると、**別のリストと同じ個数**のリストが作れる。数字を直接書かない（**ベタ打ち・ハードコード**しない）ので、data の個数が変わっても直さなくてよい

```python
data = [7, 1, 3, 6, 4, 6, 1, 8, 0, 2, 1, 0, 5, 3, 1, 3, 6, 3, 4, 0]

mylist = ['y'] * len(data)

print(len(data))
print(mylist)
```

▶ **演習**：`リストの作成と参照基礎002`、`リストの作成と参照基礎003`

### 使ってみよう：G-4 の色のリストを短く書く

- リストどうしは `+` でつなげられる

```python
mycolors = ['m'] * 3 + ['b'] * 3
print(mycolors)

plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', s=100, color=mycolors)
plt.show()
```

## H-3 要素を書き換える

- `リスト名[番号] = 新しい値` で、その番号の要素だけを書き換える

```python
mylist = ['y'] * len(data)
mylist[0] = 'r'

print(mylist)
```

- 先頭は `[0]`、最後は `[-1]` と書ける

▶ **演習**：`リストの作成と参照基礎004`

## H-4 要素を追加する

- `リスト名.append(値)` で、リストの**最後に**要素を1つ追加する
- 空のリスト `[]` から始めて、`append` で1つずつ足していくことが多い（I-1 の for 文でも使う）

```python
emptylist = []
emptylist.append('abc')
emptylist.append(123)

print(emptylist)
```

- 文字列と数値のように、**種類の違う値**を1つのリストに入れてもよい

▶ **演習**：`リストの作成と参照基礎005`

## I-1 for 文で色のリストを作る

- 「y が 6 以上なら赤、そうでなければ灰色」のように、**データに応じて**色を決めたいときは、for 文で1つずつ `append` する
- `for 変数 in リスト:` の次の行は**字下げ**（スペース4つ）する。字下げした行が、リストの要素の数だけくり返される

```python
mycolors = []
for value in y:
    if value >= 6:
        mycolors.append('r')
    else:
        mycolors.append('gray')

print(mycolors)

plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', s=100, color=mycolors)
plt.axhline(y=6, color='k', linestyle='--')
plt.show()
```

## I-3 カラーマップ

- 色のリストを手で書く代わりに、**数値**に応じて色を自動で付ける
- `c=` に**数値のリスト**、`cmap=` に色の並び（`'viridis'`、`'coolwarm'`、`'Reds'` など）を指定する
- `plt.colorbar()` で、どの色がどの数値かを示す目盛りが付く

```python
age = [18, 25, 40, 33, 60]

plt.figure(figsize=(8, 5))
plt.scatter(height_a, weight_a, c=age, cmap='viridis', s=200)
plt.colorbar(label='年齢')
plt.show()
```

# まとめ

| やりたいこと | 書き方 |
|---|---|
| グラフの大きさ（基本設定） | `plt.figure(figsize=(8, 5))`（`scatter` より前） |
| 散布図 | `plt.scatter(x, y)` |
| マーカーの種類 | `marker='o'`（`'s'` `'^'` `'*'` `'D'` など） |
| 色 | `color='r'`（短く `c='r'`） |
| 点の大きさ | `s=100`（**面積**。`markersize` ではない） |
| 点ごとの大きさ・色 | `s=リスト`、`color=リスト`（点と同じ個数） |
| 2つのグループ | `plt.scatter` を2回書く |
| 重なる順番 | `zorder=2`（大きいほど上） |
| 透明度 | `alpha=0.5`（0 透明〜1 不透明） |
| 凡例 | `label='...'` を付けて `plt.legend()` |
| 凡例の場所 | `plt.legend(loc='upper left')`、数値なら `loc=(0.05, 0.75)` |
| **以下はリスト** | |
| 同じものを並べる | `['y'] * 5` |
| 個数を数える | `len(リスト)` |
| 同じ個数のリスト | `['y'] * len(data)` |
| リストをつなぐ | `['m'] * 3 + ['b'] * 3` |
| 要素を書き換える | `mylist[0] = 'r'` |
| 要素を追加する | `mylist.append(値)` |
