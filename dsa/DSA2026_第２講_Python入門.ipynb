## 起動方法

- **① Jupyter Notebook を起動 → ② ブラウザのファイル一覧から ipynb をクリックして開く**の順番で開く

- **ipynb ファイルをダブルクリックしても起動しない**。「このファイルを開くアプリ」を聞かれても、選ばずに閉じる

## ファイルの保存場所

- Jupyter のファイル一覧に最初に表示されるのは**ホームフォルダ**（Windows：`C:\Users\ユーザー名`、Mac：家のアイコンのフォルダ）。何もしなければ、ipynb はここに保存される
- **授業用のフォルダを作る**：Jupyter のファイル一覧で「New」→「新規フォルダ」（英語表示では「New Folder」）→ 名前を `DSA2026` などに変える。以後の ipynb は、この中に保存する
  - フォルダ名やファイル名に日本語が入っていても大丈夫（ユーザー名のフォルダだけは日本語を含まないこと）
- **デスクトップにショートカットを作る**（提出のときにすぐたどり着ける）
  - Windows：エクスプローラーでホームフォルダを開く → `DSA2026` を右クリック →（Windows 11 は「その他のオプションを確認」）→「送る」→「デスクトップ（ショートカットを作成）」
  - Mac：Finder でホームフォルダを開く → `DSA2026` を **option + command を押しながら**デスクトップへドラッグ
- 前回のファイルはホームフォルダにあるので、作ったフォルダに移しておく

## A-3 計算

| 演算子 | 意味 | 例 | 結果 |
|---|---|---|---|
| `+` | 足し算 | `3 + 2` | 5 |
| `-` | 引き算 | `3 - 2` | 1 |
| `*` | 掛け算 | `3 * 2` | 6 |
| `/` | 割り算 | `3 / 2` | 1.5 |
| `**` | べき乗 | `3 ** 2` | 9 |

- 計算の順番は数学と同じ（`**` → `* /` → `+ -`）。迷ったら `( )` を付ける

````python
print(2 + 3 * 4)     # 掛け算が先
print((2 + 3) * 4)   # 括弧の中が先
````

- 変数を使って計算できる。例：税込価格、円の面積

````python
price = 1200              # 税抜価格（円）
total = price * 1.1       # 税込価格

print(total)
````

````python
r = 3                     # 半径
area = 3.14 * r**2        # 円の面積

print(area)
````

- 割り算や小数を含む計算の結果は小数になる（`1320.0` のように `.0` が付く）
- 小数の計算では、**桁がたくさん出たり、ごくわずかにずれたりすることがあるが正常**（コンピュータの小数の仕組みによるもの）

````python
print(10 / 4)
print(0.1 + 0.2)
````

## A-4 文字列

- 文字列どうしは `+` でつなげる（連結）

# B. データをまとめる

## B-1 リスト

- 複数の値を `[ ]` で囲み、カンマで区切って1つにまとめたもの
- 数値のリストも、文字列のリストも作れる
- **グラフの x 軸・y 軸のデータは、リストで渡す**

````python
temp = [22, 24, 19, 21, 25, 27, 23]               # 月〜日の最高気温
days = ['月', '火', '水', '木', '金', '土', '日']   # 曜日

print(temp)
print(days)
````

▶ **演習**：`リスト_入門001`, `リスト_入門002`

## B-2 要素数と要素の取り出し

- `len(リスト)` で要素の数を調べる

````python
print(len(temp))
````

- `リスト[番号]` で要素を取り出す。**番号は 0 から始まる**

| 曜日 | 月 | 火 | 水 | 木 | 金 | 土 | 日 |
|---|---|---|---|---|---|---|---|
| 番号 | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
| temp | 22 | 24 | 19 | 21 | 25 | 27 | 23 |

````python
print(days[2], temp[2])   # 3番目（水曜）
````

- 7つの要素の最後は `[6]`。`[7]` は存在しないので **IndexError** になる（下のセルはエラーになるのが正しい）

````python
print(temp[7])
````

- （発展）`temp[-1]` で最後の要素を取り出せる

````python
print(temp[-1])
````

## C-2 関数と引数

- `print(...)` や `len(...)` は**関数**。「**名前(引数)**」の形で使う。`( )` の中に渡す値を**引数**という
- `max`、`min` で最大値・最小値

````python
score = [65, 80, 72, 90, 58]

print(max(score))
print(min(score))
````

- 引数には2種類ある
  - **順番で意味が決まる引数**：`plt.plot(x, y)`（1番目が x、2番目が y）
  - **名前で指定する引数**（キーワード引数）：`color='b'`、`marker='s'` のように「名前=値」で書く

| 色 `color` | | マーカー `marker` | |
|---|---|---|---|
| `'r'` | 赤 | `'o'` | 丸 |
| `'g'` | 緑 | `'s'` | 四角 |
| `'b'` | 青 | `'^'` | 三角 |

````python
x = [1, 2, 3, 4, 5]

plt.plot(x, score, color='b', marker='s')
plt.show()
````

- `plt.` で始まるもの（`plot`、`title`、`show` …）は、「plt という道具箱の中の関数」
