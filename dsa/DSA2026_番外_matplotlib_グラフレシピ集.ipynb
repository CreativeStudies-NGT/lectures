## はじめに：plt.gca と plt.gcf

- matplotlib のグラフは、2つの部品でできている

| 部品 | 中身 | 取り出す命令 |
|---|---|---|
| Figure（フィギュア） | 画用紙全体。グラフを何枚並べても1枚 | `plt.gcf()`（get current figure：今の画用紙） |
| Axes（アクシーズ） | 画用紙の上の1つのグラフ。軸・目盛り・線などを持つ | `plt.gca()`（get current axes：今のグラフ） |

- `plt.title()` や `plt.xlabel()` は、裏で「今のグラフ（Axes）」に命令を出している。`plt.subplot()` で描く場所を変えると、「今のグラフ」も切り替わる
- ほとんどの命令は `plt.〜` で書けるが、plt に同じ名前の命令がないものがある。そのときは `plt.gca()` で今のグラフを取り出し、`plt.gca().命令()` と書く
- ax 形式のコードの `ax.命令()` は、plt 形式では `plt.gca().命令()` と同じ意味になる。`fig.命令()` は `plt.gcf().命令()`

**この notebook で使っているところ**

| 書き方 | すること | レシピ |
|---|---|---|
| `plt.gca().invert_yaxis()` | 縦軸を上下逆にする（横棒を上から順に並べる） | 4・11 |
| `plt.gca().set_theta_offset(…)`・`set_theta_direction(…)` | 極座標の 0° の位置と回る向きを決める | 16 |
| `plt.gca().collections` | 描いた図形（バイオリンなど）を取り出す | 39 |
| `plt.gca().xaxis.set_major_locator(…)`・`set_major_formatter(…)` | 日付の目盛りの間隔と書き方を決める | 40・42・44 |
| `plt.gca().set_facecolor(…)`・`plt.gca().spines` | グラフの内側の背景色、枠線 | 84 |
| `plt.gca().set_box_aspect(…)` | グラフの縦横比を決める | 88 |
| `transform=plt.gca().transAxes` | 文字の位置を、データの値ではなくグラフの左下 (0, 0)〜右上 (1, 1) で指定する | 102・103 |
| `plt.gca().add_collection(…)` | まとめて作った線をグラフに加える | 105 |
| `plt.gcf()` | アニメーションにする画用紙を渡す | 95 |
| `plt.gcf().canvas.draw_idle()` | スライダーを動かしたあと、画用紙を描き直す | 97 |

- 例：縦軸を上下逆にする

```python
import matplotlib.pyplot as plt
import japanize_matplotlib

plt.barh(['A', 'B', 'C'], [3, 5, 2])
plt.gca().invert_yaxis()
plt.title('A が上に来る')
plt.show()
```

## 1章 棒グラフ

### 1. 縦棒グラフ

- 変えた点：元のコードのまま（はじめから plt 形式）

```python
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

HIGHLIGHT = '#1E56A0'

df_sales = pd.DataFrame({
    '商品カテゴリ': ['ノートPC', '周辺機器', 'ソフトウェア', '保守サービス', 'クラウド', '研修'],
    '売上': [1280, 760, 540, 920, 1110, 430],
})

sales_by_category = (
    df_sales.groupby('商品カテゴリ')['売上']
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
plt.bar(sales_by_category.index, sales_by_category.values, color=HIGHLIGHT)
plt.title('商品カテゴリ別 売上合計')
plt.xlabel('商品カテゴリ')
plt.ylabel('売上（万円）')
plt.ylim(0, sales_by_category.max() * 1.15)
plt.xticks(rotation=30, ha='right')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

### 3. グループ縦棒グラフ

- 変えた点：元のコードのまま（はじめから plt 形式）

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

df_sales = pd.DataFrame({
    '地域': ['関東', '関東', '関東', '関西', '関西', '関西', '中部', '中部', '中部', '九州', '九州', '九州'],
    'チャネル': ['オンライン', '店舗', '電話'] * 4,
    '売上': [520, 430, 210, 410, 390, 180, 360, 310, 160, 300, 260, 140],
})

pivot = df_sales.groupby(['地域', 'チャネル'])['売上'].sum().unstack()

x = np.arange(len(pivot.index))
width = 0.24
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

plt.figure(figsize=(8, 5))
for i, channel in enumerate(pivot.columns):
    plt.bar(x + (i - 1) * width, pivot[channel], width, label=channel, color=colors[i])

plt.title('地域別・チャネル別 売上')
plt.xlabel('地域')
plt.ylabel('売上（万円）')
plt.xticks(x, pivot.index)
plt.ylim(0, pivot.max().max() * 1.2)
plt.legend(title='チャネル')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

### 4. グループ横棒グラフ

- 変えた点：ax.〜 を plt.〜 に。目盛りのラベルは plt.xticks(位置, ラベル)

```python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib

colors_professional = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

categories = [
    'エンタープライズ向け保守契約',
    '中堅企業向けクラウド移行',
    '小売向け分析ダッシュボード',
    '製造業向けIoT導入支援',
]
last_year = np.array([82, 64, 48, 55])
this_year = np.array([96, 71, 62, 68])

y = np.arange(len(categories))
bar_height = 0.34

plt.figure(figsize=(9, 5))
plt.barh(y - bar_height / 2, last_year, height=bar_height,
         color=colors_professional[0], label='前年売上')
plt.barh(y + bar_height / 2, this_year, height=bar_height,
         color=colors_professional[1], label='当年売上')

for values, offset in [(last_year, -bar_height / 2), (this_year, bar_height / 2)]:
    for index, value in enumerate(values):
        plt.text(value + 1.5, index + offset, f'{value}百万円',
                 va='center', fontsize=9)

plt.yticks(y, categories)
plt.gca().invert_yaxis()
plt.xlabel('売上')
plt.title('サービス別売上の前年比較')
plt.legend(loc='lower right')
plt.grid(axis='x', linestyle=':', alpha=0.4)
plt.xlim(0, 110)

plt.tight_layout()
plt.show()
```

## 2章 比較・順位のチャート

### 8. レーダーチャート

- 変えた点：ax.〜 を plt.〜 に。極座標は plt.subplot(polar=True)。目盛りのラベルは plt.xticks(位置, ラベル)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

df_scores = pd.DataFrame({
    '評価軸': ['品質', '価格', '機能性', 'サポート', '導入しやすさ', '拡張性'],
    '自社製品': [4.5, 3.8, 4.2, 4.6, 4.0, 4.1],
    '競合製品': [4.1, 4.4, 3.9, 3.7, 4.2, 3.8],
})

labels = df_scores['評価軸'].tolist()
angles = np.linspace(0, 2 * np.pi, len(labels), endpoint=False).tolist()
angles += angles[:1]

plt.figure(figsize=(6, 6))
plt.subplot(polar=True)

for column, color in [('自社製品', '#1E56A0'), ('競合製品', '#EA580C')]:
    values = df_scores[column].tolist()
    values += values[:1]
    plt.plot(angles, values, color=color, linewidth=2, label=column)
    plt.fill(angles, values, color=color, alpha=0.15)

plt.xticks(angles[:-1], labels)
plt.ylim(0, 5)
plt.yticks([1, 2, 3, 4, 5])
plt.title('製品評価の比較', pad=20)
plt.legend(loc='upper right', bbox_to_anchor=(1.2, 1.1))

plt.tight_layout()
plt.show()
```

### 9. ダンベルプロット

### 15. ゲージチャート

- 変えた点：ax.〜 を plt.〜 に。帯（Wedge）は plt.fill で描く円弧に、針の付け根（Circle）は plt.scatter に

```python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib

HIGHLIGHT = '#1E56A0'

score = 78
ranges = [
    (0, 60, '#F4B6B3', '要注意'),
    (60, 85, '#F8D9A0', '順調'),
    (85, 100, '#A8D5B5', '目標達成'),
]


def value_to_angle(value):
    return np.radians(180 - 180 * value / 100)


plt.figure(figsize=(8, 4.8))

for start, end, color, label in ranges:
    t = value_to_angle(np.linspace(start, end, 50))
    xs = np.concatenate([np.cos(t), 0.74 * np.cos(t[::-1])])
    ys = np.concatenate([np.sin(t), 0.74 * np.sin(t[::-1])])
    plt.fill(xs, ys, facecolor=color, edgecolor='white', linewidth=2)

    label_angle = value_to_angle((start + end) / 2)
    plt.text(0.87 * np.cos(label_angle), 0.87 * np.sin(label_angle),
             label, ha='center', va='center', fontsize=12, color='#1F1F1F',
             fontweight='bold')

needle_angle = value_to_angle(score)
plt.plot([0, 0.68 * np.cos(needle_angle)], [0, 0.68 * np.sin(needle_angle)],
         color='#222222', linewidth=4, solid_capstyle='round')
plt.scatter(0, 0, s=400, color='#222222', zorder=3)

for tick in [0, 20, 40, 60, 80, 100]:
    angle = value_to_angle(tick)
    plt.text(1.13 * np.cos(angle), 1.13 * np.sin(angle),
             str(tick), ha='center', va='center', fontsize=10, color='#333333')

plt.text(0, -0.22, f'顧客満足度スコア {score}',
         ha='center', va='center', fontsize=16,
         fontweight='bold', color=HIGHLIGHT)
plt.title('顧客満足度KPIゲージ', pad=14)
plt.axis('equal')
plt.xlim(-1.25, 1.25)
plt.ylim(-0.35, 1.2)
plt.axis('off')

plt.tight_layout()
plt.show()
```

### 28. サークルパッキング

- 変えた点：ax.〜 を plt.〜 に。円（Circle）は plt.fill で描く

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import circlify
import japanize_matplotlib

colors_professional = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']

df_sales = pd.DataFrame({
    'region': ['関東', '関東', '関東', '関西', '関西', '中部', '中部', '九州'],
    'category': ['PC', '周辺機器', '保守', 'PC', 'ソフトウェア', 'PC', '保守', 'ソフトウェア'],
    'sales': [6200000, 3100000, 1800000, 4200000, 2600000, 3300000, 1500000, 2100000]
})

hierarchy = []
for region, group in df_sales.groupby('region'):
    hierarchy.append({
        'id': region,
        'datum': group['sales'].sum(),
        'children': [
            {'id': row['category'], 'datum': row['sales'], 'parent': region}
            for _, row in group.iterrows()
        ]
    })

circles = circlify.circlify(
    hierarchy,
    show_enclosure=False,
    target_enclosure=circlify.Circle(x=0, y=0, r=1)
)

region_color = {item['id']: colors_professional[i] for i, item in enumerate(hierarchy)}
t = np.linspace(0, 2 * np.pi, 100)

plt.figure(figsize=(8, 8))
for circle in circles:
    xs = circle.x + circle.r * np.cos(t)
    ys = circle.y + circle.r * np.sin(t)
    if circle.level == 1:
        color = region_color[circle.ex['id']]
        plt.fill(xs, ys, color=color, alpha=0.18)
        plt.text(circle.x, circle.y + circle.r * 0.75, circle.ex['id'], ha='center', va='center', fontsize=13, weight='bold')
    elif circle.level == 2:
        color = region_color[circle.ex['parent']]
        plt.fill(xs, ys, color=color, alpha=0.65)
        plt.text(circle.x, circle.y, circle.ex['id'], ha='center', va='center', fontsize=10, color='white', weight='bold')

plt.title('地域・商品カテゴリ別売上構成（サークルパッキング）', fontsize=16, pad=16)
plt.axis('equal')
plt.xlim(-1.05, 1.05)
plt.ylim(-1.05, 1.05)
plt.axis('off')

plt.tight_layout()
plt.show()
```

### 29. デンドログラム

- 変えた点：ax.〜 を plt.〜 に

```python
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.stats import zscore

df_customers = pd.DataFrame({
    'segment': ['法人高単価', '法人高単価', '法人標準', '法人標準', '個人優良', '個人優良', '個人一般', '個人一般'],
    'city': ['東京', '大阪', '東京', '名古屋', '東京', '福岡', '大阪', '福岡'],
    'annual_income': [950, 880, 720, 690, 610, 570, 430, 390],
    'monthly_spend': [48, 44, 32, 29, 27, 24, 16, 14],
    'purchase_count': [32, 29, 24, 21, 20, 18, 12, 10]
})

profile = df_customers.groupby(['segment', 'city']).mean(numeric_only=True)
scaled_profile = profile.apply(zscore)

linked = linkage(scaled_profile, method='ward')

plt.figure(figsize=(10, 5))
dendrogram(
    linked,
    labels=[f'{segment}・{city}' for segment, city in profile.index],
    leaf_rotation=35,
    leaf_font_size=10,
)

plt.title('顧客グループの類似度（デンドログラム）', fontsize=15)
plt.ylabel('距離')
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

### 39. レインクラウドプロット

- 変えた点：ax.〜 を plt.〜 に。バイオリンの半分を消す部分だけ plt.gca() を使う。sns.set_theme に font='IPAexGothic' を足した（日本語が □ にならないように）

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.collections import PolyCollection

sns.set_theme(style='whitegrid', font='IPAexGothic')
import japanize_matplotlib

colors_professional = ['#1f77b4', '#ff7f0e', '#2ca02c']

rng = np.random.default_rng(42)
segments = ['法人', '個人', 'EC']
centers = [7.5, 6.8, 7.9]

df_score = pd.DataFrame({
    'セグメント': np.repeat(segments, 70),
    '満足度': np.concatenate([
        np.clip(rng.normal(center, 0.9, 70), 1, 10)
        for center in centers
    ]),
})

plt.figure(figsize=(10, 6))
sns.violinplot(
    data=df_score,
    x='セグメント',
    y='満足度',
    hue='セグメント',
    palette=colors_professional,
    inner=None,
    cut=0,
    linewidth=1.1,
    density_norm='width',
    legend=False,
)

for collection in plt.gca().collections:
    if isinstance(collection, PolyCollection):
        for path in collection.get_paths():
            vertices = path.vertices
            center = (vertices[:, 0].min() + vertices[:, 0].max()) / 2
            vertices[:, 0] = np.maximum(vertices[:, 0], center)

positions = np.arange(len(segments))
box_data = [
    df_score.loc[df_score['セグメント'] == segment, '満足度'].to_numpy()
    for segment in segments
]

plt.boxplot(
    box_data,
    positions=positions - 0.18,
    widths=0.12,
    patch_artist=True,
    showfliers=False,
    manage_ticks=False,
    boxprops=dict(facecolor='white', edgecolor='#333333', linewidth=1.1),
    medianprops=dict(color='#333333', linewidth=1.6),
    whiskerprops=dict(color='#333333', linewidth=1.1),
    capprops=dict(color='#333333', linewidth=1.1),
)

for i, (segment, color) in enumerate(zip(segments, colors_professional)):
    values = df_score.loc[df_score['セグメント'] == segment, '満足度'].to_numpy()
    x = rng.normal(i - 0.35, 0.035, size=len(values))
    plt.scatter(
        x,
        values,
        s=24,
        color=color,
        alpha=0.55,
        edgecolors='white',
        linewidths=0.4,
        zorder=3,
    )

score_min = df_score['満足度'].min()
score_max = df_score['満足度'].max()
margin = (score_max - score_min) * 0.1

plt.xticks(positions, segments)
plt.title('セグメント別満足度のレインクラウドプロット', fontsize=14, fontweight='bold')
plt.xlabel('顧客セグメント')
plt.ylabel('満足度（10点満点）')
plt.ylim(score_min - margin, score_max + margin)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

### 63. コードダイアグラム

- 変えた点：ax.〜 を plt.〜 に。部門の弧（Wedge）は plt.fill で、流れの曲線（PathPatch）は点を計算して plt.plot で描く

```python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib

colors_professional = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

departments = ['営業', '開発', 'マーケ', 'CS', '管理', '海外']
flows = [
    ('営業', 'CS', 18),
    ('営業', 'マーケ', 12),
    ('開発', '海外', 16),
    ('マーケ', '営業', 10),
    ('CS', '開発', 8),
    ('海外', '営業', 14),
    ('管理', '開発', 6),
    ('CS', '管理', 5),
]

node_count = len(departments)
angles = {name: 90 - index * 360 / node_count for index, name in enumerate(departments)}
color_map = {name: colors_professional[index] for index, name in enumerate(departments)}

plt.figure(figsize=(7.5, 7.5))

for name in departments:
    angle = angles[name]
    t = np.radians(np.linspace(angle - 22, angle + 22, 50))
    xs = np.concatenate([np.cos(t), 0.9 * np.cos(t[::-1])])
    ys = np.concatenate([np.sin(t), 0.9 * np.sin(t[::-1])])
    plt.fill(xs, ys, color=color_map[name], edgecolor='white', linewidth=2)

    rad = np.radians(angle)
    label_x = 1.18 * np.cos(rad)
    label_y = 1.18 * np.sin(rad)
    ha = 'left' if label_x > 0.05 else 'right' if label_x < -0.05 else 'center'
    plt.text(label_x, label_y, name, ha=ha, va='center', fontsize=12, fontweight='bold')

u = np.linspace(0, 1, 60)
for source, target, value in flows:
    a1 = np.radians(angles[source])
    a2 = np.radians(angles[target])
    px = [0.88 * np.cos(a1), 0.28 * np.cos(a1), 0.28 * np.cos(a2), 0.88 * np.cos(a2)]
    py = [0.88 * np.sin(a1), 0.28 * np.sin(a1), 0.28 * np.sin(a2), 0.88 * np.sin(a2)]
    xs = (1 - u)**3 * px[0] + 3 * (1 - u)**2 * u * px[1] + 3 * (1 - u) * u**2 * px[2] + u**3 * px[3]
    ys = (1 - u)**3 * py[0] + 3 * (1 - u)**2 * u * py[1] + 3 * (1 - u) * u**2 * py[2] + u**3 * py[3]
    plt.plot(xs, ys, color=color_map[source], linewidth=1.2 + value * 0.35, alpha=0.42, solid_capstyle='round')

plt.title('部門間の人材異動フロー', pad=18)
plt.axis('equal')
plt.xlim(-1.35, 1.35)
plt.ylim(-1.25, 1.30)
plt.axis('off')

plt.tight_layout()
plt.show()
```

### 90. 注釈とテキスト（annotate, text）

- 変えた点：plt.subplot で並べ、ax.〜 を plt.〜 に

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

rng = np.random.default_rng(42)

df_curve = pd.DataFrame({
    '期間': np.linspace(0, 10, 100)
})
df_curve['需要指数'] = np.sin(df_curve['期間']) * np.exp(-df_curve['期間'] / 5)

df_sales = pd.DataFrame({
    '月': ['1月', '2月', '3月', '4月', '5月', '6月'],
    '売上': [100, 85, 120, 95, 140, 160]
})
target = 120

colors_professional = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

plt.figure(figsize=(14, 6))
plt.suptitle('注釈・テキストシステム', fontsize=16, fontweight='bold')

plt.subplot(1, 2, 1)
plt.plot(
    df_curve['期間'],
    df_curve['需要指数'],
    linewidth=2,
    color=colors_professional[0]
)

peak_idx = df_curve['需要指数'].idxmax()
peak_x = df_curve.loc[peak_idx, '期間']
peak_y = df_curve.loc[peak_idx, '需要指数']

plt.annotate(
    f'最高点\n({peak_x:.1f}, {peak_y:.2f})',
    xy=(peak_x, peak_y),
    xytext=(peak_x + 1, peak_y + 0.2),
    arrowprops=dict(arrowstyle='->', color='red', lw=2),
    fontsize=12,
    fontweight='bold',
    bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.7)
)

plt.title('重要ポイント強調', fontweight='bold')
plt.xlabel('期間')
plt.ylabel('需要指数')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
bar_colors = [
    colors_professional[0] if sale >= target else colors_professional[3]
    for sale in df_sales['売上']
]
bars = plt.bar(df_sales['月'], df_sales['売上'], color=bar_colors, alpha=0.8)

plt.axhline(
    y=target,
    color='red',
    linestyle='--',
    linewidth=2,
    label=f'目標値: {target}'
)

for i, row in df_sales.iterrows():
    sale = row['売上']
    if sale >= target:
        plt.annotate(
            '達成!',
            xy=(i, sale),
            xytext=(i, sale + 10),
            ha='center',
            fontweight='bold',
            color='green',
            arrowprops=dict(arrowstyle='->', color='green')
        )
    else:
        shortage = target - sale
        plt.annotate(
            f'-{shortage}',
            xy=(i, sale),
            xytext=(i, sale + 10),
            ha='center',
            fontweight='bold',
            color='red'
        )

plt.title('売上目標達成状況', fontweight='bold')
plt.ylabel('売上（百万円）')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

rng = np.random.default_rng(42)

df_stream = pd.DataFrame({
    '時点': np.arange(200) * 0.1
})
df_stream['売上指数'] = np.sin(df_stream['時点']) + rng.normal(0, 0.1, len(df_stream))

colors_professional = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

plt.figure(figsize=(10, 6))
plt.xlim(0, 10)
plt.ylim(-2, 2)
plt.title('リアルタイムデータ可視化シミュレーション', fontweight='bold')
plt.xlabel('経過時間')
plt.ylabel('売上指数')
plt.grid(True, alpha=0.3)

x_data = []
y_data = []
line, = plt.plot([], [], linewidth=3, color=colors_professional[0], label='リアルタイムデータ')
plt.legend()

def animate(frame):
    row = df_stream.iloc[frame]
    x_data.append(row['時点'])
    y_data.append(row['売上指数'])

    if len(x_data) > 100:
        x_data.pop(0)
        y_data.pop(0)

    line.set_data(x_data, y_data)
    plt.xlim(max(0, max(x_data) - 10), max(x_data) + 1)
    return line,

anim = FuncAnimation(plt.gcf(), animate, frames=len(df_stream), interval=50, repeat=False)
plt.close()
HTML(anim.to_jshtml())
```

### 97. インタラクティブ要素（widgets）

- 変えた点：部品の場所を plt.axes([左, 下, 幅, 高さ]) で作り、描く場所を plt.sca で選ぶ。1行目の %matplotlib widget で動かせるようにした

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import CheckButtons, Slider
import japanize_matplotlib

rng = np.random.default_rng(42)
days = np.linspace(0, 30, 400)
baseline = 120
amp0, freq0, phase0 = 25, 0.7, 0.0

months = np.arange(1, 13)
month_labels = [f'{month}月' for month in months]
base_sales = np.array([100, 118, 105, 132, 148, 165, 178, 171, 190, 205, 218, 235], dtype=float)
regions = ['北海道', '関東', '関西', '九州']
regional_weights = {'北海道': 0.72, '関東': 1.35, '関西': 1.05, '九州': 0.86}
regional_data = {
    region: base_sales * regional_weights[region] * rng.uniform(0.94, 1.08, len(months))
    for region in regions
}

colors_professional = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']
colors_regions = dict(zip(regions, colors_professional))

plt.figure(figsize=(16, 9))

wave_area = plt.axes([0.05, 0.32, 0.42, 0.6])
demand = baseline + amp0 * np.sin(freq0 * days + phase0)
line, = plt.plot(days, demand, linewidth=2.5, color=colors_professional[0])
plt.xlim(0, 30)
plt.ylim(60, 180)
plt.title('需要変動シミュレーション', fontweight='bold', fontsize=14)
plt.xlabel('日数')
plt.ylabel('予測需要（件）')
plt.grid(True, alpha=0.3)

region_area = plt.axes([0.55, 0.32, 0.42, 0.6])

slider_freq = Slider(plt.axes([0.1, 0.14, 0.3, 0.03]), '周波数', 0.2, 1.5, valinit=freq0)
slider_amp = Slider(plt.axes([0.1, 0.07, 0.3, 0.03]), '振幅', 5, 60, valinit=amp0)
slider_phase = Slider(plt.axes([0.55, 0.14, 0.25, 0.03]), '位相', 0, 2 * np.pi, valinit=phase0)

def update_wave(_):
    demand = baseline + slider_amp.val * np.sin(slider_freq.val * days + slider_phase.val)
    line.set_ydata(demand)
    plt.sca(wave_area)
    plt.ylim(demand.min() - 15, demand.max() + 15)
    plt.gcf().canvas.draw_idle()

slider_freq.on_changed(update_wave)
slider_amp.on_changed(update_wave)
slider_phase.on_changed(update_wave)

check = CheckButtons(plt.axes([0.86, 0.02, 0.1, 0.16]), regions, [True, True, True, True])
plt.title('地域', fontweight='bold')

def draw_regions():
    plt.sca(region_area)
    plt.cla()
    active_regions = [region for region, active in zip(regions, check.get_status()) if active]

    for region in active_regions:
        plt.plot(months, regional_data[region], 'o-', linewidth=2, color=colors_regions[region], label=region)

    plt.title('地域別売上比較', fontweight='bold', fontsize=14)
    plt.xlabel('月')
    plt.ylabel('売上（百万円）')
    plt.xticks(months, month_labels, rotation=45)
    plt.grid(True, alpha=0.3)
    if active_regions:
        plt.legend()

def toggle_region(_):
    draw_regions()
    plt.gcf().canvas.draw_idle()

check.on_clicked(toggle_region)
draw_regions()

print('インタラクティブ要素の活用例:')
print('- パラメータ調整: リアルタイム結果確認')
print('- データフィルタリング: 動的表示切替')
print('- プレゼンテーション: 観客との対話型説明')
print('- 教育・トレーニング: 体験型学習支援')

plt.show()
```

- 変えた点：plt.subplot で並べ、ax.〜 を plt.〜 に。LineCollection を加える部分だけ plt.gca() を使う

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
import japanize_matplotlib

rng = np.random.default_rng(42)

n_points = 30_000
request_count = rng.gamma(shape=2.0, scale=120.0, size=n_points)
response_time = request_count * 0.35 + rng.normal(0, 45, n_points)
categories = rng.choice(['Webログ', '決済', '問い合わせ'], size=n_points, p=[0.55, 0.25, 0.20])

t = np.linspace(0, 1, 80)
segments = []
for _ in range(1000):
    phase = rng.uniform(0, 2 * np.pi)
    trend = rng.normal(0.25, 0.08) * t
    noise = rng.normal(0, 0.02, len(t))
    segments.append(np.column_stack([t, 0.4 + trend + 0.12 * np.sin(2 * np.pi * t + phase) + noise]))

large_matrix = rng.normal(120, 18, size=(450, 450))
sampled_matrix = large_matrix[::5, ::5]

time_axis = np.linspace(0, 24, 600)
daily_pattern = 100 + 18 * np.sin(2 * np.pi * time_axis / 24) + rng.normal(0, 2, len(time_axis))

colors_professional = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']
HIGHLIGHT = '#1E56A0'
ACCENT = '#EA580C'
GRAYS = ['#999999', '#BBBBBB', '#DDDDDD']

plt.figure(figsize=(18, 10))
plt.suptitle('静的グラフのパフォーマンス最適化', fontsize=16, fontweight='bold')

plt.subplot(2, 3, 1)
sample_index = rng.choice(n_points, size=5000, replace=False)
plt.scatter(
    request_count[sample_index],
    response_time[sample_index],
    s=5,
    alpha=0.35,
    color=HIGHLIGHT,
    rasterized=True,
)
plt.title('1. rasterized=True とサンプリング', fontweight='bold')
plt.xlabel('リクエスト数')
plt.ylabel('応答時間（ms）')
plt.grid(True, alpha=0.3)

plt.subplot(2, 3, 2)
collection = LineCollection(segments, colors=colors_professional[0], linewidths=0.4, alpha=0.18, rasterized=True)
plt.gca().add_collection(collection)
plt.autoscale()
plt.title('2. LineCollection による一括描画', fontweight='bold')
plt.xlabel('期間')
plt.ylabel('センサー値')
plt.grid(True, alpha=0.3)

plt.subplot(2, 3, 3)
image = plt.imshow(sampled_matrix, cmap='viridis', aspect='auto', interpolation='nearest')
plt.title('3. 行列の間引き表示', fontweight='bold')
plt.xlabel('設備列')
plt.ylabel('時間帯')
plt.colorbar(image, fraction=0.046, pad=0.04)

plt.subplot(2, 3, 4)
plt.plot(time_axis, daily_pattern, color=GRAYS[1], linewidth=1, label='静的背景')
for hour in [3, 8, 13, 18, 23]:
    center = np.argmin(np.abs(time_axis - hour))
    window = slice(max(center - 12, 0), min(center + 13, len(time_axis)))
    plt.plot(time_axis[window], daily_pattern[window], linewidth=3, color=ACCENT)
plt.title('4. 代表フレームだけを強調', fontweight='bold')
plt.xlabel('時刻')
plt.ylabel('処理件数')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(2, 3, 5)
category_colors = dict(zip(['Webログ', '決済', '問い合わせ'], colors_professional[:3]))
for category in ['Webログ', '決済', '問い合わせ']:
    idx = np.where(categories == category)[0]
    step = max(1, len(idx) // 1200)
    sampled_idx = idx[::step][:1200]
    plt.scatter(
        request_count[sampled_idx],
        response_time[sampled_idx],
        s=5,
        alpha=0.35,
        color=category_colors[category],
        label=category,
        rasterized=True,
    )
plt.title('5. カテゴリ別フィルタリングとサンプリング', fontweight='bold')
plt.xlabel('リクエスト数')
plt.ylabel('応答時間（ms）')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(2, 3, 6)
plt.scatter(request_count, response_time, s=3, alpha=0.12, color=GRAYS[0], rasterized=True, label='明細点')
bins = np.linspace(request_count.min(), request_count.max(), 24)
centers = (bins[:-1] + bins[1:]) / 2
mean_response = []
for left, right in zip(bins[:-1], bins[1:]):
    mask = (request_count >= left) & (request_count < right)
    mean_response.append(np.mean(response_time[mask]) if np.any(mask) else np.nan)
plt.plot(centers, mean_response, linewidth=3, color=ACCENT, label='平均線')
plt.title('6. ラスター点とベクター線の混在', fontweight='bold')
plt.xlabel('リクエスト数')
plt.ylabel('応答時間（ms）')
plt.legend()
plt.grid(True, alpha=0.3)

print('静的グラフのパフォーマンス最適化:')
print('- 点群は rasterized=True を使うと、PDF/SVG出力時のファイルサイズを抑えやすくなります。')
print('- 大量の線は LineCollection で一括描画すると効率的です。')
print('- ヒートマップは必要な解像度まで間引いてから描画するとメモリ使用量を抑えられます。')
print('- 注目すべき代表フレームや平均線だけをベクター要素として重ねると、読みやすさを保てます。')

plt.tight_layout()
plt.show()
```
