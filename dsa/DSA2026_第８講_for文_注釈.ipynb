# U. for 文：同じ処理をくり返す

12か月の降水量を1つずつ表示したいとき、`print` を12回書くのは大変。**for 文**を使うと、リストの要素を1つずつ取り出して、同じ処理をくり返せる。

## U-1 for 文の基本

```python
for 変数 in リスト:
    くり返す処理
```

- リストの**先頭から順に**要素を1つ取り出して `変数` に入れ、字下げした部分を実行する。これを最後の要素までくり返す

````python
fruits = ['りんご', 'みかん', 'ぶどう']

for f in fruits:
    print(f)
````

- 1回目は `f` に `'りんご'`、2回目は `'みかん'`、3回目は `'ぶどう'` が入る
- 変数の名前（ここでは `f`）は自由に決めてよい。`k`、`x`、`fruit` など
- 数と文字が混ざったリストでも、同じように取り出せる

````python
mylist = [11, 22, 'abc', 33]

for k in mylist:
    print(k)
````

▶ **演習**：`for文基礎001`

## U-2 コロンと字下げ

- 1行目の最後に **`:`（コロン）** を付ける
- くり返す処理は、行の先頭に**半角スペース4つ**を入れて**字下げ（インデント）**する。Jupyter では `:` の後で改行すると自動で字下げされる
- 字下げした行だけがくり返される。字下げをやめた行は、くり返しが**終わった後に1回だけ**実行される

````python
for f in fruits:
    print(f + 'が好き')
    print('---')
print('おわり')
````

| エラーの最後の行 | 原因 | 直し方 |
|---|---|---|
| `SyntaxError: expected ':'` | 1行目の最後の `:` を忘れた | `:` を付ける |
| `IndentationError: expected an indented block` | くり返す処理を字下げしていない | 行の先頭にスペース4つ |
| `IndentationError: unexpected indent` | 字下げしなくてよい行を字下げした | 余分なスペースを消す |

- 字下げは `Tab` キーでも入る。全角スペースは使わない（見た目では区別できないので注意）

## U-3 くり返しの中で計算する

- 取り出した要素を使って計算できる。`**` は「〜乗」（`3 ** 2` は 3 の2乗）

````python
mylist = [0, 1, 2, 3, 4, 5, 6]

for k in mylist:
    print(k ** 2)
````

▶ **演習**：`for文基礎002`

- Excel から取り出した列も、リストと同じように for で取り出せる

````python
for r in df['降水量']:
    print(r, 'mm')
````

- 合計を数える：くり返しの前に `total = 0` を作り、くり返しの中で足していく

````python
total = 0
for r in df['降水量']:
    total = total + r
print('1年間の降水量', total, 'mm')
print('np.sum でも同じ', np.sum(df['降水量']))
````

## U-4 range：番号でくり返す

- `range(5)` は `0, 1, 2, 3, 4`（**0 から始まり、5 は入らない**）
- `range(1, 13)` は `1, 2, …, 12`（1 から 12 まで）
- 「何番目か」を使いたいときは、`range(len(リスト))` で**番号**を取り出し、`リスト[番号]` で要素を取り出す

````python
for i in range(5):
    print(i)
````

````python
for i in range(len(fruits)):
    print(i, fruits[i])
````

- リストの番号は **0 から**始まる（第２講）。`fruits[0]` が `'りんご'`

## U-5 くり返しでリストを作る

- 空のリスト `[]` を作り、くり返しの中で `.append(値)` で1つずつ足していく（前回のリストの操作）
- 例：降水量を mm から cm に直したリストを作る

````python
rain_cm = []
for r in df['降水量']:
    rain_cm.append(r / 10)
print(rain_cm)
````

## U-6 くり返しでグラフを描く

## V-0 今日の完成形

月ごとの降水量の棒グラフの上に、**値を書き込んだ**もの。棒が12本あるので、`plt.text` を for で12回くり返している。

1. **全部コピーして**自分の notebook のセルに貼り付け、実行する
1. まずは**数値や文字を変えて**、グラフのどこが変わるかを確かめる

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

df = pd.read_excel('dsa2026data.xlsx', sheet_name='降水量')
x = df['月']
y = df['降水量']

plt.figure(figsize=(10, 5))
plt.bar(x, y, color='skyblue', edgecolor='k')
for i in range(len(x)):  # [1] 棒の数だけくり返す
    plt.text(x[i], y[i] + 5, y[i],  # [2] 書く場所と文字
             ha='center', fontsize=10)  # [3] 位置・大きさ
plt.text(7, 400, '7月は梅雨の終わりの大雨', ha='center', color='r', fontsize=12)  # [4] 説明の文字
plt.ylim(0, 450)
plt.xticks(x)
plt.title('新潟市の月ごとの降水量（2024年）', fontsize=20)
plt.xlabel('月', fontsize=14)
plt.ylabel('降水量 (mm)', fontsize=14)
plt.show()
````

## V-1 plt.text の基本

第４講 H-10 で、散布図の点に1つだけ名前を付けた。ここでくわしく学ぶ。

- `plt.text(x, y, '文字')` で、**グラフの目盛りで** (x, y) の位置に文字を書く
- 文字の代わりに数（変数）を渡してもよい

````python
plt.figure(figsize=(8, 5))
plt.plot(df['月'], df['気温'], marker='o')
plt.text(8, 28, '最高 28.0℃')
plt.text(1, 4.3, '最低 4.3℃')
plt.xlabel('月')
plt.ylabel('気温 (℃)')
plt.show()
````

- 点と文字が重なって読みにくい。位置を少しずらす（V-2）

## V-2 文字の大きさ・色・位置

| 書き方 | 意味 |
|---|---|
| `fontsize=12` | 文字の大きさ |
| `color='r'` | 文字の色 |
| `ha='center'` | 横の位置合わせ：`'left'`（指定した点から右へ書く。何も書かないとこれ）、`'center'`（点が文字の中央）、`'right'` |
| `va='bottom'` | 縦の位置合わせ：`'bottom'`（点の上に書く）、`'center'`、`'top'`（点の下に書く） |

````python
plt.figure(figsize=(8, 5))
plt.plot(df['月'], df['気温'], marker='o')
plt.text(8, 28.8, '最高 28.0℃', ha='center', va='bottom', color='r', fontsize=12)
plt.text(1, 3.5, '最低 4.3℃', ha='center', va='top', color='b', fontsize=12)
plt.ylim(0, 33)
plt.xlabel('月')
plt.ylabel('気温 (℃)')
plt.show()
````

### 複数行の注釈

- 第５講 M-3 で紹介した `\n`（改行）を使うと、注釈を2行以上にできる

````python
plt.figure(figsize=(8, 5))
plt.plot(df['月'], df['気温'], marker='o')
plt.text(8, 28.8, '最高気温\n28.0℃（8月）', ha='center', va='bottom', color='r', fontsize=12)
plt.text(1.2, 4, '最低気温\n4.3℃（1月）', ha='left', va='top', color='b', fontsize=12)
plt.ylim(0, 35)
plt.xlabel('月')
plt.ylabel('気温 (℃)')
plt.show()
````

- `'…\n…'` で2行になる。`ha='center'` で各行が中央にそろう（最低気温のように `ha='left'` なら左そろえ）
- `va='bottom'` は文字のかたまり全体の下端を点の位置に合わせる。行が増えると上に伸びるので、`plt.ylim` で上を空ける
- 行の間隔は `linespacing=1.5` などで変えられる

## V-3 棒の上に値を書く

- 棒が12本あれば、`plt.text` を12回書くことになる → **for 文**でくり返す
- `range(len(x))` で番号 `i` を取り出し、`x[i]`（横の位置）、`y[i]`（棒の高さ）を使う

````python
x = df['月']
y = df['降水量']

plt.figure(figsize=(10, 5))
plt.bar(x, y, color='skyblue')
for i in range(len(x)):
    plt.text(x[i], y[i], y[i], ha='center', va='bottom')
plt.xticks(x)
plt.xlabel('月')
plt.ylabel('降水量 (mm)')
plt.show()
````

- `va='bottom'` で、棒のてっぺんのすぐ上に文字が乗る
- 平均点の棒グラフ（U-6）にも、点数を書き込んでみよう。小数が長いときは `round(値, 1)` で小数第1位までにする

````python
plt.figure(figsize=(8, 5))
plt.bar(subjects, means, color='lightgreen')
for i in range(len(subjects)):
    plt.text(i, means[i], round(means[i], 1), ha='center', va='bottom', fontsize=12)
plt.ylim(0, 100)
plt.title('5教科の平均点')
plt.show()
````

- 横軸が文字（`'国語'` など）のとき、棒の位置は左から `0, 1, 2, …` になる。だから `plt.text(i, …)` で棒の上に書ける

## W-1 番号と値を同時に取り出す

- `enumerate(リスト)` を使うと、**番号と値**を一度に取り出せる。`range(len(…))` と同じことを短く書ける
- `zip(リスト1, リスト2)` を使うと、2つのリストから**同じ番号の要素を組にして**取り出せる

````python
for i, f in enumerate(fruits):
    print(i, f)

for m, r in zip(df['月'], df['降水量']):
    print(m, '月', r, 'mm')
````

## W-2 矢印付きの注釈

- `plt.annotate('文字', xy=(指したい点), xytext=(文字の位置), arrowprops={'arrowstyle': '->'})` で、矢印付きの注釈を書ける

````python
plt.figure(figsize=(8, 5))
plt.plot(df['月'], df['降水量'], marker='o')
plt.annotate('7月の大雨', xy=(7, 376.5), xytext=(3, 350), fontsize=12,
             arrowprops={'arrowstyle': '->', 'color': 'r'})
plt.ylim(0, 420)
plt.xlabel('月')
plt.ylabel('降水量 (mm)')
plt.show()
````

# まとめ

| やりたいこと | 書き方 |
|---|---|
| **以下は for 文** | |
| リストの要素を1つずつ取り出す | `for k in リスト:`（`:` を忘れない） |
| くり返す処理 | 行の先頭に**半角スペース4つ**（字下げ） |
| 番号でくり返す | `for i in range(len(リスト)):`、要素は `リスト[i]` |
| `range` | `range(5)` → 0〜4、`range(1, 13)` → 1〜12 |
| リストを作る | `new = []` → くり返しの中で `new.append(値)` |
| 2乗 | `k ** 2` |
| **以下は注釈** | |
| 文字を書く | `plt.text(x, y, '文字')` |
| 位置・大きさ・色 | `ha='center'`、`va='bottom'`、`fontsize=12`、`color='r'` |
| 複数行の注釈 | `plt.text(x, y, '1行目\n2行目')` |
| 棒の上に値を書く | `for i in range(len(x)):` の中で `plt.text(x[i], y[i], y[i], ha='center', va='bottom')` |
