# E. 折れ線グラフを整える

D で描いたグラフに、**見やすくするための指定**を足していく。演習「第３講」はどれも白紙から書く問題なので、ここのコードを写して、データを問題のものに変えればよい。

演習「第３講」の問題で使うデータ：

````python
import matplotlib.pyplot as plt
import japanize_matplotlib

x = [1, 2, 3, 4, 5, 6]
y = [2, 6, 3, 12, 10, 5]

plt.plot(x, y)
plt.show()
````

▶ **演習**：`plot_描画基礎001`（上のセルと同じ。これで1問解ける）

## E-1 マーカーの種類・サイズ・色

- `plt.plot` に**キーワード引数**（C-2）を足す

| 指定 | 書き方 | 例 |
|---|---|---|
| マーカーの種類 | `marker=` | `'o'` 丸、`'s'` 四角、`'^'` 三角、`'*'` 星、`'D'` ひし形 |
| マーカーのサイズ | `markersize=` | `10`、`15`（数値。大きいほど大きい） |
| 色 | `color=` | `'r'` 赤、`'g'` 緑、`'b'` 青、`'orange'`、`'tan'` など色の名前も使える |

````python
plt.plot(x, y, marker='o', markersize=10, color='orange')
plt.show()
````

- 記号をマーカーにすることもできる。`$` で囲んだ部分は数式の書き方（TeX）で解釈される
  - ハート `'$\\heartsuit$'`、スペード `'$\\spadesuit$'`、クラブ `'$\\clubsuit$'`、ダイヤ `'$\\diamondsuit$'`
  - `\` は Windows のキーボードでは `¥` キーで入力する。2つ重ねて書くことに注意

````python
plt.plot(x, y, marker='$\\heartsuit$', markersize=20, color='r')
plt.show()
````

▶ **演習**：`plot_描画基礎002`

## E-2 軸の範囲

- 何も指定しないと、軸の範囲はデータに合わせて自動で決まる
- `plt.xlim(最小, 最大)`、`plt.ylim(最小, 最大)` で範囲を決められる
- 例：y 軸を 0 から始めると、値の大小を正しく比べやすい

````python
plt.plot(x, y, marker='o', markersize=10, color='orange')
plt.xlim(0, 7)
plt.ylim(0, 14)
plt.show()
````

▶ **演習**：`plot_描画基礎003`

## E-3 タイトル・軸の名前とフォントサイズ

- D で使った `plt.title`、`plt.xlabel`、`plt.ylabel` に、`fontsize=` を足すと文字の大きさを変えられる

````python
plt.plot(x, y, marker='o', markersize=10, color='orange')
plt.xlim(0, 7)
plt.ylim(0, 14)
plt.title('折れ線グラフ', fontsize=20)
plt.xlabel('x軸', fontsize=14)
plt.ylabel('y軸', fontsize=14)
plt.show()
````

- 上から順に「描く → 範囲 → 文字」と並べたが、`plt.show()` より前なら順番は入れ替えてもよい

▶ **演習**：`plot_描画基礎004`

## E-4 2本の線を重ねる

- `plt.plot` を**2回書く**と、同じグラフに2本の線が重なる
- `label=` で線に名前を付け、`plt.legend()` で**凡例**（どの線が何か）を表示する
- 色を指定しないと、1本目は青、2本目はオレンジのように自動で色分けされる

````python
y1 = [2, 6, 3, 12, 10, 5]
y2 = [7, 3, 4, 8, 9, 2]

plt.plot(x, y1, marker='o', markersize=10, label='データ1')
plt.plot(x, y2, marker='s', markersize=10, label='データ2')
plt.xlim(0, 7)
plt.ylim(0, 14)
plt.title('2本の折れ線グラフ', fontsize=20)
plt.xlabel('x軸', fontsize=14)
plt.ylabel('y軸', fontsize=14)
plt.legend()
plt.show()
````

▶ **演習**：`plot_描画基礎005`（凡例は問題の条件には無いが、付けると分かりやすい）

## E-5 グラフの大きさ・凡例

関数には、ここまで使ったもの以外にも**いろいろなオプション（キーワード引数）**がある。E-5〜E-7 でよく使うものを紹介する。演習問題は無いので、自分のグラフで試してみよう。

### グラフの大きさ：`plt.figure(figsize=(横, 縦))`

- 単位はインチ（1インチ＝約2.5cm）。何も指定しないと `(6.4, 4.8)`
- **`plt.plot` より前に**書く（「これから描く紙の大きさ」を先に決める）

````python
plt.figure(figsize=(10, 3))
plt.plot(x, y1, marker='o')
plt.show()
````

- 横長にすると、時間の流れ（日ごとの気温など）が見やすくなる

### 凡例の場所：`plt.legend(loc='...')`

- 何も指定しないと `'best'`（線と重なりにくい場所を自動で選ぶ）。それでも線に重なるときは、場所を指定する

| `loc=` | 場所 |
|---|---|
| `'upper right'` | 右上 |
| `'upper left'` | 左上 |
| `'lower right'` | 右下 |
| `'lower left'` | 左下 |
| `'center'` | 中央 |

- 凡例の文字の大きさも `fontsize=` で変えられる（E-3 と同じ）

````python
plt.plot(x, y1, marker='o', label='データ1')
plt.plot(x, y2, marker='s', label='データ2')
plt.legend(loc='upper left', fontsize=14)
plt.show()
````

### 凡例の場所を数値で指定する：`loc=(横, 縦)`

- グラフの枠の**左下を `(0, 0)`、右上を `(1, 1)`** として、凡例の**左下の角**を置く位置を割合で指定する
- 例：`loc=(0.05, 0.75)` は「横は左端から少し右、縦は上から4分の1」のあたり。線と重なる場合は、数値を少しずつ変えて、空いている場所を探す
- `frameon=False` を足すと、凡例の**枠が消える**（すっきり見える）

````python
plt.plot(x, y1, marker='o', label='データ1')
plt.plot(x, y2, marker='s', label='データ2')
plt.legend(loc=(0.05, 0.75), frameon=False)
plt.show()
````

## E-6 線の種類・太さと短い書き方

### 線の種類・太さ：`linestyle=`、`linewidth=`

| 指定 | 書き方 | 例 |
|---|---|---|
| 線の種類 | `linestyle=` | `'-'` 実線、`'--'` 破線、`':'` 点線、`'-.'` 一点鎖線 |
| 線の太さ | `linewidth=` | `1`、`3`（数値。大きいほど太い） |

````python
plt.plot(x, y1, color='r', linestyle='--', linewidth=3, label='データ1')
plt.plot(x, y2, color='g', linestyle=':', linewidth=3, label='データ2')
plt.legend()
plt.show()
````

### 短い書き方：`c=`、`ls=`、`lw=`

- よく使う3つには**短い名前**がある。意味はまったく同じ

| 長い書き方 | 短い書き方 |
|---|---|
| `color='r'` | `c='r'` |
| `linestyle='--'` | `ls='--'` |
| `linewidth=3` | `lw=3` |

- 下のセルは、上のセルと**同じグラフ**になる。インターネットや生成AIのコードには短い書き方もよく出てくるので、読めるようにしておこう

````python
plt.plot(x, y1, c='r', ls='--', lw=3, label='データ1')
plt.plot(x, y2, c='g', ls=':', lw=3, label='データ2')
plt.legend()
plt.show()
````

### マーカー・線・色を1つの文字列で指定する

- `plt.plot(x, y, 'o--r')` のように、x, y の後ろに**文字列を1つ**書くと、マーカー・線の種類・色をまとめて指定できる
- 中身は「マーカー」「線の種類」「色」の記号を並べたもの（順番は入れ替えてもよい。使わないものは省略できる）

| 文字列 | 意味 |
|---|---|
| `'o--r'` | 丸マーカー＋破線＋赤 |
| `'s:g'` | 四角マーカー＋点線＋緑 |
| `'^-b'` | 三角マーカー＋実線＋青 |
| `'o'` | 丸マーカーだけ（線なし。散布図のようになる） |

````python
plt.plot(x, y1, 'o--r', label='データ1')
plt.plot(x, y2, 's:g', label='データ2')
plt.legend()
plt.show()
````

## E-7 目盛り・透明度・補助線

### 目盛りの文字を斜めにする：`plt.xticks(rotation=角度)`

- x 軸の文字が長くて重なるときに使う。角度は度（`20`、`45`、`90` など）
- `fontsize=` で目盛りの文字の大きさも変えられる

````python
city = ['新潟市', '長岡市', '上越市', '三条市', '柏崎市', '新発田市']
rain = [1845, 2212, 2755, 1916, 2117, 1852]

plt.plot(city, rain, marker='o')
plt.xticks(rotation=45, fontsize=12)
plt.ylabel('年間降水量（mm）')
plt.show()
````

- 上のデータは、目盛りの説明用に作った例

### 透明度：`alpha=`

- `0`（完全に透明）〜 `1`（不透明）の数値。重なった部分が透けて見える

````python
plt.plot(x, y1, marker='o', markersize=25, alpha=0.5)
plt.plot(x, y2, marker='o', markersize=25, alpha=0.5)
plt.xlim(0, 7)
plt.ylim(0, 14)
plt.show()
````

### 補助線：`plt.axhline(y=値)`、`plt.axvline(x=値)`

- `axhline` は**横線**（horizontal）、`axvline` は**縦線**（vertical）
- 例：平均値（C-2 の `sum / len`）に横線を引くと、平均より上か下かがひと目で分かる
- `c=`、`ls=`、`lw=` で色・線の種類・太さを指定できる（E-6 と同じ）

### 目盛り線：`plt.grid()`

- 値を読み取りやすくなる。`alpha=` で薄くすると、グラフの邪魔になりにくい

````python
plt.plot(x, y1, marker='o')
plt.axhline(y=sum(y1) / len(y1), c='m', ls=':', lw=2)
plt.axvline(x=4, c='gray', ls='--')
plt.grid(alpha=0.3)
plt.show()
````

# F. 散布図

- 折れ線グラフは点を**線でつなぐ**。散布図は**点だけを打つ**
- 線でつなぐのは、時間の流れのように**順番に意味がある**とき。2つの量の**関係**（身長と体重など）を見たいときは散布図を使う
- 散布図は `plt.scatter(x, y)` で描く。x, y の渡し方は `plt.plot` と同じ

````python
plt.scatter(x, y)
plt.show()
````

▶ **演習**：`scatter_描画基礎001`

- マーカーの種類は `marker=`、色は `color=`（または `c=`）で指定する。`plot` と同じ
- **大きさは `s=`** で指定する（`plot` の `markersize=` とは名前が違うので注意）。`s=` は点の**面積**なので、`markersize=10` と同じくらいにしたいときは `s=100` くらいになる

````python
plt.scatter(x, y, marker='^', color='g', s=100)
plt.show()
````

▶ **演習**：`scatter_描画基礎002`
