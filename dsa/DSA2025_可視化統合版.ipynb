### このテキストのコードの読み方

初めて見ると気になる書き方を、先にまとめておく。

**1. 行頭の `%` （マジックコマンド）**

`%matplotlib inline` `%config ...` は Python の文法ではなく、Jupyter への指示。
グラフをノートブック内に、きれいな解像度で表示するための「おまじない」と思ってよい。

**2. 行末の `;` （セミコロン）**

```python
plt.plot(x, y);     ←  末尾にセミコロン
```

Python では文末のセミコロンは不要だが、**Jupyter では「最後の行の戻り値を表示しない」**
という意味になる。付けないと、グラフの上に

```
[<matplotlib.lines.Line2D at 0x...>]
```

のような余計な行が出る。**付け忘れても、グラフは正しく描ける**ので気にしなくてよい。

**3. `plt.show()`**

セミコロンと同じく、余計な表示を消してグラフを確定させる命令。
`plt.plot(...);` と `plt.plot(...)` ＋ `plt.show()` は、どちらでも結果は同じ。
このテキストでは元の講義資料に合わせて両方の書き方が混在している。

**4. `#` で始まる行**

コメント（実行されない説明文）。行頭に `#` を付けると、その行を一時的に無効にできる。
`#plt.ylim(0,17)` のように**あえて無効にしてある行**は、`#` を外すと効果を確認できる。

## 0.4 データの入れ物：リスト・np.array・pd.Series

グラフに渡すデータには、次の3種類が使える。**この違いは1章以降ずっと出てくる**ので、
先に押さえておくと後で混乱しない。

| 種類 | 作り方 | このテキストでの登場例 |
|---|---|---|
| **リスト** (list) | `[1, 2, 3]` | ベタ書きするデータ全般 |
| **np.array** | `np.array([1, 2, 3])` | 4章の体重データ、8章のスコアデータ |
| **pd.Series** | `df['列名']` | Excelから読んだ列（1章以降ずっと） |

見た目はどれも「数値の並び」だが、**四則演算のふるまいが違う**。ここが最初のつまずきどころ。

````python
# list
temp = [4.3, 4.8, 6.0, 13.7, 17.1, 22.2, 26.2, 28.0, 25.0, 18.7, 11.5, 4.9]
temp
````

````python
# list に * をすると「繰り返し」になってしまう（各要素の1.5倍にはならない）
temp * 2
````

````python
# numpy の関数に渡すと、内部で array に変換されて計算される
np.sqrt(temp)
````

````python
print(np.mean(temp))
print(np.std(temp))
````

````python
# np.array に変換する
temp_array = np.array(temp)
temp_array
````

````python
# array なら、各要素が1.5倍になる
temp_array * 1.5
````

````python
# array 自身が mean() や std() の機能を持つ
print(temp_array.mean())
print(temp_array.std())
````

````python
# np.mean(array) でも array.mean() でも、どちらでもOK
print(np.mean(temp_array))
print(np.std(temp_array))
````

`pd.read_excel()` で読み込んだ dataframe の列（`df['列名']`）は pd.Series 型で、
np.array と同様に各要素への演算ができる（例：`df_accident['交通事故件数']/1000`）。

## 1.2 実データの描画（交通事故統計）

この節から `'s:c'` のような短い文字列が出てくる。これは
**色・マーカーの形・線種を一括指定する省略記法**で、`'s:c'` なら
「マーカーは四角(`s`)、線は点線(`:`)、色はシアン(`c`)」の意味。
`plt.plot(x, y, color='c', marker='s', linestyle=':')` と書いても同じ。
記号の一覧は [付録A](#付録A-fmt-指定について) にまとめてある。

````python
# 最も簡単なグラフ
# x軸に df_accident の「年」列、y軸に df_accident の「交通事故件数」列を指定
plt.plot(df_accident['年'], df_accident['交通事故件数']);
````

````python
# グラフサイズ指定
plt.figure(figsize=(4,3))

# タイトル指定
plt.title('全国交通事故統計', fontsize=16)

# x, y軸名指定
plt.xlabel('年', fontsize=14)
plt.ylabel('件数(千件)', fontsize=14)

# y軸範囲の指定
plt.ylim(0, 1300)

# グリッド線の設定
plt.grid(False)

# データプロット（折れ線グラフ、凡例：ラベル付き）
plt.plot(df_accident['年'], df_accident['交通事故件数']/1000, 's:c', label='交通事故件数')
plt.plot(df_accident['年'], df_accident['負傷者数']/1000, 's:m', label='負傷者数')
plt.legend();
````

**（発展）** 数値範囲が大きく異なる系列を重ねたい場合は、`plt.twinx()` で右軸を追加する。
ここでは「こういうこともできる」とだけ見ておけばよい。詳しくは [7.2 両y軸グラフ](#7.2-両y軸グラフ（twinx）) で扱う。

````python
# グラフサイズ指定
plt.figure(figsize=(6,4))

# タイトル指定
plt.title('全国交通事故統計', fontsize=16)

# x, y軸名指定
plt.xlabel('年', fontsize=14)
plt.ylabel('件数(千件)', fontsize=14)

# y軸範囲の指定
plt.ylim(0, 1300)

# グリッド線の設定
plt.grid(False)

# データプロット（折れ線グラフ）
plt.plot(df_accident['年'], df_accident['交通事故件数']/1000, 's:c', label='交通事故件数')
plt.plot(df_accident['年'], df_accident['負傷者数']/1000, 's:m', label='負傷者数')

# 凡例の枠線を非表示
plt.legend(frameon=False)

# 右軸の設定
plt.twinx()
plt.plot(df_accident['年'], df_accident['死者数']/1000, 's-y', label='死者数')

# y軸範囲の指定
plt.ylim(0, 10)
plt.ylabel('死者数(千件)', fontsize=10)
plt.legend(loc=(0.7,0.75), frameon=False);
````

## 1.3 複数グラフの一括描画（subplot）
`plt.subplot(行数, 列数, 何番目)` で、1枚の図の中に複数のグラフを並べられる。

````python
# 複数グラフ一括描画の例
plt.figure(figsize=(10,6))

plt.subplot(2,2,1)
plt.ylim(0, 1200)
plt.plot(df_accident['年'], df_accident['交通事故件数']/1000, 's:c')
plt.xlabel('年', fontsize=14)
plt.ylabel('交通事故件数(千件)', fontsize=14)

plt.subplot(2,2,2)
plt.ylim(0, 1300)
plt.plot(df_accident['年'], df_accident['負傷者数']/1000, 's:m')
plt.xlabel('年', fontsize=14)
plt.ylabel('負傷者数(千件)', fontsize=14)

plt.subplot(2,2,3)
plt.ylim(0, 10)
plt.plot(df_accident['年'], df_accident['死者数']/1000, 's:y')
plt.xlabel('年', fontsize=14)
plt.ylabel('死者数(千件)', fontsize=14)

plt.subplot(2,2,4)
plt.ylim(0, 100)
plt.plot(df_accident['年'], df_accident['重傷者数']/1000, 's:b')
plt.xlabel('年', fontsize=14)
plt.ylabel('重傷者数(千件)', fontsize=14)

# グラフ同士の重なりを自動調整
plt.tight_layout();
````

## 1.4 グラフを保存して PowerPoint・Word で使う

グラフを描く最終的な目的は、たいていレポートやスライドに載せることである。

### 基本形と、最大の落とし穴

`plt.savefig('ファイル名')` で保存する。ファイルはこの `.ipynb` と同じフォルダにできる。

**`plt.savefig()` は必ず `plt.show()` より前に書くこと。** 後ろに書くと、
中身が空の**真っ白なファイル**が保存される（グラフは画面に出ているのに、ファイルは白紙）。
これは非常によくある失敗で、原因に気づきにくい。

```python
plt.plot(x, y)
plt.savefig('graph.png')   # ← 先に保存
plt.show()                 # ← 後に表示
```

````python
# 保存の基本（このセルを実行すると graph_test.png ができる）
plt.figure(figsize=(4,3))
plt.plot(df_accident['年'], df_accident['交通事故件数']/1000, 's:c')
plt.title('全国交通事故統計', fontsize=14)
plt.xlabel('年', fontsize=12)
plt.ylabel('件数(千件)', fontsize=12)

# bbox_inches='tight' で余白を切り詰める（軸ラベルの見切れ防止）
plt.savefig('graph_test.png', bbox_inches='tight', dpi=200)

plt.show()
````

### 主なオプション

| オプション | 意味 | 目安 |
|---|---|---|
| `bbox_inches='tight'` | 余白を切り詰める。**ラベルの見切れ防止に必須** | 常に付けてよい |
| `dpi=200` | 解像度（png など画像形式でのみ有効） | スライド用 150〜200、印刷用 300〜360 |
| `transparent=True` | 背景を透明にする | 色付きスライドに貼るとき |

### 背景の透明化（`transparent=True`）

グラフを普通に保存すると、背景は**白く塗りつぶされている**。これを色付きのスライドや
写真の上に貼ると、グラフのまわりに**白い四角が残って浮いてしまう**。

`plt.savefig('graph.png', transparent=True)` とすると背景が透明になり、
スライドの背景色や写真がグラフの後ろに透けて、自然に馴染む。

**何が透明になるか**

図全体の背景・グラフ領域の背景・凡例の背景が、まとめて透明になる。
**残るのは線・点・文字だけ**。グラフの中身が消えることはない。

**使うときの注意**

1. **jpg では透明にできない。** jpg は透明を扱えない形式なので、`transparent=True`
   を付けても無視され、白い背景のまま保存される。透明にしたいときは **png**（または svg）にする。
2. **暗い背景のスライドでは、文字と線が見えなくなる。** 既定では文字・軸・目盛が
   黒なので、透明にして濃い色のスライドに置くと読めない。
   その場合は [11.1 スタイルファイルの適用](#11.1-スタイルファイルの適用) の
   `dark_background` スタイルで描くと、文字と線が白くなって読めるようになる。
3. **画面上では白背景と見分けがつかない。** 保存したpngをビューアで開いても、
   多くのビューアは透明部分を白く表示する。実際にスライドに貼って確かめるのが確実。

````python
# 背景を透明にして保存（色付きスライドに貼るとき用）
plt.figure(figsize=(4,3))
plt.plot(df_accident['年'], df_accident['交通事故件数']/1000, 's:c')
plt.title('全国交通事故統計', fontsize=14)
plt.xlabel('年', fontsize=12)
plt.ylabel('件数(千件)', fontsize=12)

# transparent=True で背景が透明な png になる
plt.savefig('graph_transparent.png', bbox_inches='tight', dpi=200, transparent=True)

plt.show()
````

````python
# 暗い背景のスライドに貼る場合：文字と線を白くしてから透明化する
with plt.style.context('dark_background'):
    plt.figure(figsize=(4,3))
    plt.plot(df_accident['年'], df_accident['交通事故件数']/1000, 's:c')
    plt.title('全国交通事故統計', fontsize=14)
    plt.xlabel('年', fontsize=12)
    plt.ylabel('件数(千件)', fontsize=12)

    plt.savefig('graph_transparent_dark.png', bbox_inches='tight', dpi=200, transparent=True)

    plt.show()
````

### ファイル形式の選び方

**拡張子を変えるだけ**で形式が変わる。大きく「ビットマップ」と「ベクター」に分かれる。

| 形式 | 種類 | 特徴 | おすすめの用途 |
|---|---|---|---|
| **`.png`** | ビットマップ | 手軽。どこにでも貼れる | まずはこれ。迷ったらpng |
| **`.svg`** | ベクター | 拡大しても劣化しない。**Officeで図形として編集できる** | スライド、ポスター、作り込む図 |
| `.pdf` | ベクター | 印刷・論文用。LaTeXと相性がよい | 論文、配布資料 |
| `.jpg` | ビットマップ | 写真向け。**グラフには不向き**（線や文字がにじむ） | 使わない |

- **ビットマップ**（png/jpg）は色の点の集まり。拡大するとギザギザになる。
- **ベクター**（svg/pdf）は線や文字の「設計図」。何倍に拡大しても輪郭がなめらか。

````python
# 同じグラフを png と svg の両方で保存して比べてみる
plt.figure(figsize=(4,3))
plt.plot(df_accident['年'], df_accident['交通事故件数']/1000, 's:c')
plt.title('全国交通事故統計', fontsize=14)
plt.xlabel('年', fontsize=12)
plt.ylabel('件数(千件)', fontsize=12)

plt.savefig('graph_test.png', bbox_inches='tight', dpi=200)   # ビットマップ
plt.savefig('graph_test.svg', bbox_inches='tight')            # ベクター

plt.show()

# できたファイルのサイズを確認
import os
for f in ['graph_test.png', 'graph_test.svg']:
    print(f'{f:20} {os.path.getsize(f):>8,} バイト')
````

### SVGの文字は「文字」ではない（重要）

matplotlib は既定では、SVGに保存するとき**文字を図形（パス）に変換**する。
しかも**1文字ずつ別々の図形**になる。

つまり「全国交通事故統計」というタイトルは、Illustratorで開くと
**8個のバラバラな図形**になっており、文字として打ち直すことはできない。

- **利点**：そのフォントが入っていないPCで開いても、必ず同じ見た目になる
- **欠点**：文字の修正ができない（色や大きさは変えられる）

**文字も打ち直せるようにして保存するには**、次の設定をしてから保存する。

```python
plt.rcParams['svg.fonttype'] = 'none'   # 文字を「文字」のまま保存する
```

こうすると Illustrator で文字を選んで修正できる。ただし、開いたPCに同じフォント
（`japanize_matplotlib` が使う IPAexGothic）が入っていないと、別のフォントに
置き換わって表示される。

**配るだけ・見せるだけなら、既定のまま（文字は図形）で問題ない。**
むしろ既定の方が、どのPCで開いても見た目が変わらないという利点がある。

````python
# SVG内部で文字がどう保存されるかを確かめる
import re

def check_svg(filename):
    s = open(filename, encoding='utf-8').read()
    n_text = len(re.findall(r'<text', s))
    n_path = len(re.findall(r'<path', s))
    print(f'{filename:22} 文字要素<text>={n_text:>3} 個 / 図形要素<path>={n_path:>3} 個')

# 既定（文字は図形に変換される）
plt.figure(figsize=(4,3))
plt.plot([1,2,3],[2,1,3], marker='o')
plt.title('日本語タイトル')
plt.savefig('svg_default.svg', bbox_inches='tight')
plt.close()

# 文字を文字のまま保存する設定
with plt.rc_context({'svg.fonttype': 'none'}):
    plt.figure(figsize=(4,3))
    plt.plot([1,2,3],[2,1,3], marker='o')
    plt.title('日本語タイトル')
    plt.savefig('svg_editable.svg', bbox_inches='tight')
    plt.close()

check_svg('svg_default.svg')    # <text> は 0 個 ＝ 文字が図形化されている
check_svg('svg_editable.svg')   # <text> があり、文字として編集できる
````

# 4. ヒストグラム
https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html

**棒グラフとの違い**：棒グラフは「項目ごとの値」、ヒストグラムは「一群のデータの分布」を表す。
そのため、ヒストグラムは**データを1組渡すだけ**で描ける。

## 4.1 度数分布表とヒストグラム

# 8. エラーバー

**（注意！）** エラーバーは、元の分布が正規分布でよく表されることを暗に仮定している。

## 8.1 前提：二項分布と正規分布

エラーバーの長さには、ふつう**標準偏差**を使う。なぜそれで「ばらつき」を表せるのかを
理解するために、まず正規分布の形を見ておく。

**統計の話が続くので、グラフの描き方だけ知りたい場合は [8.3 errorbar の基本](#8.3-errorbar-の基本) へ進んでよい。**

````python
# 二項分布・正規分布を扱うライブラリをインポート
from scipy.stats import norm, binom

n = 50    # 試行回数
p = [0.05, 0.1, 0.25, 0.5, 0.7]  # 成功確率(list)
x = np.arange(0, n+1)            # 成功回数(list)

# 複数のパラメータ(成功確率 p) を持つ二項分布を繰り返し作成
for k in p:
    y = binom.pmf(x, n, k)
    plt.bar(x, y, alpha=0.6, label=f'p={k}')

# 必要な情報追加
plt.title('二項分布シミュレーション', fontsize=18)
plt.xlabel('x', fontsize=16)
plt.ylabel('f(x)', fontsize=16)
plt.text(0.86, 0.5, f'n={n}', fontsize=14, transform=plt.gca().transAxes)
plt.grid(color='gray', linestyle=':')
plt.legend()

plt.show()
````

````python
# ----- you can set values below -----
n = 30      # 試行回数
p = 0.4     # 成功確率 valid range: [0, 1]
# ------------------------------------

mu = n*p                   # 期待値
std = np.sqrt(n*p*(1-p))   # 標準偏差

# 棒グラフ用データ
x = np.arange(0, n+1)  # 成功回数(list)
y = binom.pmf(x, n, p)

# 棒グラフの描画
plt.figure(figsize=(5,3))
plt.bar(x, y, color='c')  # 二項分布

# 折れ線グラフの描画
x_norm = np.linspace(0, n, 128)
plt.plot(x_norm, norm.pdf(x_norm, loc=mu, scale=std))  # 正規分布

plt.title('二項分布と正規分布の比較', fontsize=18)
plt.xlabel('確率変数(x)', fontsize=16)
plt.ylabel('f(x)', fontsize=16)
print('平均: ', mu, '  標準偏差: ', std)

plt.show()
````

#### 正規分布と、ある区間に入る確率

````python
# 正規分布のパラメータ設定
center = 0
sigma = 1
z_pos = 1

# 正規分布データ作成
x1 = np.linspace(center - 4*sigma, center + 4*sigma, 2**8)
y1 = norm.pdf(x1, loc=center, scale=sigma)
x2 = np.linspace(center - z_pos*sigma, center + z_pos*sigma, 2**8)
y2 = norm.pdf(x2, loc=center, scale=sigma)

# 折れ線グラフ作成
plt.figure(figsize=(5,3))
plt.plot(x1, y1, color='c')
plt.plot(x2, y2, color='c')
plt.fill_between(x2, 0, y2, facecolor='c', alpha=0.3)

# 垂直補助線（ある点までの補助線）
plt.vlines(x=-z_pos, ymin=-0, ymax=norm.pdf(z_pos, loc=center, scale=sigma),
           linestyles='dashed', linewidths=1)
plt.vlines(x=z_pos, ymin=-0, ymax=norm.pdf(z_pos, loc=center, scale=sigma),
           linestyles='dashed', linewidths=1)

# ラベル情報の追加
plt.xlabel('sigma')
plt.ylabel('Probability')

# 色塗り範囲の確率計算
prob = 2*norm.cdf(x=z_pos, loc=center, scale=sigma) - 1
print(f'区間 [{-z_pos: .1f}*sigma,{z_pos: .1f}*sigma]に入る確率:{prob: .3f}')

plt.show()
````

## 8.2 エラーバーとは：円周率を測る世界線
初めて人類が円周率を求める世界線を考える。
例えば、直径1mの円周の長さを紐で何度も繰り返し測ってヒストグラムを作成し、円周率の真値を追い込む。

````python
# 正規分布を扱うライブラリをインポート（8.1節で読み込み済み。8.2節から始める場合はここで必要）
from scipy.stats import norm

# 固定した乱数の生成
rng = np.random.default_rng(348309421)   # seedの設定（空も可）
pi_data = rng.normal(np.pi, 0.10, 1000)  # 平均 3.1415...，標準偏差 0.10 の正規分布の乱数 1000個の配列
pi_data
````

````python
mu = np.pi
std = 0.1

# 固定した乱数の生成
rng = np.random.default_rng(348309421)
pi_data = rng.normal(np.pi, std, 1000)  # 正規分布の乱数1000個の配列

plt.figure(figsize=(5,3))

# ヒストグラム作成
plt.hist(pi_data, bins=20)

# 折れ線グラフの描画
x_norm = np.linspace(np.pi-1/2, np.pi+1/2, 128)
plt.plot(x_norm, norm.pdf(x_norm, loc=mu, scale=std)*38)  # 正規分布

# （知らないはずの）真値に補助線を引く
plt.axvline(x=np.pi, c='y', ls=':')

plt.title('測定値の分布', fontsize=18)
plt.xlabel('測定値', fontsize=16)
plt.ylabel('回数', fontsize=16)
print('平均: ', mu, '  標準偏差: ', std)

plt.show()
````

測定精度（標準偏差）が上がると、エラーバーが短くなり真値が絞り込まれていく様子を並べて示す。

````python
# 複数グラフ全体の描画範囲を指定
plt.figure(figsize=(10,3))

# 1x2 （縦1、横2）のグラフの 1番目のグラフ作成
plt.subplot(1, 2, 1)

mu = np.pi
std = 0.1

rng = np.random.default_rng(348309421)
data = rng.normal(np.pi, std, 1000)

plt.hist(data, bins=20)

x_norm = np.linspace(np.pi-1/2, np.pi+1/2, 128)
plt.plot(x_norm, norm.pdf(x_norm, loc=mu, scale=std)*38)
plt.axvline(x=np.pi, c='y', ls=':')

# エラーバー設定
plt.errorbar(data.mean(), 20, xerr=data.std(), capsize=5, fmt='wo', markersize=6, ecolor='w', markeredgecolor='g')

plt.title('円周率の真値を絞り込む(1)', fontsize=18)
plt.xlabel('測定値', fontsize=16)
plt.ylabel('回数', fontsize=16)
print('平均: ', mu, '  標準偏差: ', std)

# 1x2 （縦1、横2）のグラフの 2番目のグラフ作成
plt.subplot(1, 2, 2)

mu = np.pi
std = 0.02

rng = np.random.default_rng(348309421)
data = rng.normal(np.pi, std, 1000)

plt.hist(data, bins=20)

x_norm = np.linspace(np.pi-1/2, np.pi+1/2, 128)
plt.plot(x_norm, norm.pdf(x_norm, loc=mu, scale=std)*8)
plt.axvline(x=np.pi, c='y', ls=':')

# エラーバー設定
plt.errorbar(data.mean(), 20, xerr=data.std(), capsize=5, fmt='wo', markersize=6, ecolor='w', markeredgecolor='g')

plt.title('円周率の真値を絞り込む(2)', fontsize=18)
plt.xlabel('測定値', fontsize=16)
plt.ylabel('回数', fontsize=16)
print('平均: ', mu, '  標準偏差: ', std)

plt.tight_layout()
plt.show()
````

datetimeモジュールの4つのクラス：

1. `datetime.timedelta`：2つのdatetimeオブジェクト間の差を表すクラス。時間の長さや期間を計算するのに用いる。

#### 書式コード ([document](https://docs.python.org/ja/3/library/datetime.html#strftime-and-strptime-format-codes))

主なもの
- `%d`: 0埋めした10進数で表記した月中の日にち
- `%m`: 0埋めした10進数で表記した月
- `%y`: 0埋めした10進数で表記した西暦の下2桁
- `%Y`: 0埋めした10進数で表記した西暦4桁
- `%H`: 0埋めした10進数で表記した時 （24時間表記）
- `%I`: 0埋めした10進数で表記した時 （12時間表記）
- `%M`: 0埋めした10進数で表記した分
- `%S`: 0埋めした10進数で表記した秒
- `%f`: 0埋めした10進数で表記したマイクロ秒（6桁）
- `%A`: ロケールの曜日名
- `%a`: ロケールの曜日名（短縮形）
- `%B`: ロケールの月名
- `%b`: ロケールの月名（短縮形）
- `%j`: 0埋めした10進数で表記した年中の日にち（正月が'001'）
- `%U`: 0埋めした10進数で表記した年中の週番号 （週の始まりは日曜日）
- `%W`: 0埋めした10進数で表記した年中の週番号 （週の始まりは月曜日）

## 10.5 データの標準化
各変数の単位（分）（年）（m$^2$）がバラバラのままでは、係数の大小を比べても意味がない。
各変数を平均0・標準偏差1に変換（標準化）してから重回帰することで、影響の強さを比較できる。

[回帰分析と標準化](https://analyze.kyoto/blog/%E5%9B%9E%E5%B8%B0%E5%88%86%E6%9E%90%E3%81%A8%E6%A8%99%E6%BA%96%E5%8C%96/)

````python
# 標準正規分布を作成するのと同じ手順
def z_transform(x):
    mu = x.mean()
    sigma = x.std(ddof=0)
    return (x - mu) / sigma
````

````python
df_rent['家賃'] = z_transform(df_rent['家賃'])
df_rent['駅徒歩'] = z_transform(df_rent['駅徒歩'])
df_rent['築年数'] = z_transform(df_rent['築年数'])
df_rent['面積'] = z_transform(df_rent['面積'])
df_rent['階'] = z_transform(df_rent['階'])
df_rent.head()
````

````python
# 重回帰（標準化後）
res = ols('家賃 ~ 駅徒歩 + 築年数 + 面積 + 階', data=df_rent).fit()
res.summary()
````

## 10.6 切片の設定について
モデル式に `0 +` を付けると、切片なし（原点を強制的に通る）回帰になる。

````python
np.random.seed(100)
x = np.arange(-10, 11)

# 直線 y = 3x + 5 に乱数でノイズを加える
noise = np.random.normal(0, 1, 21)
y = 3*x + (5 + noise*10)

# データを dataframe に格納
df_line = pd.DataFrame({'xdata': x, 'ydata': y})
df_line.head()
````

````python
plt.figure(figsize=(4,3))
plt.scatter(df_line['xdata'], df_line['ydata'])
plt.show()
````

````python
# 単回帰（線形回帰：切片あり）
res = ols('ydata ~ xdata', data=df_line).fit()
res.summary()
````

````python
a = res.params['xdata']
b = res.params['Intercept']
a, b
````

````python
# 直線は両端の2点だけ与えれば描ける
x_line = np.array([-10, 10])
y_line = (lambda x: a*x + b)(x_line)
y_line
````

````python
# 散布図の作成
plt.figure(figsize=(4,3))
plt.scatter(df_line['xdata'], df_line['ydata'])

# 傾きと切片の値を使った回帰直線描画
plt.plot(x_line, y_line, 'r-')

plt.xlabel('xdata', fontsize=12)
plt.ylabel('ydata', fontsize=12)
plt.grid()
plt.show()
````

````python
# 単回帰（線形回帰：原点強制通過）
res = ols('ydata ~ 0 + xdata', data=df_line).fit()
res.summary()
````

````python
a = res.params['xdata']

x_line = np.array([-10, 10])
y_line = (lambda x: a*x)(x_line)
y_line
````

````python
# 散布図の作成
plt.figure(figsize=(4,3))
plt.scatter(df_line['xdata'], df_line['ydata'])

# 傾きの値を使った回帰直線描画（原点を通る）
plt.plot(x_line, y_line, 'r-')

plt.xlabel('xdata', fontsize=12)
plt.ylabel('ydata', fontsize=12)
plt.grid()
plt.show()
````

## 12.3 画像をマーカーにした散布図
参考サイト：https://sabopy.com/py/matplotlib-picplot/

（事前準備）マーカーとして使う画像ファイル（この例では `sweets_montblanc.png`）を、
このノートブックと同じフォルダに入れておく。

````python
# このハコ部分は、この授業の範囲内では、こういうもんか、で構いません。そのまま使ってください。
from matplotlib.offsetbox import OffsetImage, AnnotationBbox

def imscatter(x, y, image_path, ax=None, zoom=1):
    """
    画像をマーカーとしてプロットする関数

    Parameters
    ----------
    x, y : array_like
        プロット座標
    image_path : str
        画像ファイルのパス
    ax : matplotlib.axes.Axes, optional
        プロット先の軸
    zoom : float, optional
        画像の拡大率

    Returns
    -------
    artists : list
        プロットした画像のリスト
    """

    # 画像の読み込み
    try:
        image = plt.imread(image_path)
    except FileNotFoundError:
        print(f"画像ファイル '{image_path}' が見つかりません。")
        return []

    # OffsetImageオブジェクトの作成
    im = OffsetImage(image, zoom=zoom)

    # 入力座標の確認（1次元以上のarrayに変換）
    x, y = np.atleast_1d(x, y)

    # 各点に画像をプロット
    artists = []
    for x0, y0 in zip(x, y):
        ab = AnnotationBbox(im, (x0, y0), xycoords='data', frameon=False)
        artists.append(ax.add_artist(ab))

    # 軸の調整
    ax.update_datalim(np.column_stack([x, y]))
    ax.autoscale()
    return artists


def imscatter_sized(x, y, z, image_path, ax=None, zoom_factor=0.2):
    """zの値に応じて画像サイズを変更するプロット関数"""

    image = plt.imread(image_path)

    # 正規化してサイズを調整
    z_norm = (z - np.min(z)) / (np.max(z) - np.min(z)) * 0.8 + 0.2

    artists = []
    for x0, y0, z0 in zip(x, y, z_norm):
        # z値に基づいてズームレベルを変更
        im = OffsetImage(image, zoom=z0 * zoom_factor)
        ab = AnnotationBbox(im, (x0, y0), xycoords='data', frameon=False)
        artists.append(ax.add_artist(ab))

    ax.update_datalim(np.column_stack([x, y]))
    ax.autoscale()
    return artists
````

````python
# グラフサイズの指定
plt.figure(figsize=(5, 4))

# データ
x = [0.37454012, 0.95071431, 0.73199394, 0.59865848, 0.15601864, 0.15599452,
     0.05808361, 0.86617615, 0.60111501, 0.70807258]
y = [0.02058449, 0.96990985, 0.83244264, 0.21233911, 0.18182497, 0.18340451,
     0.30424224, 0.52475643, 0.43194502, 0.29122914]

# プロットに用いる画像のファイル名指定
image_path = 'sweets_montblanc.png'

# 指定した画像をマーカーとして描画するための関数（上で定義したもの）呼び出し
imscatter(x, y, image_path, ax=plt.gca(), zoom=0.1)

# 参考のため点も表示（透明度 alpha=0で非表示）
plt.plot(x, y, 'ko', alpha=0)

# グラフの見た目を調整
plt.title('モンブランでプロットした散布図', fontsize=16)
plt.xlabel('X軸', fontsize=12)
plt.ylabel('Y軸', fontsize=12)
plt.grid(alpha=0.3)
plt.show()
````

| エラー表示 | 原因 | 対処 |
|---|---|---|
