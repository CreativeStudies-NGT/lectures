## N-1 棒グラフの基本

- `plt.plot(x, y)` を `plt.bar(x, y)` に変えるだけ。x が**横軸の項目**、y が**棒の高さ**
- 横軸の項目は、数でも**文字**（`'Thu'` など）でもよい

````python
x = ['Thu', 'Fri', 'Sat', 'Sun']
y = [170, 120, 250, 190]

plt.figure(figsize=(8, 5))
plt.bar(x, y)
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

### 折れ線グラフ・散布図との違い

| | 折れ線グラフ | 散布図 | 棒グラフ |
|---|---|---|---|
| 関数 | `plt.plot(x, y)` | `plt.scatter(x, y)` | `plt.bar(x, y)` |
| 向いているもの | 時間の流れ（順番に意味がある） | 2つの量の関係 | **項目ごとの量の比べ合い** |
| 縦軸 | 0 から始めなくてよい | 0 から始めなくてよい | **0 から始める**（棒の長さで比べるため） |

## N-2 棒の色

- 折れ線グラフ・散布図と同じく `color=`（短く `c=` は棒グラフでは使えないので `color=` と書く）
- 全部の棒が同じ色になる

````python
plt.figure(figsize=(8, 5))
plt.bar(x, y, color='orange')
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

- 前々回の散布図と同じく、`color=['r', 'g', 'b', 'gray']` のように**リスト**で書くと、棒ごとに色を変えられる

▶ **演習**：`bar_描画基礎001`

## N-3 棒の幅

- `width=` で棒の幅を変える。何も書かないと `0.8`。`1` にすると、すき間がなくなる

````python
plt.figure(figsize=(8, 5))
plt.bar(x, y, width=0.4)
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

▶ **演習**：`bar_描画基礎002`

## N-4 外枠線と網掛け

- `edgecolor=` で棒の**外枠線**の色、`linewidth=` でその太さ
- `hatch=` で**網掛け**（模様）。白黒で印刷しても区別できる

| 書き方 | 模様 | 書き方 | 模様 |
|---|---|---|---|
| `'/'` | 斜線 | `'x'` | バツ |
| `'\\'` | 逆向きの斜線 | `'.'` | 点 |
| `'-'` | 横線 | `'o'` | 丸 |
| `'+'` | 格子 | `'*'` | 星 |

- 同じ記号を重ねる（`'//'`、`'///'`）と、模様が細かくなる

````python
plt.figure(figsize=(8, 5))
plt.bar(x, y, color='w', edgecolor='k', linewidth=2, hatch='//')
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

▶ **演習**：`bar_描画基礎003`

## N-5 データを増やす

- 項目を増やすときは、x と y の**両方に**同じ数だけ足す（数が違うとエラー：`ValueError: shape mismatch`）
- 個数は `len(x)` と `len(y)` で確かめられる（第２講）

````python
x7 = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
y7 = [140, 110, 130, 170, 120, 250, 190]
print(len(x7), len(y7))

plt.figure(figsize=(8, 5))
plt.bar(x7, y7)
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

▶ **演習**：`bar_描画基礎004`（データは自分で決めてよい）

## N-6 平均の補助線

- 前回の復習。平均は `np.mean(リスト)`、横の補助線は `plt.axhline(y=値)`
- 平均の値を**自分で計算して書き込まない**。`np.mean(y7)` と書けば、データを変えても線が自動で動く

````python
plt.figure(figsize=(8, 5))
plt.bar(x7, y7)
plt.axhline(y=np.mean(y7), color='r', linestyle='--', label='平均')
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.legend()
plt.show()
````

▶ **演習**：`bar_描画基礎005`

## N-7 横棒グラフ

- `plt.bar` を `plt.barh` に変えると、棒が**横向き**になる（h は horizontal＝水平）
- 横向きになると、項目は**縦軸**、量は**横軸**に移る。だから `xlabel` と `ylabel` の中身を**入れ替える**
- 項目の名前が長いとき（都市名など）は、横棒グラフのほうが読みやすい

````python
plt.figure(figsize=(8, 5))
plt.barh(x7, y7)
plt.title('Bar Chart')
plt.xlabel('Total Bill')
plt.ylabel('Day')
plt.show()
````

▶ **演習**：`bar_描画基礎006`

- 横棒グラフでは、量が横軸にあるので、平均の補助線は**縦の線** `plt.axvline(x=値)` になる

````python
plt.figure(figsize=(8, 5))
plt.barh(x7, y7)
plt.axvline(x=np.mean(y7), color='r', linestyle='--', label='平均')
plt.title('Bar Chart')
plt.xlabel('Total Bill')
plt.ylabel('Day')
plt.legend()
plt.show()
````

▶ **演習**：`bar_描画基礎007`

# O. 円グラフ

**円グラフ**は、**全体に対する割合**（アンケートの回答の割合、予算の内訳など）を見せるグラフ。演習「円グラフの描画」のデータは次の形。

```python
x = [100, 200, 300, 400, 500]
plt.pie(x)
plt.show()
```

## O-0 今日の完成形

よく使う端末のアンケート結果（架空）。12時の位置から時計回りに、大きい順に並べている。

1. **全部コピーして**自分の notebook のセルに貼り付け、実行する
1. まずは**数値や文字を変えて**、グラフのどこが変わるかを確かめる
1. そのあと部品（[1]〜[7]）の働きを1つずつ学ぶ（O-1〜O-6）

````python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib

values = [62, 25, 9, 4]
labels = ['スマホ', 'PC', 'タブレット', 'その他']
colors = ['lightcoral', 'gold', 'lightgreen', 'lightgray']
shift = [0.1, 0, 0, 0]

plt.figure(figsize=(6, 6))  # [0] グラフの大きさ
plt.pie(values,  # [1] 円グラフ
        startangle=90, counterclock=False,  # [2] 12時から時計回り
        labels=labels,  # [3] ラベル
        colors=colors,  # [4] 色
        explode=shift,  # [5] 切り出し
        autopct='%.1f%%', pctdistance=0.7,  # [6] 割合の表示
        wedgeprops={'edgecolor': 'w'})  # 境目を白い線に
plt.title('よく使う端末', fontsize=20)  # [7] タイトル
plt.show()
````

上の `plt.pie(...)` は1つの命令なので、1行に並べて書くと次のようになる。

```python
plt.pie(values, startangle=90, counterclock=False, labels=labels, colors=colors, explode=shift, autopct='%.1f%%', pctdistance=0.7, wedgeprops={'edgecolor': 'w'})
```

- これでも同じグラフになるが、長くて読みにくい。そこで、見やすさのために `,` の後ろで**改行**してよい（`(` の中なら、次の行も続きとして読まれる）

## O-1 円グラフの基本と始まりの位置

- `plt.pie(値のリスト)` で描く。割合は**自動で計算**される（合計が 100 でなくてよい）
- 何も指定しないと、**3時の位置**から**反時計回り**に並ぶ

````python
x = [100, 200, 300, 400, 500]

plt.pie(x)
plt.show()
````

- `startangle=90` で**12時の位置**から始まる（角度は3時の位置から反時計回りに数える。90度＝12時）
- `counterclock=False` で**時計回り**になる（counter clock＝反時計回り、を False＝しない）
- 時計と同じ「12時から時計回り」にすると読みやすい

````python
plt.pie(x, startangle=90, counterclock=False)
plt.show()
````

▶ **演習**：`pie_描画基礎001`

## O-2 ラベル

- `labels=リスト` で、それぞれの部分に名前を付ける。値と**同じ個数・同じ順番**で書く

````python
label = ['Japanese', 'mathematics', 'science', 'history', 'English']

plt.pie(x, startangle=90, counterclock=False, labels=label)
plt.show()
````

- 日本語のラベルも使える（`import japanize_matplotlib` を実行してあれば）
- 個数が違うと `ValueError` になる

▶ **演習**：`pie_描画基礎002`

## O-3 色

- `colors=リスト` で、部分ごとに色を決める。棒グラフの `color=` と違い、**s が付く**ことに注意
- 色の名前は第３講の色の表（`'lightcoral'`、`'gold'` など）を使う

````python
mylabel = ['a', 'b', 'c', 'd', 'e']
mycolor = ['lightcoral', 'lightpink', 'gold', 'lightgreen', 'lightblue']

plt.pie(x, startangle=90, counterclock=False, labels=mylabel, colors=mycolor)
plt.show()
````

▶ **演習**：`pie_描画基礎003`

## O-4 一部を切り出す

- `explode=リスト` で、部分ごとに中心から**どれだけ離すか**を決める。`0` は離さない
- 目立たせたい部分だけ `0.1` くらいにする

````python
mylabel = ['Japanese', 'mathematics', 'science', 'society', 'English']
shift = [0, 0.1, 0, 0, 0]

plt.pie(x, startangle=90, counterclock=False, labels=mylabel, explode=shift)
plt.show()
````

▶ **演習**：`pie_描画基礎004`

## O-5 割合を表示する

- `autopct='%.1f%%'` で、各部分に**割合（%）**を書く。`.1f` は小数第1位まで、`.2f` は第2位まで
- 最後の `%%` は「% の記号を表示する」という意味（`%` を1つだけ書くとエラー）
- `pctdistance=` で、割合の文字を**中心からどれだけ離すか**を決める（`0` が中心、`1` が円の縁。何も書かないと `0.6`）

````python
plt.pie(x, startangle=90, counterclock=False, labels=mylabel, autopct='%.2f%%', pctdistance=0.7)
plt.show()
````

- 演習のひな形の解答例では `autopct="%1.2f%%"` と書かれている。`%1.2f` と `%.2f` は同じ表示になる

▶ **演習**：`pie_描画基礎005`

## O-6 灰色だけで描く

- 白黒で印刷する資料に使う

````python
grayscales = ['0.75', '0.65', '0.55', '0.45', '0.35']

plt.pie(x, startangle=90, counterclock=False, labels=mylabel, autopct='%.1f%%', colors=grayscales)
plt.show()
````

- `'0.75'` を `0.75`（引用符なし）にするとエラーになる。**文字列で**書くこと

````python
df_tip = pd.read_excel('dsa2026data.xlsx', sheet_name='tip_data')
count = df_tip['day'].value_counts()
count
````

### 円グラフを使うときの注意

| 注意 | 理由 |
|---|---|
| 項目は5つくらいまで | 多いと細くて読めない。少ないものは「その他」にまとめる |
| 合計に意味があるときだけ使う | 「全体のうち何 %」でないもの（月ごとの気温など）は棒グラフや折れ線グラフにする |
| 立体（3D）にしない | 手前の部分が大きく見えてしまう（第12講） |
| 2つの円グラフを並べて比べない | 角度の違いは読み取りにくい。比べるなら棒グラフ |

## P-2 カラーマップで色をまとめて決める

- 色を1つずつ選ぶ代わりに、**カラーマップ**（決まった色の組み合わせ）から取り出せる
- `plt.get_cmap('Pastel1').colors` で、Pastel1 という色の組み合わせのリストが得られる
- 名前の例：`'Pastel1'`、`'Set2'`、`'tab10'`、`'Accent'`（[色の組み合わせの一覧](https://matplotlib.org/stable/users/explain/colors/colormaps.html#qualitative)）

````python
x = [100, 200, 300, 400, 500]
mylabel = ['Japanese', 'mathematics', 'science', 'society', 'English']

mycolor = plt.get_cmap('Pastel1').colors

plt.pie(x, startangle=90, counterclock=False, labels=mylabel, colors=mycolor)
plt.show()
````

- `pie_描画基礎007` のひな形は、`with mpl.rc_context(...)` で**そのグラフの間だけ**色の組み合わせを変える書き方。`colors=` を書くとそちらが優先されるので、`colors=` を消すと Pastel1 の色になる
