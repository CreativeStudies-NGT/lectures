# AM. 回帰のしくみ（交通事故のデータ）

## AM-0 今日の完成形

全国の交通事故（2000〜2025年）の**交通事故件数**と**死者数**の散布図に、**回帰直線**（赤）を引いたもの。直線の式と決定係数は、`plt.text` でグラフの左上に書いている。

1. **全部コピーして**自分の notebook のセルに貼り付け、実行する
1. まずは**色や透明度を変えて**、グラフのどこが変わるかを確かめる
1. そのあと部品（[1]〜[5]）の働きを1つずつ学ぶ（AM-1〜AM-7）

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib
from statsmodels.formula.api import ols

df_acc = pd.read_excel('dsa2026data.xlsx', sheet_name='accident')

res = ols('死者数 ~ 交通事故件数', data=df_acc).fit()  # [1] 回帰直線を求める
a = res.params['Intercept']  # [2] 切片
b = res.params['交通事故件数']  # [2] 傾き
r2 = res.rsquared  # [3] 決定係数

plt.figure(figsize=(5.5, 3.4))
plt.scatter(df_acc['交通事故件数'], df_acc['死者数'], zorder=2)
plt.plot(df_acc['交通事故件数'], res.predict(), color='r', zorder=3)  # [4] 回帰直線
plt.title('交通事故件数と死者数（2000〜2025年）', fontsize=11)
plt.text(0.05, 0.95, f'y = {a:.1f} + {b:.4f} x\n決定係数 {r2:.2f}',
         transform=plt.gca().transAxes, va='top', fontsize=11)  # [5] 回帰式を枠の左上に書く
plt.xlabel('交通事故件数（件）', fontsize=11)
plt.ylabel('死者数（人）', fontsize=11)
plt.show()
```

### 変えてみよう（1か所変えたら、もう一度実行する）

- `plt.text` の `f'…{a:.1f}…'`（f 文字列）は、変数の値を文字の中に入れる書き方（第12講 U-3）。`:.1f` は小数点以下1桁、`:.4f` は4桁、`\n` は改行
- `transform=plt.gca().transAxes` を付けると、`plt.text` の x・y を**枠の大きさに対する割合**で指定できる。左下が (0, 0)、右上が (1, 1)。データの値の範囲が変わっても、いつも枠の同じ位置に書ける
- `plt.gca()` は「今のグラフ」（第８講）。`va='top'` で、文字の上端を y の位置にそろえる

**値で指定する書き方（これまで）とのちがい**

- これまでの `plt.text` は、x・y に**軸の目盛りの値**を入れていた
  - 第８講 V-1：`plt.text(8, 28, '最高 28.0℃')` は、x＝8月、y＝28℃ の場所
  - 第12講 U-7：`plt.text(gdp[i], life[i], country[i])` は、点と同じ場所に国名
- 上の式を値で書くと、たとえば `plt.text(300000, 9000, …)`（交通事故件数 約29万〜95万件、死者数 約2500〜9000人）。これを AN-3 の高齢化率のデータ（x は 23〜39%、y は 10〜19 くらい）にそのまま使うと、(300000, 9000) は枠のはるか外になり、式が正しい場所に出ない。データが変わるたびに、目盛りを見て値を選び直すことになる
- `transform=plt.gca().transAxes` を付けて (0.05, 0.95) にすれば、どのデータでも左上に出る。AM-0 と AN-3 は、列名を替えるだけで同じコードが使える

| | 値で指定（これまで） | 枠に対する割合で指定（`transAxes`） |
|---|---|---|
| x・y の意味 | 軸の目盛りの値 | 枠の左下 (0, 0)〜右上 (1, 1) |
| 書き方 | `plt.text(8, 28, '文字')` | `plt.text(0.05, 0.95, '文字', transform=plt.gca().transAxes)` |
| 向いている使い方 | 特定の点に添える（最高気温、国名） | グラフ全体の説明（式、決定係数） |
| データが変わると | 文字の位置も変わる。値を選び直す | いつも枠の同じ場所 |

- `color='r'` → `color='orange'`、`plt.scatter` に `alpha=0.5` を足す

## AM-1 回帰とは（目的変数と説明変数）

- 散布図の点の並びに、**いちばん合う直線**を引くこと。直線の式は **y = 切片 + 傾き × x**
  - **切片**：x が 0 のときの y
  - **傾き**：x が 1 増えると、y がどれだけ増えるか
- y と x には名前がある

| | 意味 | この例 |
|---|---|---|
| **目的変数**（y） | 説明したいもの・予測したいもの | 死者数 |
| **説明変数**（x） | 説明に使うもの・予測の手がかり | 交通事故件数 |

- この例は「死者数を、交通事故件数で説明する」。事故が多い年ほど死者も多いか、事故が1件増えると死者は何人増えるか、を直線で表す
- どちらを目的変数にするかは、**自分が何を知りたいか**で決める（AM-9）
- 「いちばん合う」とは何かは AM-6 で確かめる
- 説明変数が1つのときを**単回帰**、2つ以上のときを**重回帰**（AO-4）という

## AM-3 回帰直線を求める

- `ols('目的変数 ~ 説明変数', data=データフレーム).fit()` と書く
  - `~` の**左が目的変数**（y）、**右が説明変数**（x）。`'死者数 ~ 交通事故件数'` は「死者数を交通事故件数で説明する」
  - `.fit()` で計算する。結果を `res`（result＝結果）に入れておく
  - `data=` で表を渡す。`~` の左右に書いた列だけが使われる（`年` などの列は使われない）
- 列名に空白や記号があると使えない（くわしい条件は AN-1 の「ols で使える列名の条件」）

```python
res = ols('死者数 ~ 交通事故件数', data=df_acc).fit()
print(res.params)
```

- 表示の `Intercept` は切片、`交通事故件数` は傾き（AM-4）。最後の `dtype: float64` は「小数の数が入っている」という意味で、気にしなくてよい

**`res` から取り出せるもの**

| 書き方 | 出てくるもの | くわしくは |
|---|---|---|
| `res.params` | 切片と傾き | AM-4 |
| `res.predict()` | データの x それぞれに対する直線上の y（26個の数）。`plt.plot` で結ぶと回帰直線 | AM-5 |
| `res.rsquared` | 決定係数（1つの数） | AM-7 |

- `predict` だけ `()` を付ける（計算をする命令なので）。付け忘れると、数ではなく `<bound method …>` と出る

▶ **演習**：`reg_描画基礎001`

## AM-4 切片と傾きを読む

- `res.params` の `Intercept` が**切片**、`交通事故件数` が**傾き**
- この例では y = 19.9 + 0.0078 × x
  - 傾き 0.0078：「事故が1件増えると、死者は約0.0078人増える」。1件では小さすぎて分かりにくいので、**単位を大きくして**読む：「事故が1万件増えると、死者は約78人増える」
  - 切片 19.9：「事故が0件なら、死者は約20人」。事故0件の年はデータに無い（いちばん少ない年でも約29万件）。**切片はあまり意味を持たないことが多い**
- 1つずつ取り出すときは `res.params['Intercept']`、`res.params['交通事故件数']`。`[ ]` の中は `~` の右に書いた列名

```python
a = res.params['Intercept']
b = res.params['交通事故件数']
print('切片', a)
print('傾き', b)
```

▶ **演習**：`reg_描画基礎002`

## AM-6 ずれを線で描く（いちばん合う直線とは）

- 点は直線の上にぴったりは乗らない。点と直線の**縦方向のずれ**（実際の y − 直線上の y）を線で描いて見る
- 直線より上の点は赤、下の点は青にする：`np.where(条件, A, B)`（第13講 Y-4 の別法）
- for で1本ずつ線を引く（第12講）：`plt.plot([x, x], [点の y, 直線の y])` は、(x, 点の y) と (x, 直線の y) を結ぶ縦線

```python
pred = res.predict()                                              # 直線上の y
mycolor = np.where(df_acc['死者数'] >= pred, 'r', 'b')             # [1] 直線より上は赤、下は青

plt.figure(figsize=(5.5, 3.4))
for k in range(len(df_acc)):                                      # [2] 1点ずつ縦線を引く
    plt.plot([df_acc['交通事故件数'][k], df_acc['交通事故件数'][k]], [df_acc['死者数'][k], pred[k]],
             color=mycolor[k], linewidth=1)
plt.scatter(df_acc['交通事故件数'], df_acc['死者数'], c=mycolor, zorder=2)
plt.plot(df_acc['交通事故件数'], pred, color='k', zorder=3)
plt.xlabel('交通事故件数（件）')
plt.ylabel('死者数（人）')
plt.show()

print('ずれの2乗の合計', np.sum((df_acc['死者数'] - pred) ** 2))      # [3]
```

- ずれは、上（赤）にも下（青）にもある。直線は点の真ん中を通る
- ずれをそのまま足すと、プラスとマイナスが打ち消し合う。そこで**2乗してから足す**（[3]）。これが「ずれの2乗の合計」
- **回帰直線は、ずれの2乗の合計がいちばん小さくなる直線**。これが AM-1 の「いちばん合う」の意味（**最小二乗法**。`ols` は Ordinary Least Squares＝最小二乗法 の略）

本当にいちばん小さいか、ほかの直線とくらべて確かめる。`a`・`b` は AM-4 で求めた回帰直線の切片と傾き。

```python
x = df_acc['交通事故件数']
y = df_acc['死者数']
lines = [['回帰直線', a, b, 'r'],
         ['傾きを少し小さく', 509, 0.00703, 'g'],
         ['傾きを少し大きく', -470, 0.00860, 'orange'],
         ['2000年と2025年の点を結ぶ', -357, 0.01012, 'gray']]

plt.figure(figsize=(5.5, 3.4))
plt.scatter(x, y, zorder=2)
for name, a1, b1, c1 in lines:                                    # [1] 直線を1本ずつ
    plt.plot(x, a1 + b1 * x, color=c1, label=name)
    print(f'{name}：ずれの2乗の合計 {np.sum((y - (a1 + b1 * x)) ** 2):.0f}')   # [2]
plt.xlabel('交通事故件数（件）')
plt.ylabel('死者数（人）')
plt.legend()
plt.show()
```

- ずれの2乗の合計は、回帰直線（赤）がいちばん小さい。傾きを少し変えても、ほかの2点を結んでも、合計は大きくなる
- AM-6 の図をよく見ると、2000〜2003年（右上）と2018年以降（左下）は直線より上、その間は下に並ぶ。点の並びが少し曲がっているためで、直線ではとらえきれない部分（AM-10）

## AM-7 決定係数

**決定係数**（R²）は、**目的変数のばらつきのうち、説明変数で説明できる割合**。0〜1 の値で、1 に近いほど点が直線のまわりに集まっている。

- 死者数は年によって 2500人〜9000人とばらついている。そのばらつきのうち、事故件数の違いで説明できるのは何割か、を表す
- 次の2つの「ずれの2乗の合計」をくらべて求める
  - [1] **平均だけで予測したとき**：説明変数を使わず、どの年も「死者数の平均（約4900人）」と予測したときのずれ。もとの**ばらつきの大きさ**
  - [2] **回帰直線で予測したとき**：事故件数を使って予測したときのずれ（AM-6 の [3]）。**説明しきれずに残ったずれ**
- **決定係数 ＝ 1 − [2] ÷ [1]**。説明変数を使うことで、ずれが何割減ったか
- `plt.subplot(1, 2, 1)` は、横に2つ並べたグラフの1つ目（第10講 AI-1）。`plt.axhline(値)` は横線（第４講 G）

```python
x = df_acc['交通事故件数']
y = df_acc['死者数']
pred = res.predict()

plt.figure(figsize=(9, 3.2))
plt.subplot(1, 2, 1)
for k in range(len(df_acc)):
    plt.plot([x[k], x[k]], [y[k], y.mean()], color='gray', linewidth=1)
plt.scatter(x, y, zorder=2)
plt.axhline(y.mean(), color='k')
plt.title('[1] 平均だけで予測したときのずれ')
plt.xlabel('交通事故件数（件）')
plt.ylabel('死者数（人）')

plt.subplot(1, 2, 2)
for k in range(len(df_acc)):
    plt.plot([x[k], x[k]], [y[k], pred[k]], color='gray', linewidth=1)
plt.scatter(x, y, zorder=2)
plt.plot(x, pred, color='r')
plt.title('[2] 回帰直線で予測したときのずれ')
plt.xlabel('交通事故件数（件）')
plt.show()

s1 = np.sum((y - y.mean()) ** 2)    # [1] 平均からのずれの2乗の合計
s2 = np.sum((y - pred) ** 2)        # [2] 回帰直線からのずれの2乗の合計
print(f'[1] {s1:.0f}')
print(f'[2] {s2:.0f}')
print('[2] ÷ [1] =', s2 / s1)
print('1 − [2] ÷ [1] =', 1 - s2 / s1)
print('res.rsquared =', res.rsquared)
```

- 縦線（ずれ）は、右の回帰直線の方がずっと短い
- [2] ÷ [1] は約 0.11：事故件数を使うと、ずれ（の2乗の合計）が、平均だけのときの**約1割に減った**
- 決定係数は 1 − 0.11 ≒ **0.89**。`res.rsquared` と同じ値になる
- 読み方：「年ごとの死者数のばらつきの**約89%**は、交通事故件数の違いで説明できる。残りの約11%は、事故件数以外の理由（事故1件あたりの重さの変化など）」

**決定係数の読み方の注意**

| 注意 | 説明 |
|---|---|
| 何以上なら「よい」という決まりは無い | 分野やデータで変わる。物の測定では 0.9 以上が普通でも、人の行動や社会のデータでは 0.3 でも意味がある関係のことがある |
| 傾きの大きさとは別のもの | 傾き 0.0078 は小さく見えるが、決定係数は 0.89 と高い。傾きは「x が増えると y がどれだけ増えるか」、決定係数は「点が直線のまわりにどれだけ集まっているか」 |
| 高くても、原因とは限らない | 決定係数は「一緒に動く」度合い。原因かどうかは別に考える（AM-10） |
| 相関係数との関係 | 単回帰では、決定係数は x と y の**相関係数の2乗**と同じ（前期のデータサイエンス概説）。相関係数 0.95 なら決定係数 0.89 |

- 決定係数だけを取り出すときは `res.rsquared`

▶ **演習**：`reg_描画基礎003`

## AM-9 ols の書き方（列が多いデータ）

- `accident` シートには、`交通事故件数`・`死者数` のほかに `年`・`負傷者数`・`重傷者数` の列もある
- 列がたくさんあっても、`ols` が使うのは **`~` の左右に書いた2列だけ**。どれを目的変数、どれを説明変数にするかは、自分で決めて列名を書く

**[1] 左右を入れかえる**：交通事故件数（目的変数）を死者数（説明変数）で説明する

```python
res_r = ols('交通事故件数 ~ 死者数', data=df_acc).fit()   # [1] 左右を入れかえると別の直線
print(res_r.params)
print('決定係数', res_r.rsquared)
```

- 傾き 114：死者が1人多い年は、事故が約114件多い
- AM-4 の傾き 0.0078 を逆にした 1 ÷ 0.0078 ≒ 128 には**ならない**。左右を入れかえると**別の直線**になる（どちらも「縦方向のずれ」が最小の直線だが、縦の向きが変わる）
- 決定係数は AM-7 と同じ 0.89
- 結果の変数は `res` と別の名前（`res_r`）にした。同じ名前にすると、AM-3 の結果が消える
- `~` の左には、**予測したいもの**（目的変数）を書く。「事故の件数から死者数を見積もりたい」なら AM-3 の書き方

**[2] 説明変数を年にする**：交通事故件数（目的変数）が年（説明変数）とともにどう変わるか

```python
res_y = ols('交通事故件数 ~ 年', data=df_acc).fit()   # [2] x を年にする（1年あたりの増減）
print(res_y.params)
print('決定係数', res_y.rsquared)

plt.scatter(df_acc['年'], df_acc['交通事故件数'], zorder=2)
plt.plot(df_acc['年'], res_y.predict(), color='r', zorder=3)
plt.xlabel('年')
plt.ylabel('交通事故件数')
plt.show()
```

- 表示の `6.514341e+07` は 6.514341 × 10⁷（＝ 65143410）、`-3.205821e+04` は −3.205821 × 10⁴（＝ −32058）。大きい数や小さい数は、このように表示されることがある
- 傾き −32058：1年ごとに、事故が約3万2千件ずつ減っている。決定係数 0.97
- 切片 6514万 は「西暦0年の件数」で、意味はない（AM-4）
- 年を説明変数にすると、傾きは**1年あたりの増減**になる
- この直線を延ばすと、2032年に事故は0件、2040年には約 −25万件になってしまう（AM-10）

| 書き方 | 意味 |
|---|---|
| `'死者数 ~ 交通事故件数'` | 死者数を交通事故件数で説明する（AM-3） |
| `'交通事故件数 ~ 死者数'` | 交通事故件数を死者数で説明する。上とは別の直線 |
| `'交通事故件数 ~ 年'` | 交通事故件数の1年あたりの増減 |

- 傾きを取り出す `res.params['…']` の中も、`~` の右に書いた列名にする（例：`res_y.params['年']`）
- `~` の右に `+` で列を足すと、説明変数が2つ以上の回帰（重回帰）になる。AO-4（発展）

### ols で使える列名の条件

`ols('y の列名 ~ x の列名', …)` の式に書く列名は、次の条件を満たす必要がある。

| 条件 | だめな例 | 直した例 |
|---|---|---|
| 日本語は使える | | `負傷者数`、`一人あたりGDP` |
| 空白を入れない | `人口 総数` | `人口総数` |
| かっこを入れない | `人口（千人）`、`割合(%)` | `人口`、`割合` |
| 記号を入れない（`-` `/` `%` `・` `#` `【】` など） | `65歳以上-割合`、`男・女` | `高齢者割合`、`男女計` |
| 数字で始めない | `2020年人口` | `人口2020` |

- 条件に合わない列名のまま `ols` に書くと、`SyntaxError` や `PatsyError` などのエラーになる
- **直し方**：notebook の中で直すのではなく、**Excel の1行目（列名）を書き換えて保存し、読み直す**

## AO-1 乱数で作ったデータで確かめる

- 実データでは「正解の直線」が分からない。そこで、**正解が y = 2 + 0.5x と分かっているデータ**を乱数で作り、回帰でその直線が見つかるか確かめる
- `rng.uniform(0, 10, 50)`：0〜10 の数を50個（どの値も同じくらい出る。第13講 AW）
- `rng.normal(0, 1, 50)`：平均0・標準偏差1 のばらつきを50個（第13講 Y-3）。これを `e` に入れ、正解の直線に足す
- `pd.DataFrame({'x': x, 'y': y})` で表にすると、`ols` に渡せる

```python
rng = np.random.default_rng(348309421)
x = rng.uniform(0, 10, 50)
e = rng.normal(0, 1, 50)                 # ばらつき
y = 2 + 0.5 * x + e                      # [1] 正解の直線に、ばらつきを足す
df_sim = pd.DataFrame({'x': x, 'y': y})   # [2] 表にする

res_sim = ols('y ~ x', data=df_sim).fit()
print(res_sim.params)
print('決定係数', res_sim.rsquared)

plt.scatter(df_sim['x'], df_sim['y'], alpha=0.6, zorder=2)
plt.plot(df_sim['x'], res_sim.predict(), color='r', zorder=3, label='回帰直線')
plt.plot([0, 10], [2, 7], color='gray', linestyle='--', label='正解 y = 2 + 0.5x')
plt.legend()
plt.show()
```

- 切片・傾きは 2・0.5 に近いが、ぴったりではない。ばらつきがあるので、50個の点からは正解を少しずれて見積もる
- ずれの2乗の合計（AM-6）をくらべると、回帰直線の方が正解の直線より小さい。回帰直線は「この50個の点に**いちばん**合う直線」なので、正解の直線よりもずれが小さくなる

```python
print('回帰直線のずれの2乗の合計', np.sum((df_sim['y'] - res_sim.predict()) ** 2))
print('正解の直線のずれの2乗の合計', np.sum((df_sim['y'] - (2 + 0.5 * df_sim['x'])) ** 2))
```

- 決定係数は、ばらつきの大きさで変わる。下で、ばらつき `e` を 0.5倍・2倍・4倍にして確かめる

```python
for s in [0.5, 1, 2, 4]:
    y = 2 + 0.5 * x + s * e
    res_s = ols('y ~ x', data=pd.DataFrame({'x': x, 'y': y})).fit()
    b_s = res_s.params['x']
    print(f'ばらつき {s}倍：傾き {b_s:.3f}、決定係数 {res_s.rsquared:.2f}')
```

- ばらつきが大きいほど決定係数は下がる。正解の直線は同じでも、点が直線のまわりに散らばるほど、説明できる割合（AM-7）は小さくなる
- 傾きも正解の 0.5 から**ずれやすくなる**。決定係数が低いときは、傾きの値もあまりあてにならない。点の数を増やすと、ずれは小さくなる（`50` を `1000` にして確かめてみよう。AO-1 の最初のコードセルの2か所）

## AO-2 多項式回帰（曲線を当てはめる）

- `I(x ** 2)` を足すと2次式（放物線）：y = 切片 + 係数1 × x + 係数2 × x²
- 正解が放物線 y = 1 + 3x − 0.3x² のデータを乱数で作り（AO-1 と同じ作り方）、直線と2次式を当てはめてくらべる

```python
rng = np.random.default_rng(1)
x = np.linspace(0, 10, 50)
df_q = pd.DataFrame({'x': x, 'y': 1 + 3 * x - 0.3 * x ** 2 + rng.normal(0, 0.8, 50)})   # [1] 放物線＋ばらつき

res1 = ols('y ~ x', data=df_q).fit()
res2 = ols('y ~ x + I(x ** 2)', data=df_q).fit()   # [2] x² の項を足す
print(res2.params)
print('直線の決定係数', res1.rsquared)
print('2次式の決定係数', res2.rsquared)

plt.scatter(df_q['x'], df_q['y'], zorder=2)
plt.plot(df_q['x'], res1.predict(), color='gray', label='直線')
plt.plot(df_q['x'], res2.predict(), color='r', label='2次式')   # [3] 2次式の予測を線で結ぶ
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

- 2次式（赤）は点の山形に沿う。係数は正解の 1, 3, −0.3 に近い。x² の係数がマイナスなので、上に凸（山形）になる
- 直線（灰色）はほぼ水平で、決定係数は0に近い。**決定係数が低くても「関係が無い」とは限らない**。まず散布図を見る（AM-10）
- `np.linspace(0, 10, 50)` は、0から10までを等間隔に並べた50個の数（第10講・第12講）
- 次数を上げすぎると、点にぴったり合っても予測には役に立たない曲線になる（過学習）

▶ **演習**：`reg_描画基礎007`（直線と2次式の決定係数をくらべる）

### 係数の標準誤差（係数がどのくらいあてになるか）

- データには**ばらつき**がある。上のデータは、正解の放物線 y = 1 + 3x − 0.3x² に乱数でばらつきを足して作った。だから求めた x² の係数は −0.288 で、正解の −0.3 とぴったりは一致しない
- 同じ条件でデータを**取り直す**と、ばらつき方が変わるので、求めた係数も少し変わる。−0.29 になったり −0.31 になったりする
- **標準誤差**は、取り直したときに係数がふつうどのくらい変わるかの目安。標準誤差が小さいほど、係数はあてになる
- 結果は「係数 ± 標準誤差」の形で書く（例：x² の係数 −0.288 ± 0.013）
- うれしいのは、**データ1組だけから計算できる**こと。`res2.bse` で取り出せる（bse は standard error の略）

```python
print(pd.DataFrame({'係数': res2.params, '標準誤差': res2.bse}))   # [1] 係数と、その標準誤差を並べる
print('決定係数', res2.rsquared)
print('残差の標準偏差', np.sqrt(res2.mse_resid))   # [2] 点が曲線から、ふつうどのくらいずれるか
print('自由度', res2.df_resid)   # [3] 点の数 − 係数の数（50 − 3）
```

- 本当に「取り直したときの変わり方」になっているか、乱数を替えてデータを1000回作り直し、そのたびに x² の係数を求めて確かめる

```python
c2 = []
for seed in range(1000):   # [1] 乱数を替えて、データを1000回取り直す
    rng_r = np.random.default_rng(seed)
    df_r = pd.DataFrame({'x': x, 'y': 1 + 3 * x - 0.3 * x ** 2 + rng_r.normal(0, 0.8, 50)})
    c2.append(ols('y ~ x + I(x ** 2)', data=df_r).fit().params['I(x ** 2)'])   # [2] x² の係数だけ集める

print('最初の5回', np.round(c2[:5], 3))
print('1000回の係数の標準偏差', np.std(c2))   # [3] 取り直したときの変わり方の大きさ
print('1組のデータからの標準誤差', res2.bse['I(x ** 2)'])

plt.hist(c2, bins=30)
plt.axvline(-0.3, color='r', label='正解 −0.3')
plt.xlabel('x² の係数')
plt.ylabel('回数')
plt.legend()
plt.show()
```

- 係数は取り直すたびに −0.3 のまわりでばらつく。そのばらつきの大きさ（標準偏差）は約 0.014 で、データ1組から計算した標準誤差（約 0.013）とほぼ同じ
- 現実のデータは、ふつう1組しか取れない。それでも標準誤差を見れば、係数がどのくらいあてになるかがわかる
- 標準誤差は、次の3つで決まる
  - 点のばらつき（残差の標準偏差）が小さいほど、小さい
  - **点の数 n**（ここでは50）が多いほど、小さい。小さくなり方は √n に比例するので、点を4倍にしても標準誤差は半分にしかならない
  - x の範囲が広いほど、小さい
- 取り直しの回数（1000回）は n ではない。回数を増やすとヒストグラムがなめらかになるだけで、ばらつきの大きさ（約 0.014）は変わらない

**標準偏差と標準誤差のちがい**（第９講 AB-4・AC-4 でも、平均について同じ話をした）

- **標準偏差**は、**データ1つ1つ**のばらつき。このデータでは、点が曲線から上下にずれる大きさ（残差の標準偏差 0.72）。データを作るときに足した乱数のばらつき 0.8 に近い
- **標準誤差**は、データから**求めた値**（係数や平均）のばらつき。x² の係数なら 0.013
- 取り直し実験では、1000個の**係数**の標準偏差を計算した。それが標準誤差と一致した。つまり、標準誤差は「求めた値の標準偏差」のこと

| | 標準偏差 | 標準誤差 |
|---|---|---|
| 何のばらつきか | データ1つ1つ（点） | 求めた値（係数・平均） |
| このデータでは | 0.72（点が曲線からずれる大きさ） | 0.013（x² の係数のぶれ） |
| 点を増やすと | ほとんど変わらない（データの性質だから） | 小さくなる（√n に比例して） |
| 答える問い | 1つ1つの点は、どのくらい散らばっているか | 求めた値は、どのくらいあてになるか |
| 使う場面 | データの様子を説明する（例：点数のばらつき） | 結果を「値 ± 標準誤差」で報告する |

**テストの点数でたとえると**

ある学年 100人がテストを受け、平均 65点、標準偏差 15点だったとする。

- **標準偏差 15点**：一人ひとりの点数の散らばり。「多くの人は 50〜80点くらい（65 ± 15）にいる」という、**個人**についての話
- **標準誤差 1.5点**（＝ 15 ÷ √100）：**平均点**のぶれ。同じくらいの力の別の 100人（たとえば来年の学年）が受けたら、平均点は 65 ± 1.5、つまり 63.5〜66.5点くらいになりそう、という話
- 人数を 400人に増やすと
  - 標準偏差は 15点のまま。人を増やしても、一人ひとりの点数の散らばりは変わらない
  - 標準誤差は 15 ÷ √400 ＝ 0.75点に半分になる。人数が多いほど、平均点はあてになる
- 使い分けの場面
  - 「自分の 80点は、みんなの中でどのくらい？」→ 標準偏差と比べる。平均より 1標準偏差ほど上なので、上のほう
  - 「A組の平均 65点と B組の平均 66点（どちらも 100人、標準偏差 15点）、差はある？」→ 標準誤差と比べる。差の 1点は平均点のぶれ（1.5点）より小さいので、たまたまかもしれない
- 回帰の係数も同じ。「勉強時間が1時間増えると点数が 5点上がる（傾き 5）」で、傾きの標準誤差が 0.5 なら、傾きは 5 ± 0.5 でかなりあてになる。標準誤差が 4 なら、5 ± 4 で「上がる」とも言い切りにくい

- 平均の場合は、標準誤差 ＝ 標準偏差 ÷ √n（第９講 AC-4）。係数の標準誤差も、同じように √n で小さくなる
- エラーバーや「± いくつ」を見たら、それが標準偏差か標準誤差かを確かめる。標準誤差のほうがずっと短いので、取りちがえると印象が大きく変わる（第９講 AB-1・AB-4）

- 同じ計算は Excel の `LINEST` 関数でもできる。x が A2:A51、y が B2:B51 にあるとき、縦5行×横3列を選んで `=LINEST(B2:B51, A2:A51^{1,2}, TRUE, TRUE)` と入力する（`^{1,2}` は「x と x² を作る」、後ろの `TRUE, TRUE` は「切片も求める」「標準誤差なども出す」という意味）
- 出力と Python の対応は次のとおり。**LINEST は係数が逆順（x² → x → 切片）に並ぶ**ことに注意

| LINEST の出力 | 1列目 | 2列目 | 3列目 |
|---|---|---|---|
| 1行目：係数 | `res2.params['I(x ** 2)']` | `res2.params['x']` | `res2.params['Intercept']` |
| 2行目：標準誤差 | `res2.bse['I(x ** 2)']` | `res2.bse['x']` | `res2.bse['Intercept']` |
| 3行目 | 決定係数 `res2.rsquared` | 残差の標準偏差 `np.sqrt(res2.mse_resid)` | （#N/A） |
| 4行目 | F 値 `res2.fvalue` | 自由度 `res2.df_resid` | （#N/A） |
| 5行目 | 回帰の平方和 `res2.ess` | 残差の平方和 `res2.ssr` | （#N/A） |

- `print(res2.summary())` とすると、係数（coef）・標準誤差（std err）・決定係数などを1つの表でまとめて見られる

## AO-3 指数関数の回帰（割合で増える・減るもの）

- 預金の利息や人口の増加のように、「何ずつ」ではなく「**何%ずつ**」変わるものは、指数関数 y = A × e^(b x) が合う
- 両辺の log（自然対数）をとると **log y = log A + b x** で、x の**直線**になる。そこで `np.log(列名)` を y にして、ふつうの回帰をする
- `np.exp(b) - 1` が、x が1増えるごとの**変化の割合**（マイナスなら減る）
- 正解が「100 から、x が1増えるごとに 30% ずつ増える」（y = 100 × 1.3^x）データを乱数で作って確かめる

```python
rng = np.random.default_rng(2)
x = np.linspace(0, 10, 50)
df_e = pd.DataFrame({'x': x, 'y': 100 * 1.3 ** x * np.exp(rng.normal(0, 0.1, 50))})   # [1] 30%ずつ増える＋ばらつき

res_lin = ols('y ~ x', data=df_e).fit()
res_exp = ols('np.log(y) ~ x', data=df_e).fit()   # [2] y の log を回帰
b = res_exp.params['x']
print(f'x が1増えるごとに {(np.exp(b) - 1) * 100:.1f} % ずつ変わる')   # [3] 変化の割合

plt.scatter(df_e['x'], df_e['y'], zorder=2)
plt.plot(df_e['x'], res_lin.predict(), color='gray', label='直線')
plt.plot(df_e['x'], np.exp(res_exp.predict()), color='r', label='指数関数')   # [4] np.exp で元に戻す
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

- 変化の割合は、正解の 30% に近い
- `res_exp.predict()` は log y の予測なので、`np.exp(…)` で元の y に戻してから描く
- 直線（灰色）は、両端で点より下、真ん中で点より上になる。指数関数（赤）の方が点に沿う
- 減る場合（例：`0.7 ** x`、30% ずつ減る）も同じ。直線を延ばすといつかマイナスになるが、指数関数は0に近づくだけで、マイナスにはならない（AM-10）
- `res_exp.rsquared` は log y に対する決定係数なので、直線の `res_lin.rsquared` とそのまま比べない
- Excel では、散布図の「近似曲線の追加」→「指数近似」で同じ曲線が出る

## AO-4 重回帰と標準化

- x を2つ以上使う：`'tip ~ total_bill + size'`（食事代と人数でチップを説明する）

```python
df_tip = pd.read_excel('dsa2026data.xlsx', sheet_name='tip_data')
res2 = ols('tip ~ total_bill + size', data=df_tip).fit()
print(res2.params)
print('決定係数', res2.rsquared)
```

- `total_bill` の係数：人数が同じなら、食事代が1ドル増えるとチップが約0.09ドル増える
- `size` の係数：食事代が同じなら、1人増えるとチップが約0.19ドル増える

**3D で見る**

- 単回帰は「点に合う**直線**」だった。説明変数が2つになると、x が2つ・y が1つで3つの軸になり、「点に合う**平面**」になる
- 面より上の点を赤、下の点を青にした（AM-6 の上下の色分けと同じ）
- [1]〜[2] は面を描くための準備（格子の点ごとに予測を求める）。中身は分からなくてよい
- 3D の面・散布図・z 軸名・見る角度は `plt` に関数が無いので、`plt.gca().〜` で書く

```python
g_bill, g_size = np.meshgrid(np.linspace(0, 55, 12), np.linspace(1, 6, 6))   # [1] 面を描く格子
g_tip = res2.predict(pd.DataFrame({'total_bill': g_bill.ravel(), 'size': g_size.ravel()})).values.reshape(g_bill.shape)   # [2] 格子の点ごとの予測
c = np.where(df_tip['tip'] >= res2.predict(), 'r', 'b')   # [3] 面より上は赤、下は青

plt.figure(figsize=(5.5, 4.8))
plt.subplot(projection='3d')   # [4] 3D のグラフにする
plt.gca().plot_surface(g_bill, g_size, g_tip, color='gray', alpha=0.3)   # [5] 回帰の面
plt.gca().scatter(df_tip['total_bill'], df_tip['size'], df_tip['tip'], c=c, s=15)   # [6] 3D の散布図
plt.xlabel('食事代 total_bill（ドル）')
plt.ylabel('人数 size（人）')
plt.gca().set_zlabel('チップ tip（ドル）')
plt.title('重回帰：tip = 0.67 + 0.093 × total_bill + 0.19 × size')
plt.gca().view_init(elev=15, azim=-100)   # [7] 見る角度（elev：高さ、azim：横の回転）
plt.show()
```

- 人数は 1〜6 の整数なので、点は人数ごとに縦の列に並ぶ
- 面の傾きが係数：食事代の向きに 1 ドル進むと 0.093、人数の向きに 1 人進むと 0.19 だけ面が上がる
- 変えてみよう：[7] の `azim=` を `-60`・`-10`・`30` などに変えて、横から回して見る。`elev=` を大きくすると上から見下ろす
- 時間があれば：上のセルの1行目に次のどちらかを足して実行し直すと、図を**マウスで回せる**。終わったら、`%matplotlib inline` だけを書いたセルを実行して元に戻す（このあとの図が、ふつうの図で出るように）
  - `%matplotlib tk`：notebook の外に**別ウィンドウ**で開く。ドラッグで回す、右ドラッグで拡大・縮小。ウィンドウを閉じるとセルの実行が終わる。notebook の後ろに隠れることがある
  - `%matplotlib widget`：**notebook の中**で回る（ipympl。Anaconda に入っている）

- 「食事代と人数、どちらが効いているか」は、単位（ドルと人）が違うのでこのままでは比べられない
- 各列を**標準化**（平均0・標準偏差1）してから回帰すると、係数の大きさで比べられる

```python
cols = ['tip', 'total_bill', 'size']
df_z = (df_tip[cols] - df_tip[cols].mean()) / df_tip[cols].std()

res3 = ols('tip ~ total_bill + size', data=df_z).fit()
print(res3.params)
print('決定係数', res3.rsquared)
```

- 標準化した係数は `total_bill` の方が大きい → チップには食事代の方が強く効いている
- 決定係数は標準化しても変わらない
- 切片の `2.775558e-16` は「2.78 × 10 の −16 乗」＝ほぼ 0（標準化すると切片は 0 になる）

## AO-5 回帰のひな形を関数にする

- AN-6 のひな形は、データや列名を変えるたびに何行も書き換える。第13講の**関数**にすると、引数を変えるだけで済む
- `return res` で回帰の結果を返すので、あとで `res.params` なども見られる

```python
def reg_plot(df, x_name, y_name):
    res = ols(f'{y_name} ~ {x_name}', data=df).fit()
    plt.scatter(df[x_name], df[y_name], zorder=2)
    plt.plot(df[x_name], res.predict(), color='r', zorder=3)
    plt.title(f'{y_name} と {x_name}（決定係数 {res.rsquared:.2f}）')
    plt.xlabel(x_name)
    plt.ylabel(y_name)
    plt.show()
    return res

res = reg_plot(df_acc, '交通事故件数', '負傷者数')
res = reg_plot(df_acc, '年', '死者数')
print(res.params)
```

## AO-6 グループごとに回帰する

- `df[条件]`（第13講 AL-3）でデータを分け、for（第12講）で1つずつ回帰する。if（第13講）で色を変える
- tip_data を昼（Lunch）と夜（Dinner）に分けて、チップの傾きを比べる

```python
plt.figure(figsize=(5.5, 3.4))
for t in ['Lunch', 'Dinner']:
    if t == 'Lunch':
        c = 'orange'
    else:
        c = 'navy'
    d = df_tip[df_tip['time'] == t]
    res_t = ols('tip ~ total_bill', data=d).fit()
    b_t = res_t.params['total_bill']
    plt.scatter(d['total_bill'], d['tip'], color=c, alpha=0.4)
    plt.plot(d['total_bill'], res_t.predict(), color=c, label=f'{t}：傾き {b_t:.3f}')
plt.xlabel('食事代 (USD)')
plt.ylabel('チップ (USD)')
plt.legend()
plt.show()
```

- 凡例に、それぞれの傾きが出る。昼と夜で、食事代あたりのチップの割合を比べられる
- `'time'` を `'smoker'`（喫煙席か）、リストを `['Yes', 'No']` に変えると、喫煙・禁煙で比べられる

# まとめ

| やりたいこと | 書き方 |
|---|---|
| 準備 | `from statsmodels.formula.api import ols` |
| 回帰直線を求める | `res = ols('目的変数 ~ 説明変数', data=df).fit()` |
| `~` の左右 | 左が目的変数（y、予測したいもの）、右が説明変数（x、手がかり）。入れかえると別の直線になる（AM-9） |
| 切片・傾き | `res.params`（`res.params['Intercept']`、`res.params['説明変数の列名']`） |
| 回帰直線を描く | `plt.plot(df['説明変数の列名'], res.predict(), color='r')` |
| ずれの2乗の合計 | `np.sum((df['目的変数の列名'] - res.predict()) ** 2)` |
| 決定係数 | `res.rsquared` |
| 予測 | `a + b * x の値`（補足：`res.predict(pd.DataFrame({'説明変数の列名': [値]}))`） |
| 直線との上下で色分け | `np.where(df['目的変数の列名'] >= res.predict(), 'r', 'b')` |
| 実データを読む | Excel で整えてから `pd.read_excel(…)` |
| 列名の条件 | 日本語は可。空白・かっこ・記号・先頭の数字は不可。**Excel で直す**（AN-1） |
| ひな形 | `x_name`・`y_name` に列名を入れ、`ols(f'{y_name} ~ {x_name}', data=df)`（AN-6） |
| 決定係数を枠の左上に | `plt.text(0.05, 0.95, f'決定係数 {res.rsquared:.2f}', transform=plt.gca().transAxes, va='top')` |
| 読んだら確かめる | まず `df.info()`（型と空欄）。次に `df.head()`・`df.tail()`、`df.describe()`、`df.isna().sum()`、`list(df.columns)`、`df.shape` |

- 回帰直線は、ずれの2乗の合計がいちばん小さい直線（最小二乗法、AM-6）
- 傾きは「説明変数が 1 増えると、目的変数がどれだけ増えるか」。単位を付けて言葉で説明する
- 決定係数は「目的変数のばらつきのうち、説明変数で説明できる割合」（0〜1、AM-7）。傾きの大きさとは別のもの
- 回帰は「一緒に動く」ことしか言えない。原因かどうかは別に考える。データの範囲の外は予測しない
