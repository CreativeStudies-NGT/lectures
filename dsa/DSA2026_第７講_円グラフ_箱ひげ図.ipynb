## Q-0 今日の完成形

よく使う端末のアンケート結果（架空）。12時の位置から時計回りに、大きい順に並べている。

1. **全部コピーして**自分の notebook のセルに貼り付け、実行する

1. そのあと部品（[1]〜[7]）の働きを1つずつ学ぶ（Q-1〜Q-6）

````python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib

values = [62, 25, 9, 4]
labels = ['スマホ', 'PC', 'タブレット', 'その他']
colors = ['lightcoral', 'gold', 'lightgreen', 'lightgray']
shift = [0.1, 0, 0, 0]

plt.figure(figsize=(6, 6))  # [0] グラフの大きさ
plt.pie(values,  # [1] 円グラフ
        startangle=90, counterclock=False,  # [2] 12時から時計回り
        labels=labels,  # [3] ラベル
        colors=colors,  # [4] 色
        explode=shift,  # [5] 切り出し
        autopct='%.1f%%', pctdistance=0.7,  # [6] 割合の表示
        wedgeprops={'edgecolor': 'w'})  # 境目を白い線に
plt.title('よく使う端末', fontsize=20)  # [7] タイトル
plt.show()
````

## Q-1 円グラフの基本と始まりの位置

- `plt.pie(値のリスト)` で描く。割合は**自動で計算**される（合計が 100 でなくてよい）
- 何も指定しないと、**3時の位置**から**反時計回り**に並ぶ

````python
x = [100, 200, 300, 400, 500]

plt.pie(x)
plt.show()
````

- `startangle=90` で**12時の位置**から始まる（角度は3時の位置から反時計回りに数える。90度＝12時）
- `counterclock=False` で**時計回り**になる（counter clock＝反時計回り、を False＝しない）
- 時計と同じ「12時から時計回り」にすると読みやすい

````python
plt.pie(x, startangle=90, counterclock=False)
plt.show()
````

▶ **演習**：`pie_描画基礎001`

## Q-2 ラベル

- `labels=リスト` で、それぞれの部分に名前を付ける。値と**同じ個数・同じ順番**で書く

````python
label = ['Japanese', 'mathematics', 'science', 'history', 'English']

plt.pie(x, startangle=90, counterclock=False, labels=label)
plt.show()
````

- 日本語のラベルも使える（`import japanize_matplotlib` を実行してあれば）
- 個数が違うと `ValueError` になる

▶ **演習**：`pie_描画基礎002`

## Q-3 色

- `colors=リスト` で、部分ごとに色を決める。棒グラフの `color=` と違い、**s が付く**ことに注意
- 色の名前は第３講の色の表（`'lightcoral'`、`'gold'` など）を使う

````python
mylabel = ['a', 'b', 'c', 'd', 'e']
mycolor = ['lightcoral', 'lightpink', 'gold', 'lightgreen', 'lightblue']

plt.pie(x, startangle=90, counterclock=False, labels=mylabel, colors=mycolor)
plt.show()
````

▶ **演習**：`pie_描画基礎003`

## Q-4 一部を切り出す

- `explode=リスト` で、部分ごとに中心から**どれだけ離すか**を決める。`0` は離さない
- 目立たせたい部分だけ `0.1` くらいにする

````python
mylabel = ['Japanese', 'mathematics', 'science', 'society', 'English']
shift = [0, 0.1, 0, 0, 0]

plt.pie(x, startangle=90, counterclock=False, labels=mylabel, explode=shift)
plt.show()
````

▶ **演習**：`pie_描画基礎004`

## Q-5 割合を表示する

- `autopct='%.1f%%'` で、各部分に**割合（%）**を書く。`.1f` は小数第1位まで、`.2f` は第2位まで
- 最後の `%%` は「% の記号を表示する」という意味（`%` を1つだけ書くとエラー）
- `pctdistance=` で、割合の文字を**中心からどれだけ離すか**を決める（`0` が中心、`1` が円の縁。何も書かないと `0.6`）

````python
plt.pie(x, startangle=90, counterclock=False, labels=mylabel, autopct='%.2f%%', pctdistance=0.7)
plt.show()
````

- 演習のひな形の解答例では `autopct="%1.2f%%"` と書かれている。`%1.2f` と `%.2f` は同じ表示になる

▶ **演習**：`pie_描画基礎005`

## Q-6 灰色だけで描く

- 色を `'0.75'` のように **0〜1 の数を文字列で**書くと、灰色になる（`'0'` が黒、`'1'` が白）
- 白黒で印刷する資料に使う

````python
grayscales = ['0.75', '0.65', '0.55', '0.45', '0.35']

plt.pie(x, startangle=90, counterclock=False, labels=mylabel, autopct='%.1f%%', colors=grayscales)
plt.show()
````

- `'0.75'` を `0.75`（引用符なし）にするとエラーになる。**文字列で**書くこと

▶ **演習**：`pie_描画基礎006`

## Q-7 Excel のデータで円グラフ

- `tip_data` シートは、あるレストランの244件の会計の記録（食事代・チップ・曜日など）
- `df['列名'].value_counts()` で、**同じ値が何件あるか**を数えられる。曜日ごとの件数を円グラフにする

````python
df_tip = pd.read_excel('dsa2026data.xlsx', sheet_name='tip_data')
count = df_tip['day'].value_counts()
count
````

````python
plt.figure(figsize=(6, 6))
plt.pie(count, labels=count.index, startangle=90, counterclock=False, autopct='%.1f%%')
plt.title('曜日ごとの会計の件数')
plt.show()
````

- `value_counts()` は**件数の多い順**に並べてくれる。12時から時計回りに大きい順に並ぶ
- `count.index` は、数えた値の名前（ここでは曜日）

### 円グラフを使うときの注意

| 注意 | 理由 |
|---|---|
| 項目は5つくらいまで | 多いと細くて読めない。少ないものは「その他」にまとめる |
| 合計に意味があるときだけ使う | 「全体のうち何 %」でないもの（月ごとの気温など）は棒グラフや折れ線グラフにする |
| 立体（3D）にしない | 手前の部分が大きく見えてしまう（第12講） |
| 2つの円グラフを並べて比べない | 角度の違いは読み取りにくい。比べるなら棒グラフ |

# R. 箱ひげ図

**箱ひげ図**は、いくつかのグループの**ばらつき**を並べて比べるグラフ。ヒストグラムを横から見て、5つの数にまとめたもの。

## R-0 箱ひげ図の読み方

````python
data = [50, 52, 58, 61, 63, 65, 67, 68, 70, 71, 72, 73, 74, 75, 75,
        76, 77, 78, 79, 80, 81, 82, 84, 85, 86, 88, 90, 92, 95, 20]

plt.figure(figsize=(6, 6))
plt.boxplot(data)
plt.text(1.1, np.max(data[:-1]), '最大（ひげの上の端）', va='center')
plt.text(1.1, np.percentile(data, 75), '上から 1/4 の位置（箱の上）', va='center')
plt.text(1.1, np.median(data), '中央値（箱の中の線）', va='center')
plt.text(1.1, np.percentile(data, 25), '下から 1/4 の位置（箱の下）', va='center')
plt.text(1.1, np.min(data[:-1]), '最小（ひげの下の端）', va='center')
plt.text(1.1, 20, '外れ値（離れすぎた値）', va='center')
plt.xlim(0.5, 2.5)
plt.ylabel('点数')
plt.show()
````

| 部分 | 意味 |
|---|---|
| 箱の中の線 | **中央値**（大きさの順に並べて真ん中の値） |
| 箱 | 真ん中の半分（50%）の人が入る範囲。箱が**長いほどばらつきが大きい** |
| ひげ | 外れ値を除いた最小から最大まで |
| ○ | **外れ値**（ほかから離れすぎた値）。箱の長さの 1.5 倍より外にある値 |

- 上の図の文字は `plt.text`（第８講で学ぶ）で書いている。今は読み方だけ分かればよい

## R-1 箱ひげ図の基本

- `plt.boxplot([データ1, データ2, …])` で、グループごとの箱が**左から順に**並ぶ
- データは**リストのリスト**（`[ ]` の中にリストを並べる）で渡す
- 演習「箱ひげ図の描画」のひな形は次の形

````python
japanese = [62, 75, 58, 81, 70, 66, 90, 55, 73, 68]
math = [45, 88, 62, 95, 30, 71, 84, 59, 77, 66]
english = [70, 72, 68, 75, 74, 69, 71, 73, 100, 70]

plt.boxplot([japanese, math, english])
plt.title('Box Plot')
plt.xlabel('Subject')
plt.ylabel('Score')
plt.show()
````

- 数学（2番目）は箱が長い＝点数がばらばら。英語（3番目）は箱が短い＝みんな同じくらい。英語の 100 点は外れ値（○）
- 箱が1つだけなら `plt.boxplot(japanese)` でもよい

## R-2 項目の名前

- 横軸の下には `1, 2, 3` と番号が付く。`plt.xticks([1, 2, 3], ['名前1', '名前2', '名前3'])` で名前に置き換える
- 番号と名前は**同じ個数**にする

````python
plt.boxplot([japanese, math, english])
plt.xticks([1, 2, 3], ['Japanese', 'Math', 'English'])
plt.title('Box Plot')
plt.xlabel('Subject')
plt.ylabel('Score')
plt.show()
````

- `plt.boxplot(..., tick_labels=['Japanese', 'Math', 'English'])` と書いても同じ（matplotlib 3.9 より古いと `labels=`）。どの環境でも動くので、この授業では `plt.xticks` を使う

▶ **演習**：`box_描画基礎001`

## R-3 箱の幅と塗りつぶし

- `widths=0.6` で箱の幅を変える（何も書かないと、箱の数に合わせて 0.15〜0.5 の間で決まる。`1` で隣の箱とくっつく）
- `patch_artist=True` で箱の中を**塗りつぶす**（何も書かないと白抜き）。塗る色は `boxprops={'facecolor': '色'}`

````python
plt.boxplot([japanese, math, english], widths=0.6, patch_artist=True, boxprops={'facecolor': 'lightblue'})
plt.xticks([1, 2, 3], ['Japanese', 'Math', 'English'])
plt.title('Box Plot')
plt.xlabel('Subject')
plt.ylabel('Score')
plt.show()
````

- `{'facecolor': 'lightblue'}` は**辞書**という書き方（`{'名前': 値}`）。ここでは「箱の中の色は lightblue」と読めばよい

▶ **演習**：`box_描画基礎002`

## R-4 平均を表示する

- 箱の中の線は**中央値**で、平均ではない。`showmeans=True` で**平均**を ▲ で表示する
- 外れ値があると、平均は中央値から外れ値の側へずれる

````python
plt.boxplot([japanese, math, english], showmeans=True)
plt.xticks([1, 2, 3], ['Japanese', 'Math', 'English'])
plt.title('Box Plot')
plt.xlabel('Subject')
plt.ylabel('Score')
plt.show()
````

▶ **演習**：`box_描画基礎003`

## R-5 補助線を引く

- 第５講の `plt.axhline(y=値)` が箱ひげ図でも使える。合格ラインや全体の平均を引く

````python
plt.boxplot([japanese, math, english])
plt.axhline(y=60, color='r', linestyle='--', label='Pass line')
plt.xticks([1, 2, 3], ['Japanese', 'Math', 'English'])
plt.title('Box Plot')
plt.xlabel('Subject')
plt.ylabel('Score')
plt.legend()
plt.show()
````

▶ **演習**：`box_描画基礎004`

## R-6 Excel のデータで箱ひげ図

## T-1 カラーマップで色をまとめて決める

- 色を1つずつ選ぶ代わりに、**カラーマップ**（決まった色の組み合わせ）から取り出せる
- `plt.get_cmap('Pastel1').colors` で、Pastel1 という色の組み合わせのリストが得られる
- 名前の例：`'Pastel1'`、`'Set2'`、`'tab10'`、`'Accent'`（[色の組み合わせの一覧](https://matplotlib.org/stable/users/explain/colors/colormaps.html#qualitative)）

````python
x = [100, 200, 300, 400, 500]
mylabel = ['Japanese', 'mathematics', 'science', 'society', 'English']

mycolor = plt.get_cmap('Pastel1').colors

plt.pie(x, startangle=90, counterclock=False, labels=mylabel, colors=mycolor)
plt.show()
````

- `pie_描画基礎007` のひな形は、`with mpl.rc_context(...)` で**そのグラフの間だけ**色の組み合わせを変える書き方。`colors=` を書くとそちらが優先されるので、`colors=` を消すと Pastel1 の色になる

## T-2 箱ごとに色を変える

- `patch_artist=True` を付けて描いた結果を変数（ここでは `bp`）に入れると、`bp['boxes']` で箱を1つずつ取り出せる
- for 文（第８講）で、箱ごとに色を塗る

````python
bp = plt.boxplot([japanese, math, english], patch_artist=True)
mycolor = ['lightcoral', 'lightgreen', 'lightblue']
for box, c in zip(bp['boxes'], mycolor):
    box.set_facecolor(c)

plt.xticks([1, 2, 3], ['Japanese', 'Math', 'English'])
plt.title('Box Plot')
plt.xlabel('Subject')
plt.ylabel('Score')
plt.show()
````

▶ **演習**（発展）：`box_描画基礎005`

## T-3 TeX で数式を書く

- マークダウンのセルで `$` で囲むと、数式になる：$y = ax^2 + b$、$\bar{x} = \frac{1}{n}\sum x_i$
- グラフの文字にも使える

````python
x = np.linspace(-2, 2, 50)
plt.plot(x, x ** 2, label='$y = x^2$')
plt.plot(x, x ** 3, label='$y = x^3$')
plt.title('$y = x^n$ のグラフ')
plt.legend()
plt.show()
````

- `np.linspace(-2, 2, 50)` は、-2 から 2 までを等間隔に分けた50個の数
