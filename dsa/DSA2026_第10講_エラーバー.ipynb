## AB-0 今日の完成形

5教科の**平均点**を点で、**平均 ± 標準偏差**を上下の線（エラーバー）で示したもの。平均と標準偏差は for でまとめて計算している。

1. **全部コピーして**自分の notebook のセルに貼り付け、実行する

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

df = pd.read_excel('dsa2026data.xlsx', sheet_name='exam')
subjects = ['国語', '数学', '理科', '社会', '英語']

means = []  # [1] 平均のリスト
stds = []  # [1] 標準偏差のリスト
for s in subjects:  # [2] 教科ごとにくり返す
    means.append(np.mean(df[s]))
    stds.append(np.std(df[s]))

plt.figure(figsize=(8, 5))
plt.errorbar(subjects, means, yerr=stds,  # [3] エラーバー
             fmt='o', markersize=10, capsize=5,  # [4] 見た目
             color='w', markeredgecolor='k', ecolor='k')
plt.ylim(0, 100)
plt.title('5教科の平均点（平均 ± 標準偏差）', fontsize=16)
plt.ylabel('点数', fontsize=14)
plt.show()
```

## AB-1 平均だけでは分からないこと

- 下の2クラスは、**平均がほぼ同じ**（68点くらい）。でも、点数の散らばり方は同じとは限らない
- 散らばりの大きさを1つの数で表したものが**標準偏差**（第５講）。エラーバーは、平均の点から上下に標準偏差の長さの線を引いたもの

```python
group1 = [66, 68, 70, 67, 69, 68, 71, 65, 70, 66]
group2 = [40, 95, 55, 85, 60, 90, 45, 80, 52, 78]
m = [np.mean(group1), np.mean(group2)]
s = [np.std(group1), np.std(group2)]
print('平均', m)
print('標準偏差', s)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.scatter([0] * 10, group1, alpha=0.5)
plt.scatter([1] * 10, group2, alpha=0.5)
plt.xticks([0, 1], ['1組', '2組'])
plt.xlim(-0.5, 1.5)
plt.ylim(0, 100)
plt.title('一人ひとりの点数')

plt.subplot(1, 2, 2)
plt.errorbar([0, 1], m, yerr=s, fmt='o', capsize=8, markersize=10)
plt.xticks([0, 1], ['1組', '2組'])
plt.xlim(-0.5, 1.5)
plt.ylim(0, 100)
plt.title('平均 ± 標準偏差（エラーバー）')

plt.tight_layout()
plt.show()
```

- 2組は点数がばらばら → 標準偏差が大きい → エラーバーが**長い**
- 1組はみんな同じくらい → 標準偏差が小さい → エラーバーが**短い**
- 多くのデータでは、エラーバーの範囲（平均 ± 標準偏差）に**全体のおよそ3分の2**の人が入る
- 論文などでは、エラーバーが標準偏差なのか別のもの（AC-1 の標準誤差など）なのかを、必ず図の説明に書く

## AB-2 リストを np.array にする

- 演習「エラーバー描画」では、リストを `np.array(リスト)` で **np.array**（numpy の配列）に変えて使う
- np.array にすると、`変数.mean()` や `変数.std()` で平均・標準偏差が求められる（`np.mean(変数)` と同じ結果）

```python
classA_score_np = np.array(classA_score)
classB_score_np = np.array(classB_score)

print(classA_score_np)
print(classB_score_np)
```

- リストは `[77, 69, 63, …]` と `,` 区切りで、np.array は `[77 69 63 …]` と空白区切りで表示される

## AB-5 エラーバーを描く

- `plt.errorbar(横の位置, 平均, yerr=エラーバーの長さ)` で描く。横の位置・平均・長さは、**同じ個数のリスト**にする
- `fmt='o'` で、点だけを描く（何も書かないと、点が線でつながる）

| オプション | 意味 |
|---|---|
| `yerr=` | エラーバーの長さ（上下それぞれ）。ふつうは標準偏差 |
| `fmt='o'` | 点の形（`'o'` 丸、`'s'` 四角、`'^'` 三角）。`'o-'` で線でつなぐ |
| `markersize=10` | 点の大きさ |
| `capsize=5` | エラーバーの端の横棒の長さ |
| `color='w'` | 点の色（`'w'` は白） |
| `markeredgecolor='k'` | 点の縁の色 |
| `ecolor='k'` | エラーバーの線の色 |

```python
myclass = ['classA', 'classB']
mean_score = [mean_A, mean_B]
errorbar = [std_A, std_B]

plt.figure(figsize=(3, 3))
plt.errorbar(myclass, mean_score, yerr=errorbar, fmt='o', capsize=5, markersize=10,
             color='w', markeredgecolor='k', ecolor='k')
plt.xlim(-0.5, 1.5)
plt.ylim(0, 100)
plt.show()
```

- 横軸が文字（`'classA'` など）のとき、位置は左から `0, 1` になる。`plt.xlim(-0.5, 1.5)` で左右に余白を作っている

## AB-6 for でまとめて計算して描く

- 教科が5つあると、平均と標準偏差を10回計算することになる → **for 文**でくり返す（第８講）
- 空のリストを2つ作り、教科ごとに `.append` していく

```python
df = pd.read_excel('dsa2026data.xlsx', sheet_name='exam')
subjects = ['国語', '数学', '理科', '社会', '英語']

means = []
stds = []
for s in subjects:
    means.append(np.mean(df[s]))
    stds.append(np.std(df[s]))
print(means)
print(stds)

plt.errorbar(subjects, means, yerr=stds, fmt='o', capsize=5)
plt.ylim(0, 100)
plt.ylabel('点数（平均 ± 標準偏差）')
plt.show()
```

- 数学はエラーバーが長い＝点数のばらつきが大きい（第７講の箱ひげ図でも、数学の箱は長かった）

# AC. 発展（できる人だけ）

## AC-1 標準誤差と信頼区間

- **標準偏差**：一人ひとりの点数のばらつき
- **標準誤差** ＝ 標準偏差 ÷ √人数：**平均値そのもの**のぶれ（もう一度同じ人数でテストをしたら、平均はどれくらい変わりそうか）。人数が多いほど小さくなる
- 平均 ± 1.96 × 標準誤差 が、おおよそ **95% 信頼区間**

```python
n = len(df)
ses = []
for sd in stds:
    ses.append(1.96 * sd / np.sqrt(n))

plt.errorbar(subjects, means, yerr=stds, fmt='o', capsize=5, label='平均 ± 標準偏差')
plt.errorbar(subjects, means, yerr=ses, fmt='o', capsize=5, color='r', label='95% 信頼区間')
plt.ylim(0, 100)
plt.ylabel('点数')
plt.legend()
plt.show()
```

- 信頼区間は標準偏差より短い。2つの教科の信頼区間が重なっていなければ、「平均点に差がある」と考える目安になる
