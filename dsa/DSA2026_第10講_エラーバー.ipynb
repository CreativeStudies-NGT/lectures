# AB. エラーバー

## AB-0 今日の完成形

5教科の**平均点**を点で、**平均 ± 標準偏差**を上下の線（エラーバー）で示したもの。平均と標準偏差は for でまとめて計算している。

1. **全部コピーして**自分の notebook のセルに貼り付け、実行する
1. まずは**色や大きさを変えて**、グラフのどこが変わるかを確かめる
1. そのあと部品（[1]〜[4]）の働きを1つずつ学ぶ（AB-1〜AB-6）

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

df = pd.read_excel('dsa2026data.xlsx', sheet_name='exam')
subjects = ['国語', '数学', '理科', '社会', '英語']

means = []  # [1] 平均のリスト
stds = []  # [1] 標準偏差のリスト
for s in subjects:  # [2] 教科ごとにくり返す
    means.append(np.mean(df[s]))
    stds.append(np.std(df[s]))

plt.figure(figsize=(8, 5))
plt.errorbar(subjects, means, yerr=stds,  # [3] エラーバー
             fmt='o', markersize=10, capsize=5,  # [4] 見た目
             color='w', markeredgecolor='k', ecolor='k')
plt.ylim(0, 100)
plt.title('5教科の平均点（平均 ± 標準偏差）', fontsize=16)
plt.ylabel('点数', fontsize=14)
plt.show()
```

# AC. 発展（できる人だけ）

## AC-1 標準誤差と信頼区間

- **標準偏差**：一人ひとりの点数のばらつき
- **標準誤差** ＝ 標準偏差 ÷ √人数：**平均値そのもの**のぶれ（もう一度同じ人数でテストをしたら、平均はどれくらい変わりそうか）。人数が多いほど小さくなる
- 平均 ± 1.96 × 標準誤差 が、おおよそ **95% 信頼区間**

```python
n = len(df)
ses = []
for sd in stds:
    ses.append(1.96 * sd / np.sqrt(n))

plt.errorbar(subjects, means, yerr=stds, fmt='o', capsize=5, label='平均 ± 標準偏差')
plt.errorbar(subjects, means, yerr=ses, fmt='o', capsize=5, color='r', label='95% 信頼区間')
plt.ylim(0, 100)
plt.ylabel('点数')
plt.legend()
plt.show()
```

- 信頼区間は標準偏差より短い。2つの教科の信頼区間が重なっていなければ、「平均点に差がある」と考える目安になる

## AC-3 ax を使った書き方

- この授業では `plt.` だけで書いているが、ネットの例や AI が書くコードでは `fig, ax = plt.subplots()` と `ax.` を使う書き方がよく出てくる
- 両 y 軸を `ax` で書くと次のようになる。`plt.ylabel` が `ax.set_ylabel` になるなど、名前が少し違う

```python
fig, ax1 = plt.subplots(figsize=(8, 5))
ax1.bar(df_rain['月'], df_rain['降水量'], color='skyblue')
ax1.set_ylabel('降水量 (mm)')

ax2 = ax1.twinx()
ax2.plot(df_rain['月'], df_rain['気温'], color='r', marker='o')
ax2.set_ylabel('気温 (℃)')
plt.show()
```

| plt の書き方 | ax の書き方 |
|---|---|
| `plt.title('…')` | `ax.set_title('…')` |
| `plt.xlabel('…')`、`plt.ylabel('…')` | `ax.set_xlabel('…')`、`ax.set_ylabel('…')` |
| `plt.xlim(0, 10)` | `ax.set_xlim(0, 10)` |
| `plt.bar(…)`、`plt.plot(…)` | `ax.bar(…)`、`ax.plot(…)`（同じ） |
