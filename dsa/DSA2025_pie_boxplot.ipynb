### カラーマップ
色の指定について、matplotlibはいくつかのカラーマップと呼ばれる色の組み合わせを持っている。（デフォルトはtab10）  
これを使うと、リストによる個別の色指定は不要になる。

[Colormapsの公式ドキュメント](https://matplotlib.org/stable/users/explain/colors/colormaps.html)  
のうち、Qualitative colormaps の部分が利用可能  
Pastel1, Pastel2, Paired, Accent, Dark2, Set1, Set2, Set3, tab10, tab20, tab20b, tab20c、及びその逆順 (_r 付加)

```python
# おまじないの追加
import matplotlib as mpl

with mpl.rc_context({'axes.prop_cycle': plt.cycler(color=plt.get_cmap('Set2').colors)}):
    label = ['Japanese', 'mathematics', 'science', 'society', 'English']

    # 色指定用リスト
#    mycolor = ["lightcoral","lightpink", "gold", "lightgreen", "lightblue"]
    
    x = [100, 200, 300, 400, 500]
    plt.pie(x, labels=label, startangle=90, autopct="%1.1f%%")

    # グラフ描画
    plt.show()
```

```python
# データ
jpn =  [10, 12, 14, 15, 18, 20, 34]
math =  [8, 9, 11, 13, 17, 19, 21]
sci = [14, 16, 18, 20, 23, 25, 27]

# 複数データがある場合は以下にまとめる
data = [jpn, math, sci]  # 3つのリストを要素とするリスト

# 箱ひげ図作成
plt.boxplot(data)

# タイトル、軸名設定
plt.title("箱ひげ図の描画", fontsize=16)
plt.xlabel("科目", fontsize=12)
plt.ylabel("得点", fontsize=12)

# グラフ表示
plt.show()
```

```python
# データ
jpn =  [10, 12, 14, 15, 18, 20, 34]
math =  [8, 9, 11, 13, 17, 19, 21]
sci = [14, 16, 18, 20, 23, 25, 27]

# 複数データがある場合は以下にまとめる
data = [jpn, math, sci]

# 箱ひげ図作成
plt.boxplot(data)

# 横軸の標記を番号から科目名に変更
plt.xticks([1,2,3], ('国語', '数学', '理科'))

# タイトル、軸名設定
plt.title("箱ひげ図の描画", fontsize=16)
plt.xlabel("科目", fontsize=12)
plt.ylabel("得点", fontsize=12)

# グラフ表示
plt.show()
```

```python
# データ
jpn =  [10, 12, 14, 15, 18, 20, 34]
math =  [8, 9, 11, 13, 17, 19, 21]
sci = [14, 16, 18, 20, 23, 25, 27]

# 複数データがある場合は以下にまとめる
data = [jpn, math, sci]

# 箱ひげの胴体設定
plt.gca().boxplot(data,
                  patch_artist=True,
                  showmeans=True,        # 平均値もプロット
                  notch=False,
                  flierprops = dict(marker="o", markerfacecolor='red', markersize=8, markeredgecolor='none'),   # 外れ値マーカー設定
                  boxprops={'facecolor': 'pink'}, 
                  medianprops={'color': 'k', 'linewidth':1}
                 )

# 横軸の標記を番号から科目名に変更
plt.xticks([1,2,3], ('国語', '数学', '理科'))

# タイトル、軸名設定
plt.title("箱ひげ図の描画", fontsize=16)
plt.xlabel("科目", fontsize=12)
plt.ylabel("得点", fontsize=12)

# グラフ表示
plt.show()
```
