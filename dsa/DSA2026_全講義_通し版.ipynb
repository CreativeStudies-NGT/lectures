## C-2 関数と引数

- `print(...)` や `len(...)` は**関数**。「**名前(引数)**」の形で使う。`( )` の中に渡す値を**引数**という
- `max`、`min` で最大値・最小値、`sum` で合計
- 平均は「合計 ÷ 個数」なので、`sum(score) / len(score)` で求められる

````python
score = [65, 80, 72, 90, 58]

print(max(score))
print(min(score))
print(sum(score))
print(sum(score) / len(score))
````

- 引数には2種類ある
  - **順番で意味が決まる引数**：`plt.plot(x, y)`（1番目が x、2番目が y）
  - **名前で指定する引数**（キーワード引数）：`color='b'`、`marker='s'` のように「名前=値」で書く

| 色 `color` | | マーカー `marker` | |
|---|---|---|---|
| `'r'` | 赤 | `'o'` | 丸 |
| `'g'` | 緑 | `'s'` | 四角 |
| `'b'` | 青 | `'^'` | 三角 |

````python
x = [1, 2, 3, 4, 5]

plt.plot(x, score, color='b', marker='s')
plt.show()
````

- `plt.` で始まるもの（`plot`、`title`、`show` …）は、「plt という道具箱の中の関数」

## F-1 凡例の場所・枠

関数には、ここまで使ったもの以外にも**いろいろなオプション（キーワード引数）**がある。F-1〜F-3 でよく使うものを紹介する。

### 凡例の場所：`plt.legend(loc='...')`

- 何も指定しないと `'best'`（線と重なりにくい場所を自動で選ぶ）。それでも線に重なるときは、場所を指定する

| `loc=` | 場所 |
|---|---|
| `'upper right'` | 右上 |
| `'upper left'` | 左上 |
| `'lower right'` | 右下 |
| `'lower left'` | 左下 |
| `'center'` | 中央 |

- 凡例の文字の大きさも `fontsize=` で変えられる（E-3 と同じ）

````python
plt.figure(figsize=(8, 5))
plt.plot(x, y1, marker='o', label='データ1')
plt.plot(x, y2, marker='s', label='データ2')
plt.legend(loc='upper left', fontsize=14)
plt.show()
````

### 凡例の場所を数値で指定する：`loc=(横, 縦)`

- グラフの枠の**左下を `(0, 0)`、右上を `(1, 1)`** として、凡例の**左下の角**を置く位置を割合で指定する
- 例：`loc=(0.05, 0.75)` は「横は左端から少し右、縦は上から4分の1」のあたり。線と重なる場合は、数値を少しずつ変えて、空いている場所を探す
- `frameon=False` を足すと、凡例の**枠が消える**（すっきり見える）

````python
plt.figure(figsize=(8, 5))
plt.plot(x, y1, marker='o', label='データ1')
plt.plot(x, y2, marker='s', label='データ2')
plt.legend(loc=(0.05, 0.75), frameon=False)
plt.show()
````

## F-2 線の種類・太さと短い書き方

### 線の種類・太さ：`linestyle=`、`linewidth=`

| 指定 | 書き方 | 例 |
|---|---|---|
| 線の種類 | `linestyle=` | `'-'` 実線、`'--'` 破線、`':'` 点線、`'-.'` 一点鎖線 |
| 線の太さ | `linewidth=` | `1`、`3`（数値。大きいほど太い） |

````python
plt.figure(figsize=(8, 5))
plt.plot(x, y1, color='r', linestyle='--', linewidth=3, label='データ1')
plt.plot(x, y2, color='g', linestyle=':', linewidth=3, label='データ2')
plt.legend()
plt.show()
````

### 短い書き方：`c=`、`ms=`、`ls=`、`lw=`

- よく使う4つには**短い名前**がある。意味はまったく同じ

| 長い書き方 | 短い書き方 |
|---|---|
| `color='r'` | `c='r'` |
| `markersize=10` | `ms=10` |
| `linestyle='--'` | `ls='--'` |
| `linewidth=3` | `lw=3` |

- 下のセルは、上のセルと**同じグラフ**になる。インターネットや生成AIのコードには短い書き方もよく出てくるので、読めるようにしておこう

````python
plt.figure(figsize=(8, 5))
plt.plot(x, y1, c='r', ls='--', lw=3, label='データ1')
plt.plot(x, y2, c='g', ls=':', lw=3, label='データ2')
plt.legend()
plt.show()
````

### マーカー・線・色を1つの文字列で指定する

- `plt.plot(x, y, 'o--r')` のように、x, y の後ろに**文字列を1つ**書くと、マーカー・線の種類・色をまとめて指定できる
- 中身は「マーカー」「線の種類」「色」の記号を並べたもの（順番は入れ替えてもよい。使わないものは省略できる）

| 文字列 | 意味 |
|---|---|
| `'o--r'` | 丸マーカー＋破線＋赤 |
| `'s:g'` | 四角マーカー＋点線＋緑 |
| `'^-b'` | 三角マーカー＋実線＋青 |
| `'o'` | 丸マーカーだけ（線なし。散布図のようになる） |

````python
plt.figure(figsize=(8, 5))
plt.plot(x, y1, 'o--r', label='データ1')
plt.plot(x, y2, 's:g', label='データ2')
plt.legend()
plt.show()
````

## F-3 目盛り・透明度・補助線

### 目盛りの文字を斜めにする：`plt.xticks(rotation=角度)`

- x 軸の文字が長くて重なるときに使う。角度は度（`20`、`45`、`90` など）
- `fontsize=` で目盛りの文字の大きさも変えられる

````python
city = ['新潟市', '長岡市', '上越市', '三条市', '柏崎市', '新発田市']
rain = [1845, 2212, 2755, 1916, 2117, 1852]

plt.figure(figsize=(8, 5))
plt.plot(city, rain, marker='o')
plt.xticks(rotation=45, fontsize=12)
plt.ylabel('年間降水量（mm）')
plt.show()
````

- 上のデータは、目盛りの説明用に作った例

### 透明度：`alpha=`

- `0`（完全に透明）〜 `1`（不透明）の数値。重なった部分が透けて見える

````python
plt.figure(figsize=(8, 5))
plt.plot(x, y1, marker='o', markersize=25, alpha=0.5)
plt.plot(x, y2, marker='o', markersize=25, alpha=0.5)
plt.xlim(0, 7)
plt.ylim(0, 14)
plt.show()
````

### 補助線：`plt.axhline(y=値)`、`plt.axvline(x=値)`

- `axhline` は**横線**（horizontal）、`axvline` は**縦線**（vertical）
- 例：平均値（C-2 の `sum / len`）に横線を引くと、平均より上か下かがひと目で分かる
- `c=`、`ls=`、`lw=` で色・線の種類・太さを指定できる（F-2 と同じ）

### 目盛り線：`plt.grid()`

- 値を読み取りやすくなる。`alpha=` で薄くすると、グラフの邪魔になりにくい

````python
plt.figure(figsize=(8, 5))
plt.plot(x, y1, marker='o')
plt.axhline(y=sum(y1) / len(y1), c='m', ls=':', lw=2)
plt.axvline(x=4, c='gray', ls='--')
plt.grid(alpha=0.3)
plt.show()
````

- 関数名の後ろ（`plt.legend(` の `(` の後など）にカーソルを置いて **`Shift + Tab`** を押すと、使い方の説明が出る（英語）。Mac も同じ `Shift + Tab`

## G-3 点の大きさ

- 散布図の点の大きさは **`s=`** で指定する（**`markersize=` ではない**）
- `s=` は点の**面積**。何も書かないときは `s=36` くらい。`s=100` で折れ線の `markersize=10` と同じくらい
- 面積なので、`s=100` → `s=400`（4倍）で、見た目の幅は2倍になる

````python
plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', color='b', s=100)
plt.show()
````

### 点ごとに大きさを変える：`s=リスト`

- `s=` に**リスト**を渡すと、1つ目の点にはリストの1つ目、2つ目の点には2つ目…の大きさが付く
- x・y に続く**3つ目の情報**（人口、売上など）を、点の大きさで見せられる
- リストの個数は、**点の個数（x・y の個数）と同じ**にする

````python
markersize_list = [100, 200, 300, 400, 500, 600]

plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', color='b', s=markersize_list)
plt.show()
````

- 演習の問題文ではリストの名前が `markersize_list` だが、渡すのは **`s=markersize_list`**。`markersize=` と書くとエラーになる：

## I-1 for 文で色のリストを作る

- 「y が 6 以上なら赤、そうでなければ灰色」のように、**データに応じて**色を決めたいときは、for 文で1つずつ `append` する
- `for 変数 in リスト:` の次の行は**字下げ**（スペース4つ）する。字下げした行が、リストの要素の数だけくり返される

````python
mycolors = []
for value in y:
    if value >= 6:
        mycolors.append('r')
    else:
        mycolors.append('gray')

print(mycolors)

plt.figure(figsize=(8, 5))
plt.scatter(x, y, marker='o', s=100, color=mycolors)
plt.axhline(y=6, color='k', linestyle='--')
plt.show()
````

# K. グラフの保存

## K-1 画像ファイルに保存する

- `plt.savefig('ファイル名.png')` で保存する
- **`plt.show()` より前**に書く（後ろに書くと**白紙**が保存される）
- 保存先は、この notebook と**同じ場所**（いつもの場所）。Jupyter の画面左のファイル一覧に出てくる

````python
plt.figure(figsize=(8, 5))
plt.scatter(number, score, s=100, zorder=2)
plt.axhline(y=np.mean(score), color='r', linestyle='--', label='平均')
plt.ylim(0, 100)
plt.xlabel('出席番号')
plt.ylabel('点数')
plt.legend()
plt.savefig('第５講_点数.png')
plt.show()
````

## K-2 細かさと背景の透明化

- `dpi=300` で細かく保存する（何も書かないと 100）。数が大きいほど細かく、ファイルも大きくなる
- `transparent=True` で背景を透明にする。色の付いたスライドに貼るときに使う

## K-3 保存の形式（png・svg・eps）

ファイル名の最後（**拡張子**）で形式が決まる。

| 形式 | 種類 | 向いている使い方 | 注意 |
|---|---|---|---|
| `.png` | ラスター（点の集まり） | PowerPoint・Word・Web に貼る。**迷ったらこれ** | 拡大すると荒れる。`dpi=300` で細かく保存する |
| `.svg` | ベクター（線と文字の情報） | 拡大しても荒れない。Illustrator などで線や文字を直せる | 古い Office では貼れないことがある |
| `.eps` | ベクター | 論文・印刷物（LaTeX や印刷所への入稿） | **透明が使えない**（`transparent=True` や `alpha=` は無視される） |

- **ラスター**は写真と同じ「点（画素）の集まり」、**ベクター**は「ここからここへ線を引く」という情報。ベクターは拡大しても線がなめらか

````python
plt.figure(figsize=(8, 5))
plt.scatter(number, score, s=100, zorder=2)
plt.axhline(y=np.mean(score), color='r', linestyle='--')
plt.xlabel('出席番号')
plt.ylabel('点数')
plt.savefig('第５講_点数.svg')
plt.show()
````

## N-1 棒グラフの基本

### 折れ線グラフ・散布図との違い

| | 折れ線グラフ | 散布図 | 棒グラフ |

## N-4 外枠線と網掛け

- `edgecolor=` で棒の**外枠線**の色、`linewidth=` でその太さ
- `hatch=` で**網掛け**（模様）。白黒で印刷しても区別できる

| 書き方 | 模様 | 書き方 | 模様 |
|---|---|---|---|
| `'/'` | 斜線 | `'x'` | バツ |
| `'\\'` | 逆向きの斜線 | `'.'` | 点 |
| `'-'` | 横線 | `'o'` | 丸 |
| `'+'` | 格子 | `'*'` | 星 |

- 同じ記号を重ねる（`'//'`、`'///'`）と、模様が細かくなる

````python
plt.figure(figsize=(8, 5))
plt.bar(x, y, color='w', edgecolor='k', linewidth=2, hatch='//')
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.show()
````

▶ **演習**：`bar_描画基礎003`

## N-6 平均の補助線

- 前回の復習。平均は `np.mean(リスト)`、横の補助線は `plt.axhline(y=値)`
- 平均の値を**自分で計算して書き込まない**。`np.mean(y7)` と書けば、データを変えても線が自動で動く

````python
plt.figure(figsize=(8, 5))
plt.bar(x7, y7)
plt.axhline(y=np.mean(y7), color='r', linestyle='--', label='平均')
plt.title('Bar Chart')
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.legend()
plt.show()
````

▶ **演習**：`bar_描画基礎005`

## N-7 横棒グラフ

- `plt.bar` を `plt.barh` に変えると、棒が**横向き**になる（h は horizontal＝水平）
- 横向きになると、項目は**縦軸**、量は**横軸**に移る。だから `xlabel` と `ylabel` の中身を**入れ替える**
- 項目の名前が長いとき（都市名など）は、横棒グラフのほうが読みやすい

````python
plt.figure(figsize=(8, 5))
plt.barh(x7, y7)
plt.title('Bar Chart')
plt.xlabel('Total Bill')
plt.ylabel('Day')
plt.show()
````

▶ **演習**：`bar_描画基礎006`

- 横棒グラフでは、量が横軸にあるので、平均の補助線は**縦の線** `plt.axvline(x=値)` になる

````python
plt.figure(figsize=(8, 5))
plt.barh(x7, y7)
plt.axvline(x=np.mean(y7), color='r', linestyle='--', label='平均')
plt.title('Bar Chart')
plt.xlabel('Total Bill')
plt.ylabel('Day')
plt.legend()
plt.show()
````

▶ **演習**：`bar_描画基礎007`

# O. ヒストグラム

**ヒストグラム**は、データを**区間に分けて、それぞれの区間にいくつ入るかを数えた**グラフ。点数が**どのあたりに集まっているか**（分布）を見るときに使う。

| | 棒グラフ | ヒストグラム |
|---|---|---|
| 関数 | `plt.bar(x, y)` | `plt.hist(データ)` |
| 渡すもの | 項目（x）と高さ（y）の**2つ** | データ**1つだけ**（数えるのは matplotlib） |
| 横軸 | 項目（曜日、都市など） | **数値の区間**（0〜10点、10〜20点 …） |
| 縦軸 | 量 | **個数**（人数など） |
| 棒のすき間 | あける | **あけない**（区間がつながっているため） |

## O-1 ヒストグラムの基本

- `plt.hist(データ)` だけで描ける。区間の分け方と個数の数え方は、matplotlib が自動で決める

````python
score = [45, 52, 58, 61, 63, 65, 67, 68, 70, 71, 72, 73, 74, 75, 75,
         76, 77, 78, 79, 80, 81, 82, 84, 85, 86, 88, 90, 92, 95, 98]

plt.figure(figsize=(8, 5))
plt.hist(score)
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.show()
````

- 縦軸は**人数**。たとえば一番高い棒は「その区間の点数の人がいちばん多い」という意味
- これが演習「ヒストグラムの描画」のひな形

## O-2 区切りの数

- `bins=` で**区間の数**を決める（bin＝箱）。何も書かないと `10`
- 少なすぎると形が分からず、多すぎるとガタガタになる。データの数が30くらいなら 5〜10 が目安

````python
plt.figure(figsize=(8, 5))
plt.hist(score, bins=5)
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.show()
````

▶ **演習**：`hist_描画基礎001`

## O-3 色と枠線

- 棒グラフと同じく `color=`、`edgecolor=`。ヒストグラムは棒がくっついているので、**枠線を付けると区切りが見やすい**

````python
plt.figure(figsize=(8, 5))
plt.hist(score, bins=5, color='lightgreen', edgecolor='k')
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.show()
````

▶ **演習**：`hist_描画基礎002`

## O-4 区切りの範囲

- 何も書かないと、区間は**データの最小値から最大値まで**になる（上の図は 45〜98 点）。区切りが 55.6 点のような半端な値になる
- `range=(0, 100)` と書くと 0〜100 点を区切る。`bins=10` と合わせると、**10点刻み**（0〜10, 10〜20, …）になる

````python
plt.figure(figsize=(8, 5))
plt.hist(score, bins=10, range=(0, 100), edgecolor='k')
plt.xticks(range(0, 101, 10))
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.show()
````

- `plt.xticks(range(0, 101, 10))` は、横軸の目盛りを 0, 10, 20, …, 100 にする（なくてもよい）

▶ **演習**：`hist_描画基礎003`

## O-5 平均の補助線

- ヒストグラムでは点数が**横軸**にあるので、平均は**縦の線** `plt.axvline(x=np.mean(データ))`（横棒グラフのときと同じ）

````python
plt.figure(figsize=(8, 5))
plt.hist(score, bins=10, range=(0, 100), edgecolor='k')
plt.axvline(x=np.mean(score), color='r', linestyle='--', label='平均')
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.legend()
plt.show()
````

▶ **演習**：`hist_描画基礎004`

## O-6 2つのグループを重ねる

- 前々回の散布図と同じく、`plt.hist` を**2回**書くと重なる
- 後ろが隠れないように `alpha=0.5`（透明度）、どちらがどちらか分かるように `label=` と `plt.legend()`
- 2つの `bins=` と `range=` は**同じ値**にする（区間がずれると比べられない）

````python
score_b = [38, 47, 55, 58, 60, 62, 64, 66, 67, 69,
           70, 71, 72, 74, 76, 78, 80, 83, 87, 91]

plt.figure(figsize=(8, 5))
plt.hist(score, bins=10, range=(0, 100), alpha=0.5, label='A組')
plt.hist(score_b, bins=10, range=(0, 100), alpha=0.5, label='B組')
plt.title('Histogram')
plt.xlabel('Score')
plt.ylabel('Number of students')
plt.legend()
plt.show()
````

▶ **演習**：`hist_描画基礎005`

## P-4 CSV ファイルを読む（文字化けしたとき）

**CSV ファイル**（`.csv`）は、表をカンマ `,` 区切りの文字だけで書いたファイル。e-Stat などからダウンロードするデータは CSV のことが多い。

- 読み方は `pd.read_csv('ファイル名')`（Excel の `read_excel` とほぼ同じ）
- ただし CSV は**文字コード**（文字を数字に置き換える約束）がファイルによって違い、**日本語が読めないことがある**
  - 日本語版の Excel で「CSV（コンマ区切り）」として保存したファイルや、e-Stat の CSV は **Shift_JIS**（Windows では `cp932`）
  - pandas は何も書かないと **UTF-8** として読むので、Shift_JIS のファイルではエラーになる

まず、練習用に Shift_JIS の CSV ファイルを作る（いつもの場所に `ramen_sjis.csv` ができる）。

````python
df_ramen.to_csv('ramen_sjis.csv', encoding='cp932', index=False)
````

````python
pd.read_csv('ramen_sjis.csv')
````

- `UnicodeDecodeError: 'utf-8' codec can't decode byte …` が出る。これが「文字コードが違う」というエラー
- `encoding='cp932'` を付けると読める

````python
pd.read_csv('ramen_sjis.csv', encoding='cp932').head()
````

| 困ったこと | 原因 | 直し方 |
|---|---|---|
| `read_csv` で `UnicodeDecodeError: 'utf-8' codec …` | CSV が Shift_JIS | `pd.read_csv('ファイル名', encoding='cp932')` |
| `encoding='cp932'` でも `UnicodeDecodeError: 'cp932' codec …` | CSV が UTF-8 | `encoding=` を外す（または `encoding='utf-8'`） |
| Python で書き出した CSV を Excel で開くと**文字化け** | Excel は UTF-8 の CSV を正しく判別できないことがある | 書き出すときに `df.to_csv('ファイル名', encoding='utf-8-sig', index=False)` |
| グラフの日本語が **□□□** になる | 文字コードではなく**フォント**の問題 | `import japanize_matplotlib` を実行する |

- この授業では、**CSV を Excel で開いて「Excel ブック（.xlsx）」で保存し直し、`read_excel` で読む**ことに統一している（M-4）。xlsx なら文字コードを気にしなくてよい。ここの `encoding=` は、それができないときのための方法
- `shift_jis` ではなく `cp932` と書くのは、`①` や `髙` など Windows で増えた文字も読めるようにするため

# R. 箱ひげ図

**箱ひげ図**は、いくつかのグループの**ばらつき**を並べて比べるグラフ。ヒストグラムを横から見て、5つの数にまとめたもの。

## R-0 箱ひげ図の読み方

````python
data = [50, 52, 58, 61, 63, 65, 67, 68, 70, 71, 72, 73, 74, 75, 75,
        76, 77, 78, 79, 80, 81, 82, 84, 85, 86, 88, 90, 92, 95, 20]

plt.figure(figsize=(6, 6))
plt.boxplot(data)
plt.text(1.1, np.max(data[:-1]), '最大（ひげの上の端）', va='center')
plt.text(1.1, np.percentile(data, 75), '上から 1/4 の位置（箱の上）', va='center')
plt.text(1.1, np.median(data), '中央値（箱の中の線）', va='center')
plt.text(1.1, np.percentile(data, 25), '下から 1/4 の位置（箱の下）', va='center')
plt.text(1.1, np.min(data[:-1]), '最小（ひげの下の端）', va='center')
plt.text(1.1, 20, '外れ値（離れすぎた値）', va='center')
plt.xlim(0.5, 2.5)
plt.ylabel('点数')
plt.show()
````

| 部分 | 意味 |
|---|---|
| 箱の中の線 | **中央値**（大きさの順に並べて真ん中の値） |
| 箱 | 真ん中の半分（50%）の人が入る範囲。箱が**長いほどばらつきが大きい** |
| ひげ | 外れ値を除いた最小から最大まで |
| ○ | **外れ値**（ほかから離れすぎた値）。箱の長さの 1.5 倍より外にある値 |

# U. for 文：同じ処理をくり返す

12か月の降水量を1つずつ表示したいとき、`print` を12回書くのは大変。**for 文**を使うと、リストの要素を1つずつ取り出して、同じ処理をくり返せる。

## U-1 for 文の基本

```python
for 変数 in リスト:
    くり返す処理
```

- リストの**先頭から順に**要素を1つ取り出して `変数` に入れ、字下げした部分を実行する。これを最後の要素までくり返す

````python
fruits = ['りんご', 'みかん', 'ぶどう']

for f in fruits:
    print(f)
````

- 1回目は `f` に `'りんご'`、2回目は `'みかん'`、3回目は `'ぶどう'` が入る
- 変数の名前（ここでは `f`）は自由に決めてよい。`k`、`x`、`fruit` など
- 数と文字が混ざったリストでも、同じように取り出せる

````python
mylist = [11, 22, 'abc', 33]

for k in mylist:
    print(k)
````

▶ **演習**：`for文基礎001`

## U-2 コロンと字下げ

- 1行目の最後に **`:`（コロン）** を付ける
- くり返す処理は、行の先頭に**半角スペース4つ**を入れて**字下げ（インデント）**する。Jupyter では `:` の後で改行すると自動で字下げされる
- 字下げした行だけがくり返される。字下げをやめた行は、くり返しが**終わった後に1回だけ**実行される

````python
for f in fruits:
    print(f + 'が好き')
    print('---')
print('おわり')
````

| エラーの最後の行 | 原因 | 直し方 |
|---|---|---|
| `SyntaxError: expected ':'` | 1行目の最後の `:` を忘れた | `:` を付ける |
| `IndentationError: expected an indented block` | くり返す処理を字下げしていない | 行の先頭にスペース4つ |
| `IndentationError: unexpected indent` | 字下げしなくてよい行を字下げした | 余分なスペースを消す |

````python
rng = np.random.default_rng(348309421)
data = rng.normal(65, 15, 20)
mean = np.mean(data)

p = []
for k in data:
    if k > mean:
        p.append('m')
    else:
        p.append('c')

plt.scatter(range(len(data)), data, c=p)
plt.axhline(y=mean, color='m', linestyle=':')
plt.show()
````

▶ **演習**：`for~if文応用001`（ひな形の `data.mean()` は `np.mean(data)` と同じ意味）

## AB-1 平均だけでは分からないこと

- 下の2クラスは、**平均がほぼ同じ**（68点くらい）。でも、点数の散らばり方は同じとは限らない
- 散らばりの大きさを1つの数で表したものが**標準偏差**（第５講）。エラーバーは、平均の点から上下に標準偏差の長さの線を引いたもの

````python
group1 = [66, 68, 70, 67, 69, 68, 71, 65, 70, 66]
group2 = [40, 95, 55, 85, 60, 90, 45, 80, 52, 78]
m = [np.mean(group1), np.mean(group2)]
s = [np.std(group1), np.std(group2)]
print('平均', m)
print('標準偏差', s)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.scatter([0] * 10, group1, alpha=0.5)
plt.scatter([1] * 10, group2, alpha=0.5)
plt.xticks([0, 1], ['1組', '2組'])
plt.xlim(-0.5, 1.5)
plt.ylim(0, 100)
plt.title('一人ひとりの点数')

plt.subplot(1, 2, 2)
plt.errorbar([0, 1], m, yerr=s, fmt='o', capsize=8, markersize=10)
plt.xticks([0, 1], ['1組', '2組'])
plt.xlim(-0.5, 1.5)
plt.ylim(0, 100)
plt.title('平均 ± 標準偏差（エラーバー）')

plt.tight_layout()
plt.show()
````

- **標準誤差** ＝ 標準偏差 ÷ √人数：**平均値そのもの**のぶれ（もう一度同じ人数でテストをしたら、平均はどれくらい変わりそうか）。人数が多いほど小さくなる
- 平均 ± 1.96 × 標準誤差 が、おおよそ **95% 信頼区間**

````python
n = len(df)
ses = []
for sd in stds:
    ses.append(1.96 * sd / np.sqrt(n))

plt.errorbar(subjects, means, yerr=stds, fmt='o', capsize=5, label='平均 ± 標準偏差')
plt.errorbar(subjects, means, yerr=ses, fmt='o', capsize=5, color='r', label='95% 信頼区間')
plt.ylim(0, 100)
plt.ylabel('点数')
plt.legend()
plt.show()
````

- 信頼区間は標準偏差より短い。2つの教科の信頼区間が重なっていなければ、「平均点に差がある」と考える目安になる

## AK-1 関数を作る：def と return

- 関数は「材料（**引数**）を入れると、結果を返してくれる箱」。`print()`、`len()`、`np.mean()` も関数
- `def 名前(引数):` で自分の関数を作る。下の行は**字下げ**して書き、`return` で結果を返す（for・if と同じく `:` と字下げ）
- 作っただけでは何も起きない。`名前(値)` と書いて**使う（呼び出す）**と、はじめて動く

````python
def tax_in(price):
    return round(price * 1.1)

print(tax_in(100))
print(tax_in(250))
````

- `tax_in(100)` と書くと、`price` に 100 が入って `round(100 * 1.1)` が計算され、110 が返ってくる
- `round()` は四捨五入（小数点以下を丸める）

▶ **演習**：`関数01`（入力した値の3倍を返す）

## AK-2 引数を2つにする

- 引数は `,` で区切っていくつでも書ける。呼び出すときも同じ順番で値を入れる

````python
def rect_area(a, b):
    return a * b

print(rect_area(3, 5))
print(rect_area(10, 2))
````

- `x ** n` は「x の n 乗」（例：`2 ** 3` は 8）

▶ **演習**：`関数02`（x の n 乗を返す）

## AK-3 結果を2つ返す

- `return` の後に `,` で区切って書くと、結果を2つ返せる。受け取るときは `変数1, 変数2 = 関数(…)`

````python
def min_max(values):
    return min(values), max(values)

print(min_max([62, 75, 58, 81, 70]))

low, high = min_max([62, 75, 58, 81, 70])
print('最低点', low)
print('最高点', high)
````

- そのまま `print` すると `(58, 81)` と丸かっこでまとめて表示される
- 2つの変数で受け取ると、`low` に 58、`high` に 81 が入る

▶ **演習**：`関数03`（和と差を返す）、`関数04`（2つの変数で受け取る）

## AK-4 色のリストを返す関数

- 第９講の「for と if で色のリストを作る」処理を、関数にまとめる。データと境目を変えても、同じ関数を使い回せる
- 関数の中の for・if は、さらに字下げする（スペース4つずつ）

````python
def get_mycolor(values, border):
    mycolor = []
    for v in values:
        if v >= border:
            mycolor.append('tomato')
        else:
            mycolor.append('gray')
    return mycolor

scores = [62, 75, 58, 81, 70, 66, 90, 55, 73, 68]
print(get_mycolor(scores, 70))

plt.bar(range(len(scores)), scores, color=get_mycolor(scores, 70))
plt.axhline(y=70, color='k', linestyle='--')
plt.show()
````

- `return` は for の外（`for` と同じ字下げ）に書く。for の中に書くと、1回目で関数が終わってしまう
- 「上限と下限の間」は `if lim1 < v < lim2:` と書ける

▶ **演習**：`関数05`（上限と下限の間だけ色を変える）

# AL. 内包表記

## AL-1 for でリストを作る処理を1行で

- **内包表記**：for でリストを作る処理を、`[ ]` の中に1行で書いたもの。下の2つは同じ結果

````python
prices = [100, 250, 400]

result = []
for p in prices:
    result.append(tax_in(p))
print(result)

print([tax_in(p) for p in prices])
````

- 読み方：`[tax_in(p) for p in prices]` は「prices から p を1つずつ取り出し、`tax_in(p)` を計算して並べたリスト」
- **後ろから読む**とわかりやすい：`for p in prices`（取り出し方）→ 前の `tax_in(p)`（何を並べるか）
- `range(-10, 10)` の中身をリストで見るには、`list(range(-10, 10))` でもよい

````python
print([k for k in range(-10, 10)])
print(list(range(-10, 10)))
````

▶ **演習**：`内包表記01`

## AL-2 計算したリストを作る

- 前の部分に計算を書くと、計算した結果のリストになる
- np.array にすれば、内包表記を使わなくても全部に同じ計算ができる（第10講）

````python
print([k * 2 for k in range(5)])
print(np.array(range(5)) * 2)
````

▶ **演習**：`内包表記02`（2乗する。内包表記と np.array の2通り）

## AL-3 条件に合うものだけ取り出す

- 後ろに `if 条件` を付けると、条件に合うものだけが残る
- `%` は割り算の**余り**。`k % 2 == 0` は「2で割り切れる」＝偶数、`k % 2 == 1` は奇数

````python
scores = [62, 75, 58, 81, 70, 66, 90, 55, 73, 68]
print([s for s in scores if s >= 70])

print([k for k in range(10) if k % 2 == 0])
````

▶ **演習**：`内包表記03`（奇数だけ取り出す）

## AL-4 条件で値を変える

- 前の部分に `A if 条件 else B` を書くと、条件に合えば A、合わなければ B を並べる
- **AL-3 との違い**：AL-3 は `if` が後ろ（取り出す）。AL-4 は `if … else` が前（すべて残して、値を変える）
- AK-4 の `get_mycolor` の中身（for・if・else の7行）は、これで1行になる（AK-0 の `get_color`）

````python
scores = [62, 75, 58, 81, 70, 66, 90, 55, 73, 68]
print(['tomato' if s >= 70 else 'gray' for s in scores])

rng = np.random.default_rng()
dice = rng.integers(1, 7, 10)
print(dice)
print(['大' if d >= 4 else '小' for d in dice])
````

- `rng.integers(1, 7, 10)` は、1以上7未満の整数を10個（サイコロを10回振る）。実行するたびに変わる

## AN-1 lambda（無名関数）

## AO-1 回帰とは

- 散布図の点の並びに、**いちばん合う直線**を引くこと。直線の式は **y = 切片 + 傾き × x**
  - **切片**：x が 0 のときの y
  - **傾き**：x が 1 増えると、y がどれだけ増えるか
- 「いちばん合う」＝ 点と直線の**縦方向のずれ（の2乗）の合計が最小**になる直線（最小二乗法。`ols` は Ordinary Least Squares の略）
- x が1つのときを**単回帰**、2つ以上のときを**重回帰**（第15講の発展）という

````python
df_tip = pd.read_excel('dsa2026data.xlsx', sheet_name='tip_data')
df_tip.head()
````

````python
plt.scatter(df_tip['total_bill'], df_tip['tip'], alpha=0.5)
plt.xlabel('食事代 (USD)')
plt.ylabel('チップ (USD)')
plt.show()
````

- 右上がり：食事代が高いほど、チップも多い傾向がある。この傾向を直線で表す

## AO-3 回帰直線を求める

- `ols('y の列名 ~ x の列名', data=データフレーム).fit()` と書く
  - `~` の**左が y**（予測したいもの）、**右が x**（手がかり）。`'tip ~ total_bill'` は「チップを食事代で説明する」
  - `.fit()` で計算する。結果を `res` に入れておく
- 列名に空白や記号があると使えない。Excel の段階で短い列名にしておく（第13講）

````python
res = ols('tip ~ total_bill', data=df_tip).fit()
print(res.params)
````

## AO-4 切片と傾きを読む

- `res.params` の `Intercept` が**切片**、`total_bill` が**傾き**
- この例では y = 0.92 + 0.105 × x
  - 傾き 0.105：「食事代が1ドル増えると、チップは約0.1ドル増える」＝ チップはだいたい食事代の1割
  - 切片 0.92：「食事代0ドルなら、チップ0.92ドル」。実際には食事代0の会計は無いので、**切片はあまり意味を持たないことが多い**
- 1つずつ取り出すときは `res.params['Intercept']`、`res.params['total_bill']`

````python
a = res.params['Intercept']
b = res.params['total_bill']
print('切片', a)
print('傾き', b)
````

## AO-5 決定係数

- **決定係数**（R²、0〜1）：直線がどれくらい点に合っているか。1 に近いほどよく合う
- 目安：0.5 以上ならまあまあ合っている、0.2 より小さければ直線ではほとんど説明できていない（分野によって目安は違う）

````python
print('決定係数', res.rsquared)
````

- 0.46 くらい：チップの多さの半分くらいは食事代で説明できる。残りは人数・曜日・人の気前など、ほかの理由

# AP. 発展（できる人だけ）

## AP-1 多項式回帰（曲線を当てはめる）

- `I(x ** 2)` を足すと2次式（放物線）：y = 切片 + 係数1 × x + 係数2 × x²
- 一人あたりGDPと平均寿命：GDP が低いうちは急に伸び、高くなると伸びが鈍る → 直線より曲線が合う

````python
df_wb = pd.read_excel('dsa2026data.xlsx', sheet_name='worldbank')
df_wb = df_wb.rename(columns={'一人あたりGDP': 'gdp', '出生時平均寿命': 'life'})

res1 = ols('life ~ gdp', data=df_wb).fit()
res2 = ols('life ~ gdp + I(gdp ** 2)', data=df_wb).fit()
print('直線の決定係数', res1.rsquared)
print('2次式の決定係数', res2.rsquared)

x = np.linspace(df_wb['gdp'].min(), df_wb['gdp'].max(), 100)
plt.scatter(df_wb['gdp'], df_wb['life'])
plt.plot(x, res1.predict(pd.DataFrame({'gdp': x})), color='gray', label='直線')
plt.plot(x, res2.predict(pd.DataFrame({'gdp': x})), color='r', label='2次式')
plt.xlabel('一人あたりGDP (USD)')
plt.ylabel('平均寿命 (歳)')
plt.legend()
plt.show()
````

- `df.rename(columns={'元の名前': '新しい名前'})` で列名を短くしている
- `np.linspace(最小, 最大, 100)` は、最小から最大までを100等分した数（なめらかな曲線を描くため）
- 次数を上げすぎると、点にぴったり合っても予測には役に立たない曲線になる（過学習）

## AR-1 第３回課題のひな形

1. 興味のあるデータを e-Stat などから探す（**2つの量の関係**を見られるもの）
1. 第13講の手順で Excel を整え、notebook と同じフォルダ（いつもの場所）に置く
1. 下のセルの `ファイル名`・`シート名`・`列名`（`x_name`、`y_name`）を書き換える
1. 散布図と回帰直線を見て、**傾き**と**決定係数**を言葉で説明する

- 下は `accident` シート（交通事故件数と負傷者数、2000〜2023年）で書いた例

````python
df = pd.read_excel('dsa2026data.xlsx', sheet_name='accident')
x_name = '交通事故件数'
y_name = '負傷者数'

res = ols(f'{y_name} ~ {x_name}', data=df).fit()
print(res.params)
print('決定係数', res.rsquared)

plt.figure(figsize=(8, 5))
plt.scatter(df[x_name], df[y_name], zorder=2)
plt.plot(df[x_name], res.predict(), color='r', zorder=3)
plt.title(f'{y_name} と {x_name}（決定係数 {res.rsquared:.2f}）')
plt.xlabel(x_name + '（件）')
plt.ylabel(y_name + '（人）')
plt.show()
````

- 書き換えるのは、上の3行（`sheet_name`、`x_name`、`y_name`）と、軸の単位だけ
- 傾き 約1.26：「事故が1件増えると、負傷者は約1.3人増える」。決定係数は 1 に近い（ほぼ直線に並ぶ）

## AS-2 重回帰と標準化

- x を2つ以上使う：`'tip ~ total_bill + size'`（食事代と人数でチップを説明する）

````python
df_tip = pd.read_excel('dsa2026data.xlsx', sheet_name='tip_data')
res2 = ols('tip ~ total_bill + size', data=df_tip).fit()
print(res2.params)
print('決定係数', res2.rsquared)
````

- `total_bill` の係数：人数が同じなら、食事代が1ドル増えるとチップが約0.09ドル増える
- `size` の係数：食事代が同じなら、1人増えるとチップが約0.19ドル増える
- 「食事代と人数、どちらが効いているか」は、単位（ドルと人）が違うのでこのままでは比べられない
- 各列を**標準化**（平均0・標準偏差1）してから回帰すると、係数の大きさで比べられる

````python
cols = ['tip', 'total_bill', 'size']
df_z = (df_tip[cols] - df_tip[cols].mean()) / df_tip[cols].std()

res3 = ols('tip ~ total_bill + size', data=df_z).fit()
print(res3.params)
print('決定係数', res3.rsquared)
````

- 標準化した係数は `total_bill` の方が大きい → チップには食事代の方が強く効いている
- 決定係数は標準化しても変わらない
- 切片の `2.775558e-16` は「2.78 × 10 の −16 乗」＝ほぼ 0（標準化すると切片は 0 になる）
