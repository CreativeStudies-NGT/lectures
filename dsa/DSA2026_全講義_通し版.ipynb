# 第４講
## 散布図：複数系列・凡例
- `plt.scatter` を2回書くと重なる。`label=` と `plt.legend()` で凡例

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

height_a = [160, 165, 170, 175, 180]
weight_a = [52, 58, 61, 68, 72]
height_b = [158, 163, 168, 172, 178]
weight_b = [48, 50, 57, 60, 66]

plt.scatter(height_a, weight_a, label='Aクラス')
plt.scatter(height_b, weight_b, marker='s', label='Bクラス')
plt.xlabel('身長 (cm)')
plt.ylabel('体重 (kg)')
plt.legend()
plt.show()
```

## 色とサイズのリスト：3つ目の情報を見せる
- `color=` や `s=` に**リスト**を渡すと、点ごとに色・大きさが変わる
- 下の例は「年齢」を点の大きさで表している（x・y に続く3つ目の情報）
- リストは**手で書けばよい**（点の数と同じ個数にする）

```python
age = [18, 25, 40, 33, 60]
size = [180, 250, 400, 330, 600]
colors = ['r', 'b', 'b', 'b', 'r']

plt.scatter(height_a, weight_a, s=size, color=colors)
plt.xlabel('身長 (cm)')
plt.ylabel('体重 (kg)')
plt.show()
```

## 重なり：zorder と alpha
- `alpha=` は透明度（0 で透明〜1 で不透明）。点が重なっても下が見える
- `zorder=` は重ねる順番（大きいほど上）

```python
plt.plot(height_a, weight_a, color='gray', zorder=1)
plt.scatter(height_a, weight_a, s=size, color='orange', alpha=0.6, zorder=2)
plt.show()
```

## リストの操作
| やりたいこと | 書き方 |
|---|---|
| 最後に1つ足す | `a.append(値)` |
| 一部を取り出す | `a[1:3]`（1番〜2番。3番は含まない） |
| 後ろから取り出す | `a[-2:]`（最後の2つ） |

```python
a = [10, 20, 30]
a.append(40)
print(a)
print(a[1:3])
print(a[-2:])
```

▶ **演習**：「第４講」scatter_描画基礎003〜010、「第５講（リスト）」

### 発展編：色のリストを for で作る → 内包表記で作る
- 体重が 60 以上なら赤、それ以外は青、というリストを自動で作る（for・if は第８・９講で詳しく扱う）

```python
colors = []
for w in weight_a:
    if w >= 60:
        colors.append('r')
    else:
        colors.append('b')
print(colors)

colors2 = ['r' if w >= 60 else 'b' for w in weight_a]
print(colors2)
```

### 発展編：カラーマップ
- `c=` に**数値**のリストを渡し、`cmap=` で色の並びを選ぶ。`plt.colorbar()` で目盛り

```python
plt.scatter(height_a, weight_a, c=age, cmap='viridis', s=200)
plt.colorbar(label='年齢')
plt.show()
```

# 第５講
## 平均・標準偏差と補助線
- 平均は `np.mean(リスト)`、標準偏差は `np.std(リスト)`（numpy は計算の道具箱）
- `plt.axhline(y=値)` で横線、`plt.axvline(x=値)` で縦線

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

score = [72, 85, 64, 90, 58, 77, 81, 69]
number = [1, 2, 3, 4, 5, 6, 7, 8]

mean = np.mean(score)
std = np.std(score)
print('平均', mean, '標準偏差', std)

plt.scatter(number, score)
plt.axhline(y=mean, color='r', linestyle='--', label='平均')
plt.ylim(0, 100)
plt.xlabel('出席番号')
plt.ylabel('点数')
plt.legend()
plt.show()
```

## グラフの保存
- `plt.savefig('ファイル名')` は **`plt.show()` より前**に書く（後ろに書くと白紙が保存される）
- `.png`：写真のような形式。PowerPoint・Word に貼るならこれ。`dpi=300` できれいになる
- `.svg`：拡大しても荒れない形式。Illustrator で編集できる
- `transparent=True` で背景を透明にできる（色付きスライドに貼るとき）
- 保存先は、この notebook と同じフォルダ

```python
plt.scatter(number, score)
plt.axhline(y=mean, color='r', linestyle='--')
plt.xlabel('出席番号')
plt.ylabel('点数')
plt.savefig('第５講_点数.png', dpi=300, transparent=True)
plt.show()
```

- `np.array` にすると、リスト全体にまとめて計算できる（リストではできない）

```python
score_np = np.array(score)
print(score_np + 5)
print(score_np / 100)
```

# 第10講
## エラーバー
- 平均だけでは「ばらつき」が分からない。**平均 ± 標準偏差**を線で添えたものがエラーバー
- `plt.errorbar(x, 平均, yerr=標準偏差, fmt='o', capsize=5)`
- 科目ごとの平均と標準偏差を、**for でまとめて**計算する

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

df_exam = pd.read_excel('dsa2025data.xlsx', sheet_name='exam')
subjects = ['国語', '数学', '理科', '社会', '英語']

means = []
stds = []
for s in subjects:
    means.append(np.mean(df_exam[s]))
    stds.append(np.std(df_exam[s]))

plt.errorbar(subjects, means, yerr=stds, fmt='o', capsize=5, markersize=8)
plt.ylim(0, 100)
plt.ylabel('点数（平均 ± 標準偏差）')
plt.show()
```

- 棒グラフなら `plt.bar(..., yerr=標準偏差, capsize=5)`

```python
plt.bar(subjects, means, yerr=stds, capsize=5, color='lightgreen')
plt.ylim(0, 100)
plt.show()
```

▶ **演習**：「エラーバー描画」

> **第２回課題（11/10 出題・11/16 締切）**：円グラフ・両 y 軸・エラーバー・for/if（ひな形の条件や色を直す形）

### 発展編：標準誤差と信頼区間
- 標準偏差：データのばらつき。標準誤差＝標準偏差 ÷ √人数：**平均値**のぶれ
- 平均 ± 1.96 × 標準誤差 が、おおよそ 95% 信頼区間

```python
n = len(df_exam)
ses = [sd / np.sqrt(n) for sd in stds]
plt.errorbar(subjects, means, yerr=[1.96 * se for se in ses], fmt='o', capsize=5)
plt.ylabel('平均点（95%信頼区間）')
plt.show()
```

# 第11講
## 総復習：グラフの型の一覧
困ったら、この表の形を写してデータを入れ替える。

| グラフ | 使う場面 | 書き方 |
|---|---|---|
| 折れ線 | 時間の流れ・順番に意味がある | `plt.plot(x, y)` |
| 散布図 | 2つの量の関係 | `plt.scatter(x, y)` |

# 第14講
## 回帰（単回帰）
- 散布図の点の並びに、いちばん合う直線を引く。**y = 切片 + 傾き × x**
- 例：食事代（total_bill）が増えるとチップ（tip）はどれくらい増える？
- 回帰の道具 `statsmodels` を使う。`ols('y の列名 ~ x の列名', data=df).fit()`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

from statsmodels.formula.api import ols

df_tip = pd.read_excel('dsa2025data.xlsx', sheet_name='tip_data')
df_tip.head()
```

```python
res = ols('tip ~ total_bill', data=df_tip).fit()
print(res.params)
print('決定係数 R2', res.rsquared)
```

- `res.params` の `Intercept` が切片、`total_bill` が傾き。「食事代が1ドル増えるとチップが約0.1ドル増える」
- 決定係数（0〜1）：直線がどれくらい点に合っているか。1 に近いほどよく合う
- 回帰直線は `res.predict()` で描ける

```python
plt.scatter(df_tip['total_bill'], df_tip['tip'], alpha=0.5, zorder=2)
plt.plot(df_tip['total_bill'], res.predict(), color='r', zorder=3)
plt.xlabel('食事代 (USD)')
plt.ylabel('チップ (USD)')
plt.show()
```

### 発展編：多項式回帰
- 曲線を当てはめる：`I(x ** 2)` を足すと2次式

```python
df_wb = pd.read_excel('dsa2025data.xlsx', sheet_name='worldbank')
df_wb = df_wb.rename(columns={'一人あたりGDP': 'gdp', '出生時平均寿命': 'life'})
res2 = ols('life ~ gdp + I(gdp ** 2)', data=df_wb).fit()

x = np.linspace(df_wb['gdp'].min(), df_wb['gdp'].max(), 100)
plt.scatter(df_wb['gdp'], df_wb['life'])
plt.plot(x, res2.predict(pd.DataFrame({'gdp': x})), color='r')
plt.xlabel('一人あたりGDP')
plt.ylabel('平均寿命')
plt.show()
```

## 自分で選んだデータでグラフ＋回帰（第３回課題のひな形）
1. 興味のあるデータを e-Stat などから探す（2つの量の関係を見られるもの）
1. 第13講の手順で Excel を整え、同じフォルダに置く
1. 下のセルの `ファイル名`・`シート名`・`列名` を書き換える

> **第３回課題（11/27 出題・12/3 締切）**：自分で選んだデータでグラフと単回帰。重回帰は任意

```python
df = pd.read_excel('dsa2025data.xlsx', sheet_name='accident')
x_name = '交通事故件数'
y_name = '負傷者数'

res = ols(f'{y_name} ~ {x_name}', data=df).fit()
print(res.params)
print('決定係数 R2', res.rsquared)

plt.scatter(df[x_name], df[y_name], zorder=2)
plt.plot(df[x_name], res.predict(), color='r')
plt.xlabel(x_name)
plt.ylabel(y_name)
plt.show()
```

### 発展編：重回帰と標準化
- x を2つ以上使う：`'tip ~ total_bill + size'`
- 単位の違う x の影響を比べるには、各列を**標準化**（平均0・標準偏差1）してから回帰する

```python
df_tip = pd.read_excel('dsa2025data.xlsx', sheet_name='tip_data')
cols = ['tip', 'total_bill', 'size']
df_z = (df_tip[cols] - df_tip[cols].mean()) / df_tip[cols].std()

res3 = ols('tip ~ total_bill + size', data=df_z).fit()
print(res3.params)
print('決定係数 R2', res3.rsquared)
```

## よくあるエラーと対処
| エラー（最後の行） | よくある原因 | 対処 |
|---|---|---|
| `SyntaxError` | `'` や `)` の閉じ忘れ、`:` の付け忘れ | 指された行の前後を見る |
| `IndentationError` | for・if の下の字下げがずれている | スペース4つにそろえる |
| `NameError` | 変数名の打ち間違い／上のセルを実行していない | 上のセルから実行し直す |
| `ModuleNotFoundError` | ライブラリが入っていない | `pip install japanize-matplotlib` など |
| `FileNotFoundError` | Excel が notebook と別のフォルダにある／名前の打ち間違い | `DSA2026` フォルダに置く |
| `KeyError` | `df['列名']` の列名が違う（空白・全角半角） | `df.head()` で列名を確かめる |
| `ValueError: x and y must have same first dimension` | x と y の個数が違う | `len()` で数をそろえる |
| 日本語が □ になる | `import japanize_matplotlib` を忘れた／手書き風（xkcd）でフォントの指定を忘れた | import のセルを実行する／第12講のフォント指定の行を写す |
| 保存した画像が白紙 | `savefig` を `show` の後に書いた | `savefig` を `show` の前に |
